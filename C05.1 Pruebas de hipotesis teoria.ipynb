# Pruebas de hipótesis

## Introducción a las pruebas de hipótesis

En los capítulos anteriores estudiamos cómo utilizar la información contenida en una muestra para realizar inferencias acerca de una población.

Primero aprendimos a construir **estimadores puntuales**, que nos permiten aproximar parámetros poblacionales desconocidos mediante estadísticos muestrales.

Posteriormente estudiamos los **intervalos de confianza**, que nos permiten expresar la incertidumbre asociada con nuestras estimaciones.

Sin embargo, en muchas situaciones nuestro interés no consiste únicamente en estimar un parámetro, sino también en **evaluar afirmaciones acerca de su valor**.

Por ejemplo, supongamos que una empresa fabrica botellas de agua cuya capacidad nominal es de 600 mililitros. La empresa afirma que su maquinaria llena las botellas con un promedio de 600 mililitros.

Para investigar esta afirmación, seleccionamos una muestra aleatoria de 100 botellas y medimos la cantidad de agua contenida en cada una. Imaginemos que obtenemos una media muestral de 597 mililitros. Surge entonces una pregunta importante: **¿Podemos concluir que la maquinaria está funcionando incorrectamente porque la media muestral es diferente de 600 mililitros?**

La respuesta no es inmediata. Recordemos que la media muestral es una variable aleatoria. Incluso si la maquinaria realmente llena las botellas con un promedio de 600 mililitros, no debemos esperar que todas las muestras produzcan exactamente esa media. Algunas muestras tendrán medias superiores a 600 mililitros, mientras que otras tendrán medias inferiores.

Por lo tanto, observar una media muestral de 597 mililitros no necesariamente contradice la afirmación de la empresa. Lo que necesitamos determinar es si una diferencia de 3 mililitros resulta razonable como consecuencia de la variabilidad muestral o si, por el contrario, constituye evidencia estadística contra la afirmación original.

Este es precisamente el problema que estudian las **pruebas de hipótesis**.

## ¿Qué es una prueba de hipótesis?

Una **prueba de hipótesis**, también llamada **contraste de hipótesis**, es un procedimiento estadístico que utiliza información muestral para evaluar una afirmación acerca de una población.

La idea consiste en establecer una afirmación inicial, suponer provisionalmente que es verdadera y estudiar qué tan compatibles son los resultados observados con dicha afirmación. Si los datos resultan suficientemente incompatibles con lo que esperaríamos bajo esa suposición, tendremos evidencia estadística para rechazar la afirmación inicial. En caso contrario, no tendremos evidencia suficiente para rechazarla.

Es importante comprender que una prueba de hipótesis no constituye una demostración matemática de que una afirmación es verdadera o falsa. A diferencia de una demostración matemática, las conclusiones de una prueba estadística están sujetas a incertidumbre, porque se construyen a partir de información muestral.

## La importancia de la variabilidad muestral

Volvamos al ejemplo de las botellas. Supongamos que la media poblacional realmente es de 600 mililitros y que la desviación estándar poblacional es de 9 mililitros. Seleccionamos una muestra aleatoria de tamaño $n=100$. Asumimos, además, que las botellas se eligen de forma independiente unas de otras, de modo que sus contenidos no se influyen mutuamente.

Como estudiamos anteriormente, la media muestral satisface:

$$E(\overline{X})=\mu\mbox{ y }SD(\overline{X})=\frac{\sigma}{\sqrt{n}}.$$

Por lo tanto $$E(\overline{X})=600\mbox{ y }SD(\overline{X})=\frac{9}{\sqrt{100}}=0.9.$$

Es decir, si la afirmación de la empresa es correcta, **las medias muestrales fluctuarán alrededor de 600 mililitros, con una desviación estándar de 0.9 mililitros**.

¿Por qué la dispersión de las medias es menor que la de las mediciones individuales? Porque al promediar muchas observaciones, los errores que van en una dirección se compensan con los que van en la dirección opuesta. Cuantas más mediciones promediemos, más se "cancelan" esas fluctuaciones y más se concentra el promedio alrededor del verdadero valor de la media poblacional. Esa es la razón intuitiva detrás del factor $1/\sqrt{n}$.

Además, bajo las condiciones del Teorema del Límite Central (muestra aleatoria e independiente, y $n$ suficientemente grande), podemos aproximar la distribución de la media muestral mediante una distribución gaussiana.

En nuestro ejemplo:

$$
\overline{X}\approx N(600,\,0.9^2).
$$

Ahora podemos interpretar la media observada de 597 mililitros desde otra perspectiva.

La diferencia entre la media observada y la media poblacional propuesta es:

$$
597-600=-3.
$$

Esta diferencia equivale a **3.33 desviaciones estándar** de la **distribución muestral de la media**:

$$
\frac{-3}{0.9}=-3.33\ldots
$$

Conviene subrayar que nos referimos a las 0.9 ml de dispersión que tiene el *promedio* de 100 botellas, no a las 9 ml de dispersión que tiene una botella individual. Es una distinción fácil de confundir, pero crucial: la misma diferencia de 3 ml se ve muy distinta según cuál de las dos variabilidades tomemos como referencia.

Por lo tanto, una media muestral de 597 mililitros se encuentra a más de tres desviaciones estándar por debajo de la media propuesta.

Se trata de un resultado poco frecuente si la maquinaria realmente llena las botellas con un promedio de 600 mililitros. Esto constituye evidencia estadística contra la afirmación original.

Observemos que nuestra conclusión no se fundamenta únicamente en que 597 es diferente de 600.

Lo verdaderamente importante es **la magnitud de esa diferencia en relación con la variabilidad muestral**.

Por ejemplo, si la desviación estándar poblacional fuera de 90 mililitros, entonces:

$$
SD(\overline{X})=\frac{90}{\sqrt{100}}=9.
$$

En este caso, la misma diferencia de 3 mililitros representaría solamente 0.33 desviaciones estándar:

$$
\frac{-3}{9}=-0.33\ldots
$$

Por lo tanto, una media muestral de 597 mililitros ya no resultaría especialmente sorprendente bajo la afirmación original.

Esto nos muestra que no basta con comparar directamente un estadístico muestral con el parámetro poblacional propuesto.

También debemos considerar qué tan probable sería observar una discrepancia de esa magnitud, o mayor, si la afirmación original fuera correcta.

## La lógica de las pruebas de hipótesis

El razonamiento anterior puede resumirse en cuatro etapas:

1. Establecemos una afirmación acerca de un parámetro poblacional.

2. Suponemos provisionalmente que dicha afirmación es correcta.

3. Estudiamos cómo deberían comportarse los resultados muestrales bajo esa suposición.

4. Comparamos los resultados observados con ese comportamiento esperado para determinar si existe evidencia estadística suficiente contra la afirmación.

La característica fundamental de este procedimiento es que comenzamos suponiendo que la afirmación inicial es verdadera.

No porque sepamos que lo es, sino porque necesitamos un punto de referencia para evaluar los datos.

## Hipótesis nula e hipótesis alternativa

En una prueba de hipótesis se plantean dos afirmaciones contrapuestas acerca de un parámetro poblacional.

Estas afirmaciones reciben los nombres de **hipótesis nula** e **hipótesis alternativa**.

### Hipótesis nula

La **hipótesis nula**, denotada por $H_0$, es la afirmación que suponemos provisionalmente verdadera y contra la cual evaluamos la evidencia proporcionada por nuestra muestra.

No significa que sepamos que dicha afirmación es correcta. Simplemente la utilizamos como punto de partida para construir nuestro procedimiento estadístico.

### Hipótesis alternativa

La **hipótesis alternativa**, denotada por $H_1$, representa la afirmación para la cual buscamos evidencia estadística.

Nuestro objetivo será determinar si los datos proporcionan evidencia suficiente para rechazar $H_0$ en favor de $H_1$.

Por ejemplo, en el caso de las botellas, podemos plantear:

$$
\begin{cases}
H_0:\mu=600\\
H_1:\mu\neq600
\end{cases}
$$

Aquí, $\mu$ representa la cantidad promedio de agua que realmente contienen las botellas fabricadas por la empresa.

La hipótesis nula establece que la maquinaria llena las botellas con un promedio de 600 mililitros.

La hipótesis alternativa establece que ese promedio es diferente de 600 mililitros.

Observemos que ambas hipótesis se refieren al parámetro poblacional $\mu$, no a la media muestral $\overline{X}$.

Esta distinción es importante: la media muestral es la información que utilizamos para evaluar las afirmaciones sobre la población.

### ¿Qué decisiones podemos tomar?

Una prueba de hipótesis conduce a una de dos decisiones:

1. **Rechazar $H_0$:** cuando los datos proporcionan evidencia estadística suficiente contra la hipótesis nula.

2. **No rechazar $H_0$:** cuando los datos no proporcionan evidencia estadística suficiente para rechazarla.

La segunda decisión merece especial atención.

**No rechazar $H_0$ no significa que hayamos demostrado que $H_0$ es verdadera.**

Significa únicamente que, con la información disponible y el procedimiento utilizado, no encontramos evidencia suficiente para rechazarla.

### Una analogía: la presunción de inocencia

Consideremos un juicio en el que una persona es acusada de cometer un delito.

De acuerdo con el principio de presunción de inocencia, la persona se considera inocente mientras no se demuestre su culpabilidad conforme al procedimiento judicial.

Podemos establecer la siguiente analogía:

$$
\begin{cases}
H_0:\text{El acusado es inocente}\\
H_1:\text{El acusado es culpable}
\end{cases}
$$

La fiscalía debe presentar evidencia contra la presunción inicial.

Si la evidencia es suficientemente sólida, el tribunal puede declarar culpable al acusado.

Si la evidencia es insuficiente, el acusado será declarado no culpable.

Pero observemos que **no culpable no significa necesariamente inocente**.

La declaración solamente indica que no se demostró la culpabilidad bajo las reglas del proceso.

De manera semejante, cuando no rechazamos una hipótesis nula, no estamos demostrando que es verdadera.

Simplemente reconocemos que la evidencia disponible no es suficiente para rechazarla.

Esta analogía es útil para comprender la lógica de las pruebas de hipótesis, aunque un procedimiento judicial y una prueba estadística no utilizan necesariamente los mismos criterios para evaluar la evidencia.

## Selección de las hipótesis y tipos de pruebas

Supongamos que deseamos evaluar una afirmación acerca de un parámetro poblacional $\theta$.

En las pruebas clásicas que estudiaremos, la hipótesis alternativa puede plantearse de tres maneras:

$$
H_1:\theta<\theta_0
$$

$$
H_1:\theta>\theta_0
$$

$$
H_1:\theta\neq\theta_0.
$$

Estas tres posibilidades conducen a dos grandes tipos de pruebas: **unilaterales y bilaterales**.

### Pruebas unilaterales

Una prueba es unilateral cuando buscamos evidencia de que un parámetro poblacional es mayor o menor que cierto valor.

Por ejemplo, supongamos que una empresa afirma que la duración promedio de sus baterías es de al menos 10 horas.

Si nuestro interés consiste en investigar si las baterías duran menos de lo anunciado, podemos plantear:

$$
\begin{cases}
H_0:\mu\geq10\\
H_1:\mu<10
\end{cases}
$$

Esta es una **prueba unilateral izquierda**, porque buscamos evidencia de que la media poblacional es menor que 10.

Por otra parte, si deseamos investigar si la duración promedio supera las 10 horas, podemos plantear:

$$
\begin{cases}
H_0:\mu\leq10\\
H_1:\mu>10
\end{cases}
$$

Esta es una **prueba unilateral derecha**.

En ambos casos, la dirección de la hipótesis alternativa determina qué resultados muestrales constituyen evidencia contra la hipótesis nula.

### Pruebas bilaterales

Una prueba es bilateral cuando buscamos evidencia de que un parámetro poblacional es diferente de cierto valor, independientemente de si resulta mayor o menor.

Por ejemplo, si deseamos investigar si la cantidad promedio de agua contenida en las botellas es distinta de 600 mililitros, planteamos:

$$
\begin{cases}
H_0:\mu=600\\
H_1:\mu\neq600
\end{cases}
$$

En este caso, tanto una media muestral demasiado pequeña como una demasiado grande pueden proporcionar evidencia contra $H_0$.

### Resumen

| Tipo de prueba | Hipótesis alternativa | Dirección de la evidencia |
|---|---|---|
| Unilateral izquierda | $H_1:\theta<\theta_0$ | Valores suficientemente pequeños |
| Unilateral derecha | $H_1:\theta>\theta_0$ | Valores suficientemente grandes |
| Bilateral | $H_1:\theta\neq\theta_0$ | Valores suficientemente alejados en ambas direcciones |

**Observación importante:** la elección de la hipótesis alternativa debe realizarse antes de analizar los resultados muestrales.

No sería correcto observar primero si nuestra estimación es mayor o menor que el valor propuesto y después seleccionar la prueba que favorezca el rechazo de $H_0$.

## El estadístico de contraste

Ya sabemos cómo plantear las hipótesis. Sin embargo, todavía necesitamos una herramienta que nos permita evaluar qué tan compatibles son los datos con la hipótesis nula.

Esta herramienta recibe el nombre de **estadístico de contraste** o **estadístico de prueba**.

Un estadístico de contraste es una variable aleatoria construida a partir de la muestra, cuya distribución bajo la hipótesis nula conocemos o podemos aproximar.

Su utilidad consiste en transformar la información muestral en una cantidad que podamos comparar con el comportamiento esperado cuando $H_0$ es verdadera.

### Ejemplo: contraste para una media poblacional

Supongamos que tenemos una muestra aleatoria:

$$
X_1,X_2,\ldots,X_n
$$

de una población con media $\mu$ y desviación estándar conocida $\sigma$.

Deseamos contrastar:

$$
\begin{cases}
H_0:\mu=\mu_0\\
H_1:\mu\neq\mu_0
\end{cases}
$$

Sabemos que:

$$
E(\overline{X})=\mu
$$

y

$$
SD(\overline{X})=\frac{\sigma}{\sqrt{n}}.
$$

Bajo la hipótesis nula, suponemos que $\mu=\mu_0$.

Por lo tanto, podemos construir el estadístico:

$$
Z=\frac{\overline{X}-\mu_0}{\sigma/\sqrt{n}}.
$$

Observemos que este estadístico mide cuántos errores estándar separan a la media muestral del valor propuesto por la hipótesis nula.

Si la población es gaussiana, entonces bajo $H_0$:

$$
Z\sim N(0,1).
$$

Si la población no es gaussiana, pero se cumplen las condiciones del Teorema del Límite Central y el tamaño muestral es suficientemente grande, podemos utilizar la aproximación:

$$
Z\approx N(0,1).
$$

Por lo tanto, el estadístico de contraste nos permite evaluar nuestros resultados utilizando una distribución conocida.

### Retomando el ejemplo de las botellas

Recordemos que:

$$
\mu_0=600,\qquad \sigma=9,\qquad n=100.
$$

También observamos:

$$
\overline{x}=597.
$$

El valor observado del estadístico de contraste es:

$$
z_{\mathrm{obs}}=\frac{597-600}{9/\sqrt{100}}.
$$

Por lo tanto:

$$
\boxed{z_{\mathrm{obs}}=-3.333}.
$$

Esto significa que la media observada se encuentra 3.333 errores estándar por debajo del valor propuesto por la hipótesis nula.

Como veremos más adelante, este resultado proporciona evidencia considerable contra $H_0$.

Sin embargo, todavía necesitamos establecer una regla formal que nos indique cuándo debemos rechazar una hipótesis nula.

## Nivel de significancia y errores en las pruebas de hipótesis

Como las pruebas de hipótesis utilizan información muestral, sus conclusiones están sujetas a incertidumbre.

Esto significa que, incluso si aplicamos correctamente un procedimiento estadístico, podemos tomar una decisión equivocada.

Existen dos errores fundamentales.

### Error de tipo I

Un **error de tipo I** ocurre cuando rechazamos la hipótesis nula siendo verdadera.

Por ejemplo, supongamos que la maquinaria realmente llena las botellas con un promedio de 600 mililitros.

Si nuestra prueba concluye que existen diferencias estadísticamente significativas y rechazamos $H_0$, habremos cometido un error de tipo I.

### Error de tipo II

Un **error de tipo II** ocurre cuando no rechazamos la hipótesis nula siendo falsa.

Por ejemplo, supongamos que la maquinaria realmente llena las botellas con un promedio de 595 mililitros.

Si nuestra prueba no detecta evidencia suficiente contra $H_0$, habremos cometido un error de tipo II.

Podemos resumir estas posibilidades en la siguiente tabla:

| Realidad | Rechazamos $H_0$ | No rechazamos $H_0$ |
|---|---|---|
| $H_0$ verdadera | Error de tipo I | Decisión correcta |
| $H_0$ falsa | Decisión correcta | Error de tipo II |

### Una analogía: los errores judiciales

Retomemos la analogía del juicio penal.

Recordemos que planteamos:

$$
\begin{cases}
H_0:\text{El acusado es inocente}\\
H_1:\text{El acusado es culpable}
\end{cases}
$$

La fiscalía tiene la responsabilidad de presentar evidencia suficiente para demostrar la culpabilidad del acusado. Si no lo consigue, el acusado será declarado no culpable.

Sin embargo, como las decisiones judiciales se toman a partir de evidencias que pueden ser incompletas, también existe la posibilidad de cometer errores.

**Error de tipo I: condenar a un inocente.**

Supongamos que el acusado realmente es inocente, pero la fiscalía presenta pruebas que llevan al tribunal a declararlo culpable.

En este caso, hemos rechazado $H_0$ siendo verdadera.

Por lo tanto, hemos cometido un error de tipo I.

**Error de tipo II: no condenar a un culpable.**

Supongamos ahora que el acusado realmente cometió el delito, pero la fiscalía no consigue reunir evidencia suficiente para demostrar su culpabilidad.

En consecuencia, el tribunal lo declara no culpable.

En este caso, no hemos rechazado $H_0$ siendo falsa.

Por lo tanto, hemos cometido un error de tipo II.

Podemos resumir la analogía en la siguiente tabla:

| Realidad | Declarar culpable | Declarar no culpable |
|---|---|---|
| El acusado es inocente | Error de tipo I | Decisión correcta |
| El acusado es culpable | Decisión correcta | Error de tipo II |

Observemos que ambos errores tienen consecuencias diferentes.

Condenar a una persona inocente no es lo mismo que dejar sin condena a una persona culpable.

Esta distinción nos permite comprender por qué, en una prueba de hipótesis, es importante controlar la probabilidad de cometer un error de tipo I.

En el contexto estadístico, el nivel de significancia $\alpha$ establece precisamente el riesgo máximo permitido de rechazar la hipótesis nula cuando esta es verdadera.

La analogía judicial también nos recuerda una idea fundamental: **exigir evidencia suficientemente sólida para rechazar $H_0$ protege contra los errores de tipo I, pero puede aumentar el riesgo de cometer errores de tipo II**.

Por ejemplo, si exigimos pruebas cada vez más contundentes para condenar a una persona, reducimos el riesgo de condenar inocentes, pero podemos aumentar el riesgo de no condenar culpables.

De manera semejante, en una prueba estadística, si mantenemos fijo el tamaño de la muestra y utilizamos un nivel de significancia más pequeño, generalmente reducimos la probabilidad de cometer un error de tipo I, pero aumentamos la probabilidad de cometer un error de tipo II frente a una alternativa determinada.

Esta relación entre ambos errores es una de las razones por las que debemos seleccionar cuidadosamente el nivel de significancia antes de realizar una prueba de hipótesis.

### Nivel de significancia

El **nivel de significancia**, denotado por $\alpha$, es un valor que fijamos antes de realizar la prueba y que controla la probabilidad de cometer un error de tipo I.

Formalmente, buscamos que:

$$
P(\text{Rechazar }H_0\mid H_0\text{ verdadera})\leq\alpha.
$$

Cuando la hipótesis nula es simple y la prueba está calibrada para tener tamaño exactamente $\alpha$, se cumple la igualdad.

Los valores convencionales más utilizados son:

$$
\alpha=0.10,\qquad \alpha=0.05,\qquad \alpha=0.01.
$$

Por ejemplo, si seleccionamos:

$$
\alpha=0.05,
$$

estamos diseñando una prueba cuya probabilidad de rechazar incorrectamente una hipótesis nula verdadera no supere el 5%, bajo los supuestos del procedimiento.

Esto no significa que, una vez rechazada la hipótesis nula, exista una probabilidad del 5% de que nuestra conclusión sea equivocada.

El nivel de significancia describe una propiedad del procedimiento estadístico cuando se utiliza repetidamente bajo la hipótesis nula.

### La probabilidad de cometer un error de tipo II

La probabilidad de cometer un error de tipo II suele denotarse por $\beta$.

Esta probabilidad depende del verdadero valor del parámetro poblacional cuando la hipótesis nula es falsa.

La cantidad:

$$
1-\beta
$$

recibe el nombre de **potencia estadística**.

La potencia representa la probabilidad de rechazar correctamente una hipótesis nula falsa, bajo una alternativa específica.

Una prueba con potencia elevada tiene mayor capacidad para detectar diferencias reales respecto a la hipótesis nula.

En general, aumentar el tamaño de la muestra puede incrementar la potencia de una prueba manteniendo fijo el nivel de significancia.

## El valor $p$

Hasta ahora hemos aprendido que las pruebas de hipótesis se construyen suponiendo provisionalmente que $H_0$ es verdadera.

También sabemos que necesitamos comparar el resultado de nuestra muestra con el comportamiento esperado bajo esa suposición.

Sin embargo, surge una pregunta fundamental:

**¿Cómo podemos cuantificar qué tan sorprendente resulta el estadístico observado si la hipótesis nula fuera verdadera?**

Para responderla utilizamos el **valor $p$**.

### Definición

El valor $p$ es la probabilidad, calculada bajo la hipótesis nula, de obtener un valor del estadístico de contraste tan extremo como el observado o más extremo, en la dirección indicada por la hipótesis alternativa.

La definición contiene tres ideas fundamentales:

1. La probabilidad se calcula suponiendo que $H_0$ es verdadera.

2. Se considera el valor observado del estadístico de contraste.

3. Se incluyen los resultados al menos tan extremos como el observado, de acuerdo con la hipótesis alternativa.

Por lo tanto, el valor $p$ nos permite medir qué tan incompatibles resultan nuestros datos con la hipótesis nula, según el estadístico y la prueba utilizados.

Un valor $p$ pequeño indica que, si $H_0$ fuera verdadera, sería poco frecuente obtener un resultado tan extremo como el observado.

Un valor $p$ grande indica que el resultado observado no es especialmente sorprendente bajo $H_0$.

Sin embargo, un valor $p$ grande no demuestra que $H_0$ sea verdadera.

### Cálculo del valor $p$ en pruebas unilaterales

Supongamos que utilizamos un estadístico de contraste $Z$ que sigue una distribución gaussiana estándar bajo $H_0$.

Si nuestra prueba es unilateral derecha:

$$
H_1:\mu>\mu_0,
$$

entonces los valores grandes de $Z$ constituyen evidencia contra $H_0$.

Si obtenemos el valor observado $z_{\mathrm{obs}}$, calculamos:

$$
\boxed{p=P(Z\geq z_{\mathrm{obs}})}.
$$

Por otra parte, si nuestra prueba es unilateral izquierda:

$$
H_1:\mu<\mu_0,
$$

los valores pequeños del estadístico constituyen evidencia contra $H_0$.

En este caso:

$$
\boxed{p=P(Z\leq z_{\mathrm{obs}})}.
$$

Observemos que la dirección de la desigualdad cambia dependiendo de la hipótesis alternativa.

### Cálculo del valor $p$ en pruebas bilaterales

Consideremos ahora:

$$
\begin{cases}
H_0:\mu=\mu_0\\
H_1:\mu\neq\mu_0
\end{cases}
$$

En este caso, tanto los valores demasiado grandes como los demasiado pequeños del estadístico proporcionan evidencia contra $H_0$.

Si utilizamos un estadístico $Z$ con distribución gaussiana estándar bajo $H_0$, el valor $p$ se calcula mediante:

$$
\boxed{p=P(|Z|\geq|z_{\mathrm{obs}}|)}.
$$

Por la simetría de la distribución gaussiana:

$$
\boxed{p=2P(Z\geq|z_{\mathrm{obs}}|)}.
$$

Es importante observar que, en una prueba bilateral, consideramos ambas colas de la distribución.

### Regla de decisión

Una vez calculado el valor $p$, lo comparamos con el nivel de significancia $\alpha$ elegido previamente.

La regla de decisión es:

$$
\boxed{
\begin{cases}
p\leq\alpha &\Rightarrow \text{Rechazar }H_0\\
p>\alpha &\Rightarrow \text{No rechazar }H_0
\end{cases}
}
$$

Por ejemplo, si trabajamos con:

$$
\alpha=0.05
$$

y obtenemos:

$$
p=0.003,
$$

entonces:

$$
0.003<0.05.
$$

Por lo tanto, rechazamos $H_0$.

En cambio, si obtenemos:

$$
p=0.28,
$$

entonces:

$$
0.28>0.05.
$$

Por lo tanto, no rechazamos $H_0$.

### Lo que el valor $p$ no significa

Existen varias interpretaciones incorrectas del valor $p$ que debemos evitar.

El valor $p$ **no es**:

- La probabilidad de que $H_0$ sea verdadera.
- La probabilidad de que $H_1$ sea falsa.
- La probabilidad de que los resultados se deban exclusivamente al azar.
- La probabilidad de que nuestra decisión sea equivocada.
- Una medida directa del tamaño o de la importancia práctica de un efecto.

Por ejemplo, obtener un valor $p=0.01$ no significa que exista una probabilidad del 1% de que la hipótesis nula sea verdadera.

Significa que, bajo la hipótesis nula y los supuestos del modelo, la probabilidad de obtener un resultado al menos tan extremo como el observado es del 1%.

Esta distinción es fundamental para interpretar correctamente las pruebas de hipótesis.

## Ejemplo completo: contraste bilateral para una media poblacional

Retomemos el ejemplo de la empresa que fabrica botellas de agua.

La empresa afirma que la cantidad promedio de agua contenida en sus botellas es de 600 mililitros.

Sabemos que la desviación estándar poblacional es de 10 mililitros.

Seleccionamos una muestra aleatoria de 100 botellas y obtenemos:

$$
\overline{x}=597.
$$

Deseamos determinar si existe evidencia estadística de que la media poblacional es diferente de 600 mililitros.

Trabajaremos con un nivel de significancia:

$$
\alpha=0.05.
$$

**Paso 1. Planteamiento de las hipótesis**

Como queremos investigar si la media poblacional es diferente de 600 mililitros, planteamos:

$$
\begin{cases}
H_0:\mu=600\\
H_1:\mu\neq600
\end{cases}
$$

Se trata de una prueba bilateral.

**Paso 2. Selección del estadístico de contraste**

Como conocemos la desviación estándar poblacional y suponemos que se cumplen las condiciones para utilizar la distribución gaussiana de la media muestral, empleamos:

$$
Z=\frac{\overline{X}-\mu_0}{\sigma/\sqrt{n}}.
$$

Bajo $H_0$:

$$
Z\approx N(0,1).
$$

**Paso 3. Cálculo del estadístico observado**

Sustituimos los valores:

$$
z_{\mathrm{obs}}=\frac{597-600}{10/\sqrt{100}}.
$$

Por lo tanto:

$$
\boxed{z_{\mathrm{obs}}=-3}.
$$

**Paso 4. Cálculo del valor $p$**

Como nuestra prueba es bilateral:

$$
p=P(|Z|\geq3).
$$

Utilizando la simetría de la distribución gaussiana estándar:

$$
p=2P(Z\geq3).
$$

Sabemos que:

$$
P(Z\geq3)\approx0.00135.
$$

Por lo tanto:

$$
p\approx2(0.00135).
$$

De donde:

$$
\boxed{p\approx0.0027}.
$$

Esto significa que, si la media poblacional realmente fuera de 600 mililitros, la probabilidad de observar una media muestral que difiera de 600 en al menos 3 mililitros sería aproximadamente del 0.27%, bajo los supuestos del modelo.

**Paso 5. Regla de decisión**

Comparamos el valor $p$ con el nivel de significancia:

$$
p=0.0027<0.05=\alpha.
$$

Por lo tanto:

$$
\boxed{\text{Rechazamos }H_0}.
$$

**Paso 6. Interpretación**

Con un nivel de significancia del 5%, encontramos evidencia estadística suficiente para concluir que la cantidad promedio de agua contenida en las botellas es diferente de 600 mililitros.

Observemos que nuestra conclusión no afirma que la media poblacional sea exactamente 597 mililitros.

El valor 597 corresponde a nuestra estimación puntual.

La prueba de hipótesis nos permite concluir que los datos proporcionan evidencia contra el valor poblacional propuesto de 600 mililitros.

## Ejemplo: contraste unilateral para una proporción poblacional

Supongamos que, para cierto estudio, se establece como criterio considerar que un estado presenta una situación de seguridad aceptable cuando menos del 10% de sus habitantes han sido víctimas de algún delito durante un periodo determinado.

Este umbral es un criterio hipotético establecido para el ejemplo, no una definición universal de seguridad.

Deseamos investigar si la proporción real de víctimas supera el 10%.

Como no podemos entrevistar a todos los habitantes, seleccionamos una muestra aleatoria de 400 personas.

De ellas, 56 manifiestan haber sido víctimas de algún delito durante el periodo considerado.

Por lo tanto, la proporción muestral es:

$$
\widehat{p}=\frac{56}{400}=0.14.
$$

Es decir, encontramos que el 14% de las personas entrevistadas han sido víctimas de algún delito.

Sin embargo, debemos preguntarnos:

**¿Es suficiente observar un 14% en nuestra muestra para concluir que la proporción poblacional supera el 10%?**

Para responder utilizaremos una prueba de hipótesis.

**Paso 1. Planteamiento de las hipótesis**

Sea $p$ la verdadera proporción poblacional de víctimas de delito.

Planteamos:

$$
\begin{cases}
H_0:p\leq0.10\\
H_1:p>0.10
\end{cases}
$$

Se trata de una prueba unilateral derecha.

**Paso 2. Selección del estadístico de contraste**

Bajo las condiciones que permiten utilizar la aproximación gaussiana para una proporción muestral, empleamos:

$$
Z=\frac{\widehat{p}-p_0}{\sqrt{\frac{p_0(1-p_0)}{n}}}.
$$

Aquí:

$$
p_0=0.10.
$$

La distribución del estadístico bajo el valor frontera de la hipótesis nula se aproxima mediante:

$$
Z\approx N(0,1).
$$

**Paso 3. Cálculo del estadístico observado**

Sustituimos:

$$
z_{\mathrm{obs}}=\frac{0.14-0.10}{\sqrt{\frac{0.10(0.90)}{400}}}.
$$

El denominador es:

$$
\sqrt{\frac{0.09}{400}}=0.015.
$$

Por lo tanto:

$$
z_{\mathrm{obs}}=\frac{0.04}{0.015}.
$$

De donde:

$$
\boxed{z_{\mathrm{obs}}\approx2.667}.
$$

**Paso 4. Cálculo del valor $p$**

Como se trata de una prueba unilateral derecha:

$$
p\text{-valor}=P(Z\geq2.667).
$$

Utilizando la distribución gaussiana estándar:

$$
\boxed{p\text{-valor}\approx0.0038}.
$$

**Paso 5. Regla de decisión**

Si utilizamos:

$$
\alpha=0.05,
$$

entonces:

$$
0.0038<0.05.
$$

Por lo tanto, rechazamos $H_0$.

**Paso 6. Interpretación**

Con un nivel de significancia del 5%, encontramos evidencia estadística suficiente para concluir que la proporción poblacional de víctimas de delito supera el 10%, suponiendo que el muestreo y las respuestas permiten una inferencia válida sobre la población.

Observemos que no estamos afirmando que la proporción poblacional sea exactamente del 14%.

Ese porcentaje corresponde únicamente a nuestra estimación muestral.

La conclusión de nuestra prueba es que existe evidencia estadística de que la proporción poblacional supera el umbral establecido.

## Pruebas de hipótesis para la media poblacional

Hasta ahora hemos trabajado con el estadístico $Z$ para contrastar afirmaciones acerca de una media poblacional cuando conocemos la desviación estándar $\sigma$.

Sin embargo, en la práctica es frecuente que la desviación estándar poblacional también sea desconocida.

Esto nos conduce a dos situaciones diferentes.

### Caso 1. Desviación estándar poblacional conocida

Supongamos que deseamos contrastar:

$$
H_0:\mu=\mu_0.
$$

Si conocemos la desviación estándar poblacional $\sigma$ y la población es gaussiana, entonces:

$$
Z=\frac{\overline{X}-\mu_0}{\sigma/\sqrt{n}}
$$

sigue una distribución gaussiana estándar bajo $H_0$.

Para poblaciones no gaussianas podemos recurrir a una aproximación basada en el Teorema del Límite Central, siempre que sus condiciones sean adecuadas.

### Caso 2. Desviación estándar poblacional desconocida

Si no conocemos $\sigma$, podemos estimarla mediante la desviación estándar muestral $S$.

En este caso construimos:

$$
T=\frac{\overline{X}-\mu_0}{S/\sqrt{n}}.
$$

Si la muestra procede de una población gaussiana, entonces, bajo $H_0$:

$$
\boxed{T\sim t_{n-1}}.
$$

Es decir, el estadístico sigue una distribución $t$ de Student con $n-1$ grados de libertad.

Esta distribución ya la estudiamos al construir intervalos de confianza para una media poblacional con desviación estándar desconocida.

A medida que aumenta el número de grados de libertad, la distribución $t$ de Student se aproxima a la distribución gaussiana estándar.

Por lo tanto, la elección entre los estadísticos $Z$ y $T$ depende de la información que tenemos acerca de la población y de los supuestos que podemos justificar.

### Comparación

| Situación | Estadístico | Distribución bajo $H_0$ |
|---|---|---|
| Media con $\sigma$ conocida, población gaussiana | $Z=\frac{\overline{X}-\mu_0}{\sigma/\sqrt{n}}$ | $N(0,1)$ |
| Media con $\sigma$ desconocida, población gaussiana | $T=\frac{\overline{X}-\mu_0}{S/\sqrt{n}}$ | $t_{n-1}$ |
| Proporción poblacional, con aproximación gaussiana válida | $Z=\frac{\widehat{p}-p_0}{\sqrt{p_0(1-p_0)/n}}$ | Aproximadamente $N(0,1)$ |

## Relación entre pruebas de hipótesis e intervalos de confianza

Las pruebas de hipótesis y los intervalos de confianza no son procedimientos independientes.

Ambos utilizan información muestral para realizar inferencias acerca de parámetros poblacionales y ambos se apoyan en las distribuciones muestrales de los estadísticos.

De hecho, existe una relación particularmente importante entre las pruebas bilaterales y los intervalos de confianza.

Supongamos que deseamos contrastar:

$$
\begin{cases}
H_0:\mu=\mu_0\\
H_1:\mu\neq\mu_0
\end{cases}
$$

con un nivel de significancia $\alpha$.

También podemos construir un intervalo de confianza bilateral para $\mu$ con nivel de confianza $1-\alpha$, utilizando el mismo modelo y estadístico.

En este caso, la decisión de la prueba bilateral correspondiente puede obtenerse directamente del intervalo de confianza.

- Si $\mu_0$ pertenece al intervalo de confianza, no rechazamos $H_0$.
- Si $\mu_0$ no pertenece al intervalo de confianza, rechazamos $H_0$.

### Ejemplo

Volvamos al caso de las botellas.

Recordemos que:

$$
\overline{x}=597,\qquad \sigma=10,\qquad n=100.
$$

Construimos un intervalo de confianza del 95% para la media poblacional.

Como conocemos $\sigma$, utilizamos:

$$
IC_{95\%}=\overline{x}\pm z_{0.975}\frac{\sigma}{\sqrt{n}}.
$$

Sustituyendo:

$$
IC_{95\%}=597\pm1.96\frac{10}{\sqrt{100}}.
$$

Por lo tanto:

$$
IC_{95\%}=597\pm1.96.
$$

De donde:

$$
\boxed{IC_{95\%}=[595.04,598.96]}.
$$

Observemos que el valor propuesto por la hipótesis nula:

$$
\mu_0=600
$$

no pertenece al intervalo de confianza.

Por lo tanto, rechazamos $H_0$ al nivel de significancia del 5%.

Esta es precisamente la misma decisión que obtuvimos mediante el valor $p$.

Sin embargo, mientras que la prueba de hipótesis nos permite evaluar el valor propuesto de 600 mililitros, el intervalo de confianza también proporciona información sobre los valores plausibles de la media poblacional, bajo el procedimiento y sus supuestos.

## Consideraciones finales

Las pruebas de hipótesis constituyen una herramienta fundamental de la inferencia estadística porque permiten evaluar afirmaciones sobre una población a partir de información muestral.

Su razonamiento se basa en suponer provisionalmente que una hipótesis nula es verdadera y estudiar qué tan compatibles son los resultados observados con esa suposición.

Para llevar a cabo este procedimiento necesitamos establecer una hipótesis nula, una hipótesis alternativa, un estadístico de contraste y una regla de decisión.

El valor $p$ nos permite cuantificar qué tan extremo resulta el estadístico observado bajo la hipótesis nula, mientras que el nivel de significancia $\alpha$ establece el criterio para decidir cuándo rechazarla.

Es importante recordar que las conclusiones de una prueba de hipótesis no son demostraciones matemáticas.

Rechazar una hipótesis nula significa que encontramos evidencia estadística suficiente contra ella, mientras que no rechazarla significa únicamente que la evidencia disponible no fue suficiente para hacerlo.

Finalmente, una diferencia estadísticamente significativa no necesariamente representa una diferencia importante desde el punto de vista práctico.

Con muestras suficientemente grandes, incluso diferencias muy pequeñas pueden resultar estadísticamente significativas.

Por esta razón, al interpretar los resultados de una prueba de hipótesis también debemos considerar la magnitud de las diferencias observadas, su contexto y los supuestos del procedimiento utilizado.

En síntesis, **las pruebas de hipótesis no eliminan la incertidumbre, sino que nos proporcionan un procedimiento sistemático para tomar decisiones estadísticas controlando determinados riesgos de error**.
