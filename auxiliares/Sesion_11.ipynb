Recordemos que, dados los datos $x_1$, $x_2$,...,$x_n$, su desviación estándar muestral viene dada por $$\sqrt{\frac{(x_1-\overline{x})^2+(x_2-\overline{x})^2+...+(x_n-\overline{x})^2}{n-1}}$$



Se define la **varianza muestral** como el cuadrado de la desviación. Es decir $$\frac{(x_1-\overline{x})^2+(x_2-\overline{x})^2+...+(x_n-\overline{x})^2}{n-1}$$

Análogamente tenemos la desviación poblacional y la varianza poblacional dadas por $$\sqrt{\frac{(x_1-\overline{x})^2+(x_2-\overline{x})^2+...+(x_N-\overline{x})^2}{N}}\mbox{ y }\frac{(x_1-\overline{x})^2+(x_2-\overline{x})^2+...+(x_N-\overline{x})^2}{N}$$



Y a la varianza de una población se le denota por $\sigma^2$ y entonces su desviación es $\sigma$

Sea $\sigma^2_m$ la varianza muestral.

$$\sigma^2=\frac{n-1}{N}\sigma^2_m$$

**Estadísticos** Digamos que $\mu$ es la estatura media de la población de estudio.

Tomaré una muestra aleatoria $X_1,X_2,...,X_n$ de la población y,  **solo porque quiero**, voy a definir $\mathbb{X}=X_1+X_2+...+X_n+\mu$

Esta $\mathbb{X}$ **no es un estadístico** porque para calcularse necesito saber el parámetro poblacional $\mu$

Lo que sí es un estadístico es, por ejemplo, $\mathcal{X}=\cos(X_1)+\sin(X_2)-\log(X_n)+\exp(\overline{X})$

Ejemplos de estadísticos:

- $\overline{X}=\frac{X_1+X_2+...+X_n}{n}$

- $S^2=\frac{(X_1-\overline{X})^2+(X_2-\overline{X})^2+...+(X_n-\overline{X})^2}{n-1}$

- $\min\{X_1,X_2,...,X_n\}$ y $\max\{X_1,X_2,...,X_n\}$
