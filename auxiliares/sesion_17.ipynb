# Población gaussiana

Si la característica es gaussiana, **no importa el tamaño de la muestra**.
- $\sigma$ conocida: usamos ``z.test``
- $\sigma$ desconocida: usamos ``t.test``

Ambos métodos son **exactos**

Y si la población no es gaussiana pero $n$ es grande, los criteriores anteriores siguen aplicando. Pero los métodos son **aproximativos**
