## El p-valor

Es una medida de qué tan sorprendente es que observemos lo que estamos observando dando por hecho que $H_0$ es verdadera:

**Ejemplo**
$$P(\mbox{observar lo que observas }|\mbox{ la moneda es justa})$$

Digamos que obtuvimos 80 caras y 20 cruces. Entonces el p-valor me indica qué tan probable es que observe 80 caras y 20 cruces dando por hecho que la moneda es justa

$$P(X=80)=\binom{100}{80}0.5^{80}0.5^{20}=4.22816×10^{-10}=0.000000000422$$

Si el p-valor es pequeñito, significa que la probabilidad de observar lo que me dice la realidad (la muestra) **dando por hecho $H_0$** es muy baja

Es decir, tendré evidencia suficiente para decir que $H_0$ es falsa

**Ejemplo**

En el caso de las botellas, ellos dicen que cada botella tiene mas o menos 600ml. Pero tú tomas 100 botellas y te dio que en promedio tienen 597ml.

El p-valor es $$P(\mbox{mi muestra tiene media 597 }|\mbox{ la media poblacional es 600})$$
