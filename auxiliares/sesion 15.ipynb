**Observación** no confundamos la varianza de la media muestral con la varianza muestral.

**Varianza muestral**. Recordemos que nuestros tres problemas típicos en estadística son 
- Estimar la media poblacional
- Estimar la proporción de individuos que cumplen una condición
- Estimar la varianza de una población respecto de su media

Para el tercer problema, el estimador que se utiliza es la varianza muestral:  $$S^2=\frac{(X_1-\overline{X})^2+(X_2-\overline{X})^2+...+(X_n-\overline{X})^2}{n-1}$$

**Varianza de la media muestral**. Este me sirve para el primer problema. ¿Cómo me sirve? ah pues cuando tome una muestra, si $n$ es grande, sabré que su media muestral se parece mucho a $\mu$.

Se calcula como $$Var(\overline{X})=\frac{\sigma^2}{n}$$

**Varianza poblacional** Esta es $\sigma^2$ y se calcula así: $\boldsymbol{x}_1,\boldsymbol{x_2},...,\boldsymbol{x}_N$ son toda la población y  $$\sigma^2=\frac{(\boldsymbol{x}_1-\mu)^2+(\boldsymbol{x}_2-\mu)^2+...+(\boldsymbol{x}_N-\mu)^2}{N}$$
