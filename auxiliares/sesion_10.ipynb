Tomemos $$f(x)=\frac{|x_1-x|+|x_2-x|+...+|x_n-x|}{n}$$

La mediana es el mejor representante cuando se utiliza como medida de dispersión a la $f(x)$ anterior.

PEEEEERO.... si esa dispersión es grande significa que, aunque la mediana es el mejor representante, **NO ES UN BUEN REPRESENTANTE** (y por lo tanto, cualquier otro valor no es buen representante según este criterio)

Es decir, tomemos ahora $$g(x)=\sqrt{\frac{(x_1-x)^2+(x_2-x)^2+...+(x_n-x)^2}{n-1}}$$

Digamos que esta es nuestra medida de dispersión. La media aritmética es el mejor representante cuando utilizamos esta medida de dispersión. 

PEEEEERO.... si esa dispersión es grande significa que, aunque la media aritmética es el mejor representante, **NO ES UN BUEN REPRESENTANTE** (y por lo tanto, cualquier otro valor no es buen representante según este criterio)

**Ojo:** a la media aritmética se le denota por $\overline{x}$. Es decir, si mis datos son $x_1,x_2,..,x_n$ se denota a su media aritmética por $\overline{x}=\frac{x_1+x_2+...+x_n}{n}$

**Importante** Supongamos que nuestra población tiene $N$ elementos y nuestra muetra tiene $n$ elementos.

Al valor $$\sqrt{\frac{(x_1-\overline{x})^2+(x_2-\overline{x})^2+...+(x_n-\overline{x})^2}{n-1}}$$

se le llama **desviación estandándar MUESTRAL**

Al valor $$\sqrt{\frac{(x_1-\overline{x})^2+(x_2-\overline{x})^2+...+(x_N-\overline{x})^2}{N}}$$ se le llama se le llama **desviación estandándar POBLACIONAL**
