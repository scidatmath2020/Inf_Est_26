Sabemos que si $\mu$ es la media poblacional, entonces $$\sigma^2=\frac{(x_1-\mu)^2+(x_2-\mu)^2+...+(x_N-\mu)^2}{N}$$ es la varianza de la población respecto de esa media.

Digamos que 1000 veces me hago la misma pregunta: ¿Cuál es la edad promedio $\mu$ de la población?

El decir que $\hat{\mu}_1=X_1$ y $\hat{\mu}_2=\overline{X}$ significa que con el primer método, diré que la edad promedio estimada es el primer valor de la muestra, en tanto que con el segundo método diré que la edad promedio estimada es el promedio de la muestra.

Por cada vez que me lo pregunte, tomo una muestra de tamaño $n$. Entonces voy a tomar 1000 muestras. 

El hecho de que $Var(X_1)=\sigma^2$ y $Var(\overline{X})=\frac{\sigma^2}{n}$ implica que la gran mayoría de las veces $\overline{X}$ es más cercano a $\mu$ que $X_1$


**ECM** Que el error cuadrático medio sea pequeño implica que la varianza del estimador y el sesgo son pequeños.

- Si el sesgo del estimador es pequeño, significa que la estimación se parece mucho al valor verdadero.
- Si la varianza del estimador es pequeña, significa que si cambio de muestra, la estimación no va a cambiar mucho

El símbolo $$\prod_{i=1}^n p^{x_i}(1-p)^{1-x_i}$$

representa $$p^{x_1}(1-p)^{1-x_1}\times p^{x_2}(1-p)^{1-x_2}\times ...\times p^{x_n}(1-p)^{1-x_n}$$

La expresión anterior depende de $p$ (es una función de $p$):

$$L(p)=p^{x_1}(1-p)^{1-x_1}\times p^{x_2}(1-p)^{1-x_2}\times ...\times p^{x_n}(1-p)^{1-x_n}$$



---

A veces es mejor trabajar con la función $\log(L)$ en lugar en lugar de $L$. Por ejemplo, en el caso de la Bernoull tenemos 
\begin{eqnarray*}
\ell(p)&=&\log(L(p))\\
&=&\log\left(\prod_{i=1}^n p^{x_i}(1-p)^{1-x_i}\right)\\
&=&\sum_{i=1}^n\log(p^{x_i}(1-p)^{1-x_i})\\
&=&\sum_{i=1}^nx_i\log(p)+(1-x_i)\log(1-p)\\
\end{eqnarray*}



por lo tanto $$\ell^\prime(p)=\sum_{i=1}^n\frac{x_i}{p}-\frac{1-x_i}{1-p}=\frac{1}{p}\sum x_i-\frac{1}{1-p}\left(n-\sum x_i\right)$$
