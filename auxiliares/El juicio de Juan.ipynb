# El juicio de Juan

## Introducción

Para comprender el funcionamiento de las pruebas de hipótesis, imaginemos un sistema judicial ficticio en el que las decisiones sobre la culpabilidad de una persona se toman mediante procedimientos estadísticos.

En este sistema, la ley establece qué tan contundente debe ser la evidencia para declarar culpable a un acusado. El fiscal se encarga de investigar y presentar las pruebas, mientras que el juez debe aplicar la ley para emitir un veredicto.

Nuestro objetivo será comprender cómo funcionan las hipótesis estadísticas, el nivel de significancia $\alpha$, el valor-p y los errores de tipo I y II.

## Procedimiento general

Antes de comenzar, describamos las etapas que seguiremos durante el juicio.

1. **Se presenta a un acusado.** Una persona es llevada ante el juez. Se presume su inocencia y corresponde al fiscal presentar evidencia de su culpabilidad. En términos estadísticos, se establecen las hipótesis $H_0$ y $H_1$.

2. **La ley establece el nivel de significancia.** La legislación determina qué tan exigente debe ser el criterio para declarar culpable a una persona. En estadística, esta exigencia se representa mediante $\alpha$. Cuanto menor sea su valor, más contundente deberá ser la evidencia.

3. **El fiscal realiza su investigación.** Utiliza una técnica previamente definida para recopilar y analizar la evidencia. Esta técnica es general: puede aplicarse a distintos acusados y no depende del nivel de significancia establecido por la ley.

4. **El fiscal presenta la evidencia.** A partir de los resultados de su investigación, calcula un estadístico de prueba y obtiene el valor-p. Este valor permite evaluar qué tan probable sería encontrar evidencia igual o más desfavorable para el acusado que la observada, suponiendo que es inocente.

5. **El juez emite su veredicto.** Compara el valor-p con el nivel de significancia establecido por la ley. Si $p<\alpha$, rechaza $H_0$. En caso contrario, no la rechaza.

6. **Se analizan los posibles errores.** El juez podría condenar a un inocente (error de tipo I) o no condenar a un culpable (error de tipo II). El nivel de significancia $\alpha$ controla la probabilidad del primer error.

Este procedimiento permite distinguir tres elementos fundamentales: **la ley establece el criterio de decisión, el fiscal presenta la evidencia y el juez emite el veredicto**.

A continuación, desarrollaremos cada una de estas etapas mediante el juicio de Juan.

---

## 1. El crimen y la presentación del acusado

Se ha cometido un asesinato en el que se utilizaron dos armas: una pistola y un cuchillo.

La policía investiga a un grupo de amigos que podrían estar relacionados con el crimen. Entre ellos se encuentran Juan, Luis y Pedro.

Sin embargo, solamente consigue detener a Juan, quien es llevado a juicio como principal sospechoso.

El fiscal tiene la responsabilidad de presentar evidencia sobre su posible culpabilidad, mientras que el juez deberá examinar las pruebas y emitir un veredicto.

Para estudiar este caso desde el punto de vista estadístico, establecemos dos hipótesis:

$$
\boxed{H_0:\text{Juan es inocente}.}
$$

$$
\boxed{H_1:\text{Juan es culpable}.}
$$

La hipótesis nula representa la presunción de inocencia. Por lo tanto, Juan no tiene la obligación de demostrar que es inocente. Es el fiscal quien debe presentar evidencia suficiente para que el juez pueda rechazar $H_0$.

Aquí aparece una distinción fundamental: **el objetivo del fiscal en este juicio consiste exclusivamente en presentar evidencia sobre la posible culpabilidad de Juan**.

Supongamos que, después de una investigación exhaustiva, el fiscal no consigue reunir evidencia suficiente para demostrar que Juan participó en el asesinato.

¿Significa esto que ahora está obligado a investigar a Luis o Pedro y presentar a alguno de ellos como posible culpable?

**¡Por supuesto que no!**

El fiscal no tiene la obligación de proponer un culpable alternativo para que el juez pueda resolver el caso de Juan. Si no consigue reunir pruebas suficientes, el juez simplemente no podrá rechazar la hipótesis de inocencia.

Esto no demuestra que Juan sea inocente, pero tampoco proporciona evidencia alguna de que Luis o Pedro sean culpables.

**En una prueba de hipótesis sucede exactamente lo mismo: nuestro objetivo es determinar si existe evidencia suficiente para rechazar $H_0$, no encontrar otra hipótesis que debamos aceptar cuando no conseguimos rechazarla.**

Por lo tanto,

$$
\boxed{\text{No rechazar }H_0\neq\text{Demostrar que }H_0\text{ es verdadera}.}
$$

## 2. La ley y el nivel de significancia

En nuestro país ficticio existe una ley muy particular: para declarar culpable a una persona, la evidencia presentada debe satisfacer un criterio estadístico previamente establecido.

Este criterio se conoce como **nivel de significancia** y se representa mediante $\alpha$.

La ley busca proteger a las personas inocentes. Por ello, exige que el procedimiento utilizado para determinar la culpabilidad tenga una probabilidad de cometer un error de tipo I que no supere $\alpha$.

Es decir,

$$
\boxed{P(\text{Declarar culpable}\mid\text{Inocente})\leq\alpha.}
$$

Pero ¿qué significa esto en la práctica?

Imaginemos que existen tres países con leyes diferentes.

En el primer país, la ley establece

$$
\alpha=0.10.
$$

En el segundo país,

$$
\alpha=0.05.
$$

Y en el tercero,

$$
\alpha=0.01.
$$

En los tres países se utiliza exactamente el mismo procedimiento estadístico para evaluar la evidencia.

Sin embargo, **la ley del tercer país es mucho más exigente que la del primero**.

¿Por qué?

Porque el tercer país solamente permite procedimientos que rechacen equivocadamente la hipótesis de inocencia, como máximo, en el $1\%$ de los casos en que esta sea verdadera.

En cambio, el primero permite una probabilidad de hasta el $10\%$.

Esto tiene una consecuencia importante: **cuanto menor sea $\alpha$, más contundente deberá ser la evidencia para declarar culpable al acusado**.

La ley no determina qué pruebas debe buscar el fiscal ni cómo debe realizar su investigación.

El fiscal utiliza su procedimiento habitual y presenta los resultados obtenidos.

La ley solamente establece el criterio estadístico que deberá satisfacer la evidencia para que el juez pueda rechazar la hipótesis de inocencia.

Por lo tanto, una misma investigación podría conducir a una condena en un país y a una absolución en otro, simplemente porque sus leyes establecen niveles de significancia diferentes.

### El significado del nivel de significancia

Para comprender mejor el significado de $\alpha$, imaginemos que nuestro sistema judicial procesa una gran cantidad de casos semejantes, utilizando siempre el mismo procedimiento estadístico.

Supongamos, además, que todos los acusados son realmente inocentes.

En algunos casos, la evidencia presentada será suficientemente desfavorable para que el juez condene equivocadamente a una persona inocente.

Esta equivocación se conoce como **error de tipo I**.

El nivel de significancia $\alpha$ establece la máxima probabilidad permitida para cometer este error.

Formalmente,

$$
\boxed{P(\text{Rechazar }H_0\mid H_0\text{ verdadera})\leq\alpha.}
$$

Por ejemplo, si la ley establece

$$
\alpha=0.05,
$$

significa que el procedimiento completo, formado por la técnica estadística del fiscal y la regla de decisión del juez, debe rechazar equivocadamente la hipótesis de inocencia en no más del $5\%$ de los casos en los que esta sea verdadera, bajo las condiciones del modelo estadístico.

**El nivel de significancia no representa la probabilidad de que Juan sea inocente.**

Tampoco representa la probabilidad de que el juez se equivoque específicamente en este juicio.

Representa una propiedad del procedimiento general de decisión.

Es importante señalar que el nivel de significancia debe establecerse **antes de examinar la evidencia**.

El juez no puede modificar el nivel de exigencia establecido por la ley simplemente porque las pruebas presentadas por el fiscal resulten más o menos contundentes de lo esperado.

## 3. El procedimiento del fiscal

Una vez establecido el nivel de significancia, comienza la investigación estadística del caso.

Nuestro fiscal es un profesional experimentado. Ha participado en numerosos juicios por asesinato y, con el paso de los años, ha desarrollado una técnica de investigación bien definida.

Cuando recibe un caso como el de Juan, no comienza buscando pruebas al azar. Sigue un procedimiento sistemático.

Primero, mide la cantidad de residuos de pólvora presentes en las manos del acusado.

Después, investiga si recientemente ha comprado cuchillos o armas de fuego.

También reconstruye sus movimientos durante las horas cercanas al asesinato.

En particular, sabemos que Juan se encontraba cerca del lugar del crimen. Por lo tanto, el fiscal investiga qué tan cerca estaba, cuánto tiempo permaneció allí y si tenía oportunidad de cometer el asesinato.

Finalmente, examina la relación entre Juan y la víctima, buscando posibles conflictos o motivos.

Lo importante es que **el fiscal utiliza este mismo procedimiento general en todos los casos semejantes**.

A partir de la información recopilada, su técnica produce una medida estadística que resume qué tan desfavorable resulta la evidencia para el acusado.

Pero nuestro fiscal sabe que su técnica no es perfecta.

En ocasiones, una persona inocente puede presentar residuos de pólvora, haber comprado un cuchillo, encontrarse cerca de la escena del crimen y mantener una relación conflictiva con la víctima.

Por lo tanto, existe la posibilidad de que su procedimiento encuentre evidencia aparentemente incriminatoria contra una persona inocente.

Precisamente por esta razón, la técnica del fiscal debe estar correctamente calibrada: necesitamos conocer cómo se comportan sus resultados cuando el acusado es inocente.

Además, el procedimiento de investigación y la forma de combinar sus resultados deben establecerse previamente. De lo contrario, el fiscal podría seleccionar únicamente las pruebas que resulten desfavorables para Juan, alterando las propiedades estadísticas del procedimiento.

**El fiscal no necesita conocer el nivel de significancia establecido por la ley para realizar su investigación.**

Su técnica es independiente de $\alpha$.

De hecho, podemos imaginar que el fiscal termina su investigación y calcula sus resultados antes de conocer el nivel de significancia que se utilizará para emitir el veredicto.

## 4. La evidencia y el valor-p

El fiscal aplica su procedimiento al caso de Juan.

Primero, examina sus manos y encuentra residuos de pólvora.

Después, revisa sus movimientos financieros y descubre que recientemente compró un cuchillo.

También confirma que Juan se encontraba cerca del lugar del asesinato a la hora en que ocurrió.

Finalmente, descubre que Juan y la víctima habían tenido una discusión días antes del crimen.

La evidencia parece comprometedora.

Sin embargo, ninguna de estas observaciones demuestra, por sí sola, que Juan sea culpable.

Una persona inocente podría tener residuos de pólvora porque practicó tiro deportivo. También podría comprar un cuchillo por motivos completamente ajenos al asesinato.

Estar cerca de la escena del crimen tampoco demuestra participación alguna.

Incluso una discusión con la víctima podría ser una simple coincidencia.

Por lo tanto, el fiscal debe evaluar **qué tan compatible es el conjunto de evidencias con la hipótesis de inocencia**.

### El valor-p

Recordemos que nuestra hipótesis nula es

$$
H_0:\text{Juan es inocente}.
$$

El fiscal se plantea la siguiente pregunta:

**Si Juan fuera inocente, ¿qué tan probable sería que mi procedimiento encontrara evidencia igual o más contundente en su contra que la que he encontrado?**

Esta pregunta nos conduce al concepto de valor-p.

La técnica del fiscal produce un estadístico de prueba $T$, cuyo valor aumenta conforme la evidencia resulta más desfavorable para el acusado.

Supongamos que, después de investigar a Juan, obtiene el valor observado $t_{\text{obs}}$.

El valor-p se define como

$$
\boxed{p=P(T\geq t_{\text{obs}}\mid H_0).}
$$

Es decir, el valor-p representa la probabilidad de obtener, suponiendo que Juan es inocente, un resultado del procedimiento con evidencia igual o más contundente en su contra que la observada.

Si el valor-p es pequeño, significa que sería poco frecuente encontrar evidencia tan desfavorable contra una persona inocente, de acuerdo con el modelo estadístico utilizado.

Por el contrario, si el valor-p es grande, significa que la evidencia encontrada no resulta especialmente sorprendente bajo la hipótesis de inocencia.

Una vez concluida su investigación, **el fiscal presenta la evidencia y el valor-p al juez**.

El fiscal no decide si Juan debe ser condenado.

Su responsabilidad consiste en presentar los resultados de su investigación. La decisión corresponde exclusivamente al juez, quien debe aplicar el criterio establecido por la ley.

### Una distinción fundamental

El valor-p no responde a la pregunta:

*¿Cuál es la probabilidad de que Juan sea inocente, dada la evidencia encontrada?*

Esa probabilidad sería

$$
P(H_0\mid \text{evidencia}).
$$

En cambio, el valor-p responde a una pregunta diferente:

*Suponiendo que Juan es inocente, ¿qué tan probable sería que el procedimiento encontrara evidencia igual o más contundente en su contra que la observada?*

Es decir,

$$
P(T\geq t_{\text{obs}}\mid H_0).
$$

Estas dos probabilidades no son equivalentes.

En particular, **un valor-p pequeño no significa que la probabilidad de que Juan sea inocente sea pequeña**.

Significa que la evidencia observada sería poco frecuente si la hipótesis de inocencia fuera verdadera, de acuerdo con el procedimiento y el modelo estadístico establecidos.

## 5. ¿Por qué el juez compara el valor-p con $\alpha$?

Hasta este momento hemos introducido dos conceptos diferentes.

Por una parte, el nivel de significancia $\alpha$ establece el criterio legal para controlar la probabilidad de cometer un error de tipo I.

Por otra parte, el valor-p es calculado por el fiscal a partir de la evidencia encontrada durante su investigación.

Pero entonces surge una pregunta:

**Si el valor-p no depende de $\alpha$, ¿por qué el juez debe compararlos para tomar una decisión?**

La respuesta está en una propiedad fundamental de los valores-p correctamente calculados: cuando la hipótesis nula es verdadera, la probabilidad de obtener un valor-p menor que $\alpha$ no supera $\alpha$.

Es decir,

$$
\boxed{P(p<\alpha\mid H_0)\leq\alpha.}
$$

Por lo tanto, si la ley establece que debe rechazarse $H_0$ cuando $p<\alpha$, la probabilidad de cometer un error de tipo I estará controlada por el nivel de significancia.

**Esta es la razón por la que comparamos el valor-p con $\alpha$.**

### Tres países, una misma evidencia

Imaginemos ahora que el fiscal presenta exactamente la misma evidencia en tres países diferentes.

En los tres casos, el valor-p obtenido es

$$
p=0.03.
$$

Sin embargo, cada país tiene una legislación diferente.

En el primer país,

$$
\alpha_1=0.10.
$$

En el segundo país,

$$
\alpha_2=0.05.
$$

Y en el tercero,

$$
\alpha_3=0.01.
$$

En el primer país,

$$
p=0.03<0.10=\alpha_1.
$$

Por lo tanto, el juez rechaza $H_0$.

En el segundo país,

$$
p=0.03<0.05=\alpha_2.
$$

Por lo tanto, el juez también rechaza $H_0$.

Sin embargo, en el tercer país,

$$
p=0.03>0.01=\alpha_3.
$$

Por lo tanto, el juez no rechaza $H_0$.

**El fiscal presentó exactamente la misma evidencia. Lo que cambió fue el nivel de exigencia establecido por la ley de cada país.**

Este ejemplo permite comprender que el valor-p depende de los resultados de la investigación, mientras que el nivel de significancia representa un criterio de decisión establecido previamente.

En consecuencia, **cuanto menor sea $\alpha$, más exigente será el criterio para rechazar la hipótesis de inocencia**.

## 6. Los errores del juez

A pesar de que el juez aplica un criterio establecido por la ley, sus decisiones no están libres de errores.

Existen dos tipos de equivocaciones.

### Error de tipo I: condenar a un inocente

Supongamos que Juan no participó en el asesinato.

Sin embargo, el fiscal presenta evidencia suficientemente contundente para que el juez rechace $H_0$.

En consecuencia, Juan es condenado a pesar de ser inocente.

Este es un error de tipo I.

Su probabilidad está controlada por el nivel de significancia:

$$
\boxed{P(\text{Error I})\leq\alpha.}
$$

### Error de tipo II: no condenar a un culpable

Ahora supongamos que Juan sí participó en el asesinato.

Sin embargo, la evidencia presentada por el fiscal no resulta suficientemente contundente para que el juez rechace $H_0$.

En consecuencia, Juan no es condenado a pesar de ser culpable.

Este es un error de tipo II.

Para una alternativa específica, su probabilidad se representa mediante

$$
\boxed{\beta=P(\text{No rechazar }H_0\mid H_1).}
$$

En general, esta probabilidad depende de las características de la alternativa considerada. No tiene por qué ser la misma para todas las formas posibles en que Juan pudo participar en el crimen.

La probabilidad complementaria,

$$
\boxed{1-\beta,}
$$

se conoce como **potencia de la prueba**.

Representa la probabilidad de rechazar $H_0$ cuando la alternativa considerada es verdadera.

En nuestra analogía, es la probabilidad de que el juez rechace la hipótesis de inocencia cuando Juan realmente es culpable, bajo una alternativa específica.

## 7. El veredicto

Después de que el fiscal presenta la evidencia y el valor-p, el juez debe emitir su veredicto.

Para ello, compara el valor-p con el nivel de significancia establecido por la ley.

El procedimiento contempla dos decisiones.

**Primera decisión: $p<\alpha$.**

La evidencia encontrada sería suficientemente poco frecuente bajo la hipótesis de inocencia.

Por lo tanto,

$$
\boxed{\text{Se rechaza }H_0.}
$$

En nuestra analogía, el juez considera que existe evidencia estadísticamente suficiente para condenar a Juan.

**Segunda decisión: $p\geq\alpha$.**

La evidencia encontrada no satisface el criterio establecido para rechazar la hipótesis de inocencia.

Por lo tanto,

$$
\boxed{\text{No se rechaza }H_0.}
$$

En nuestra analogía, Juan no es condenado.

Pero debemos tener cuidado con la interpretación.

No rechazar $H_0$ no significa que Juan sea inocente. Significa que la evidencia presentada no fue suficiente para rechazar la hipótesis de inocencia mediante el procedimiento establecido.

Tampoco significa que alguno de sus amigos sea culpable.

El fiscal no tiene la obligación de proponer otro sospechoso simplemente porque no consiguió demostrar la culpabilidad de Juan.

## 8. Conclusión

Nuestro juicio permite comprender la lógica fundamental de las pruebas de hipótesis.

Primero, establecemos una hipótesis nula y una alternativa.

Después, la ley fija un nivel de significancia $\alpha$, que controla la probabilidad de cometer un error de tipo I mediante un procedimiento estadístico previamente definido.

Posteriormente, el fiscal recopila la evidencia, calcula un estadístico de prueba y obtiene su valor-p.

Finalmente, el juez compara el valor-p con $\alpha$ para determinar si existe evidencia estadísticamente suficiente para rechazar $H_0$.

La relación entre ambos conceptos es especialmente importante:

**El nivel de significancia controla el comportamiento del procedimiento de decisión a largo plazo, mientras que el valor-p permite evaluar la evidencia obtenida en un caso particular.**

Y, sobre todo, debemos recordar que

$$
\boxed{\text{No rechazar }H_0\neq\text{Demostrar }H_0.}
$$

Así como no encontrar evidencia suficiente para condenar a Juan no demuestra su inocencia ni señala automáticamente a otro culpable, una prueba de hipótesis que no rechaza $H_0$ tampoco demuestra que esta sea verdadera.

La conclusión correcta es, sencillamente, que **no contamos con evidencia estadísticamente suficiente para rechazarla**.

---

**Nota sobre la analogía.** En un juicio real, la culpabilidad se determina mediante estándares jurídicos de prueba, no mediante valores-p. La ley estadística descrita en este ejemplo es ficticia. Su finalidad es ilustrar el funcionamiento de las pruebas de hipótesis, el nivel de significancia, el valor-p y los errores de decisión.
