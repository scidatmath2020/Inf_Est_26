# Ejercicios de Análisis de Varianza (ANOVA)

## Ejercicio 1. Comparación de métodos de enseñanza

Una escuela desea comparar tres métodos de enseñanza. Para ello, selecciona cuatro estudiantes por método y registra sus calificaciones.

| Método A | Método B | Método C |
|---:|---:|---:|
| 6 | 8 | 10 |
| 7 | 9 | 11 |
| 8 | 10 | 12 |
| 9 | 11 | 13 |

Utiliza ANOVA con un nivel de significancia de $\alpha=0.05$.

**Preguntas:**

a) ¿Cuáles son las hipótesis nula y alternativa?

b) ¿Cuáles son las medias muestrales?

c) ¿Existe evidencia de diferencias entre las medias poblacionales?

d) ¿Podemos afirmar que los tres métodos son diferentes entre sí?

### Respuestas

a) La hipótesis nula establece que las tres medias poblacionales son iguales:

$$
H_0:\mu_A=\mu_B=\mu_C.
$$

La alternativa establece que al menos dos medias son diferentes.

b) Las medias muestrales son 7.5, 9.5 y 11.5, respectivamente.

c) El ANOVA produce un p-valor aproximado de 0.0059. Como es menor que 0.05, rechazamos la hipótesis nula.

Existe evidencia estadística de que no todas las medias poblacionales son iguales.

d) No. ANOVA solamente permite concluir que al menos dos medias son diferentes. Para identificar cuáles, necesitamos comparaciones posteriores.

## Ejercicio 2. Tres fertilizantes

Un investigador desea comparar el crecimiento de plantas sometidas a tres fertilizantes. Después de varias semanas, registra sus alturas en centímetros.

| Fertilizante A | Fertilizante B | Fertilizante C |
|---:|---:|---:|
| 18 | 19 | 18 |
| 20 | 21 | 20 |
| 22 | 23 | 22 |
| 24 | 25 | 24 |
| 26 | 27 | 26 |

Se utiliza un ANOVA con $\alpha=0.05$.

**Preguntas:**

a) ¿Cuáles son las medias muestrales?

b) ¿Las diferencias observadas entre los promedios garantizan que las medias poblacionales son distintas?

c) ¿Qué conclusión obtenemos mediante ANOVA?

### Respuestas

a) Las medias son 22, 23 y 22 centímetros.

b) No. Las diferencias entre medias muestrales pueden ser consecuencia del azar.

c) El estadístico F es aproximadamente 0.233 y el p-valor es aproximadamente 0.796.

Como el p-valor es mayor que 0.05, no rechazamos la hipótesis nula.

No tenemos evidencia suficiente para afirmar que los fertilizantes producen diferentes alturas promedio.

Esto no demuestra que los tres fertilizantes tengan exactamente el mismo efecto.

## Ejercicio 3. Interpretación del p-valor

Una empresa compara la productividad de trabajadores que utilizan cuatro herramientas diferentes.

Después de realizar ANOVA, obtiene:

$$
p=0.032.
$$

**Preguntas:**

a) ¿Qué decisión tomarías si $\alpha=0.05$?

b) ¿Qué decisión tomarías si $\alpha=0.01$?

c) ¿Por qué cambia la decisión si los datos son los mismos?

d) ¿El p-valor representa la probabilidad de que la hipótesis nula sea verdadera?

### Respuestas

a) Como $0.032<0.05$, rechazamos la hipótesis nula.

b) Como $0.032>0.01$, no rechazamos la hipótesis nula.

c) Porque estamos utilizando diferentes niveles de significancia. Con $\alpha=0.01$ exigimos evidencia más fuerte para rechazar la hipótesis nula.

d) No. El p-valor representa la probabilidad, suponiendo verdadera la hipótesis nula, de obtener un estadístico de prueba al menos tan extremo como el observado.

## Ejercicio 4. Comparaciones posteriores: Tukey

Una investigación compara el tiempo promedio necesario para realizar una tarea utilizando tres programas informáticos.

Se aplica ANOVA y se obtiene un p-valor de 0.002.

Posteriormente, se realiza una prueba de Tukey, cuyos resultados son:

| Comparación | p-valor ajustado |
|---|---:|
| A frente a B | 0.018 |
| A frente a C | 0.001 |
| B frente a C | 0.420 |

Considera $\alpha=0.05$.

**Preguntas:**

a) ¿Qué nos permite concluir ANOVA?

b) ¿Qué parejas presentan diferencias estadísticamente significativas?

c) ¿Podemos afirmar que los tres programas tienen tiempos promedio diferentes entre sí?

### Respuestas

a) ANOVA permite concluir que no todas las medias poblacionales son iguales.

b) Existen diferencias estadísticamente significativas entre A y B, así como entre A y C.

No encontramos evidencia suficiente de diferencias entre B y C.

c) No. Aunque A presenta diferencias significativas respecto de B y C, no encontramos evidencia suficiente para afirmar que B y C tienen medias diferentes.

## Ejercicio 5. ¿Cuándo utilizar ANOVA de Welch?

Un investigador compara los salarios mensuales de tres sectores económicos.

Obtiene la siguiente información:

| Sector | Tamaño muestral | Media | Desviación estándar |
|---|---:|---:|---:|
| A | 20 | 15,000 | 1,000 |
| B | 35 | 17,000 | 4,000 |
| C | 50 | 19,000 | 8,000 |

**Preguntas:**

a) ¿Los grupos presentan una dispersión semejante?

b) ¿Qué condición del ANOVA clásico podría resultar problemática?

c) ¿Qué alternativa sería conveniente considerar?

d) ¿Es correcto elegir automáticamente el grupo C como el mejor sector para trabajar?

### Respuestas

a) No. Las desviaciones estándar muestran diferencias considerables entre los sectores.

b) La homogeneidad de varianzas, ya que las poblaciones podrían presentar varianzas muy distintas.

c) Sería conveniente considerar ANOVA de Welch, que permite comparar las medias sin exigir igualdad de varianzas.

d) No necesariamente. Aunque el sector C tiene la mayor media muestral, necesitamos estudiar la incertidumbre de las estimaciones y, además, considerar otros factores laborales.

## Ejercicio 6. Significancia estadística e importancia práctica

Un laboratorio compara tres tratamientos médicos.

Después de estudiar a 3,000 pacientes, obtiene las siguientes reducciones promedio de presión arterial:

| Tratamiento | Reducción promedio |
|---|---:|
| A | 10.0 |
| B | 10.1 |
| C | 10.2 |

El ANOVA produce:

$$
p=0.0004.
$$

**Preguntas:**

a) ¿Existe evidencia estadística de diferencias entre las medias poblacionales?

b) ¿Podemos afirmar que las diferencias son importantes desde el punto de vista médico?

c) ¿Por qué un tamaño muestral grande puede producir un p-valor pequeño incluso cuando las diferencias son reducidas?

d) ¿Qué información adicional necesitaríamos?

### Respuestas

a) Sí. Como $0.0004<0.05$, rechazamos la hipótesis nula.

b) No necesariamente. Las diferencias observadas son pequeñas y su importancia clínica debe evaluarse por separado.

c) Porque, al aumentar el tamaño muestral, generalmente aumenta la precisión de las estimaciones y la capacidad de detectar diferencias pequeñas.

d) Necesitaríamos conocer la magnitud de los efectos, su incertidumbre, los posibles efectos secundarios y los criterios médicos para considerar relevante una diferencia.

## Reflexión final

Estos ejercicios muestran que ANOVA no consiste solamente en obtener un p-valor.

Antes de aplicar la prueba debemos comprender el problema, identificar las poblaciones que queremos comparar y examinar si los supuestos del procedimiento son razonables.

Después debemos interpretar correctamente los resultados.

En particular, es importante recordar que:

- Rechazar la hipótesis nula no significa que todas las medias sean diferentes.
- No rechazarla no demuestra que las medias sean iguales.
- Una diferencia estadísticamente significativa no necesariamente tiene importancia práctica.
- Cuando necesitamos identificar qué grupos difieren, podemos recurrir a comparaciones posteriores como Tukey.
- Si las varianzas poblacionales son diferentes, ANOVA de Welch puede ser una alternativa más adecuada.

**El objetivo de la inferencia estadística no es simplemente obtener resultados significativos, sino utilizar las muestras para formular conclusiones razonables sobre las poblaciones.**
