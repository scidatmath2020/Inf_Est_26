La hipótesis nula en un ANOVA es

$$H_0: \mu_1=\mu_2=\mu_3=...=\mu_k$$

Por lo tanto, la alternativa es $$H_1:\mbox{hay al menos dos poblaciones donde los promedios son diferentes}$$

Por ejemplo, negar la proposición "$a=b=c$" no significa que todas sean distintas, sino que existen al menos dos diferentes.

Es decir, cuando negamos "$a=b=c$" se aceptan casos como que "$a\neq b$ pero puede suceder $a=c$"

**Ejemplo 1**

En un estudio clínico se prueban tres medicamentos A, B, C para una misma enfermedad. Entonces tomas tres grupos de pacientes. Al primero le aplicas A, al segundo le aplicas B y al tercero le aplicas C. En este caso la hipótesis nula "informalmente" significa

$$H_0: \mbox{los tres medicamentos dan los mismos resultados}$$





Por lo tanto $$H_1:\mbox{al menos hay dos medicamentos que dan diferente resultado}$$
