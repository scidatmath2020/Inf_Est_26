Recordemos que, si **la población es gaussiana** o $n$ es grande, los estadísticos $$Z=\frac{\overline{X}-\mu}{\sigma/\sqrt{n}}\mbox{ y }T=\frac{\overline{X}-\mu}{S/\sqrt{n}}$$

siguen una distribución exactamente $N(0,1)$ y $T_{n-1}$ o al menos aproximadamente.

En las pruebas de gaussianidad, no rechazar $H_0$ **no significa que la población sea gaussiana**, sino que significa que no hay evidencia de que no lo sea

El error de tipo II significa algo como *al parecer no hay evidencia para rechazar la gaussianidad. Sin embargo, dios sabe que la población en realidad no es gaussiana*

Un $p$-valor "grande" significa NO rechazar $H_0$.

Un $p$-valor "pequeño" significa que tienes evidencia significativa de que la población **no es gaussiana**

## Función de distribución de probabilidad

Si $X$ es una variable aleatoria, se define su función de distribución $F$ como $$F(x)=P(X\le x)$$

**Ejemplo: lanzamiendo de un dado**

Sea $X$ el valor que se obtiene al lanzar un dado justo. 


$$F(4.5)=P(X\le 4.5)=P(1)+P(2)+P(3)+P(4)=\frac{4}{6}$$
