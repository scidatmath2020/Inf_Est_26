# ANOVA

¿Qué es un ANOVA? **¿Qué problemas me va a resolver?**

Hasta ahora hemos utilizado pruebas de hipótesis para investigar el comportamiento de parámetros poblacional y también para establecer si la característica de una población es gaussiana.

¿Qué sucede si quiero comparar varias poblaciones?

Imaginemos una escuela donde se aplican tres métodos de enseñanza. Después de aplicar cada método a tres grupos de estudiantes, se obtienen los siguientes resultados

| Método A | Método B | Método C |
|---:|---:|---:|
| 6 | 8 | 10 |
| 7 | 9 | 11 |
| 8 | 10 | 12 |
| 9 | 11 | 13 |

Las medias muestrales son

$$
\overline X_1=7.5,\qquad
\overline X_2=9.5,\qquad
\overline X_3=11.5.
$$

El problema que resuelve ANOVA es el siguiente: se tienen $k$ poblaciones independientes ($k\ge 2$). En cada una mido cierta característica. Por lo tanto cada una tiene su propia media: $\mu_1$, $\mu_2$,...,$\mu_k$. ¿Será que hay al menos una de esas medias poblacionales diferente de las otras?

$$\begin{array}{l}H_0: \mu_1=\mu_2=...=\mu_k\\H_1:\mbox{existen al menos dos medias diferentes}\end{array}$$

Se llama ANOVA porque significa "análisis de la varianza"

En el caso de la escuela, observamos que los promedios son diferentes. Sin embargo esto no signifia que los métodos produzcan resultados distintos en las poblaciones de los estudiantes. Recordemos que estamos tratando con muestras. Por lo tanto las diferencias observadas bien podrían deberse al azar.

¿Tenemos evidencia suficiente para afirmar que al menos de los métodos produce un rendimiento promedio diferente?

## Obervación 1.

Notemos que si rechazamos $H_0$ entonces hay al menos 2 medias poblaciones diferentes. El problema es que el ANOVA no nos dice quiénes son esos diferentes. Para hallarlos se requiere lo que se conce como Prueba Ad hoc.


## Observación 2.

El hecho de tener evidencia estadísticamente fuerte de que hay al menos dos medias diferentes no significa en la práctica que se trate de una diferencia importante.




## Observación 3.

Para poder aplicar un ANOVA se deben cumplir ciertas condiciones:
- Las poblaciones son independientes (y por lo tanto las muestras)
- Cada población debe ser gaussiana
- Para el ANOVA clásico se pide que las varianzas poblaciones sean iguales (lo que se conoce como homocedasticidad). Pero en el caso en que esto no se cumpla, se aplica lo que se conoce como ANOVA de Welch. Es decir, es un tipo de ANOVA que no exige homocedasticidad (esto se conoce como heterocedasticidad).
