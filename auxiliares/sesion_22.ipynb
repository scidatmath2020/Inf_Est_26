## Prueba Z

La **prueba Z** la vamos a aplicar cuando conocemos la desviación $\sigma$ siempre que
- La población es gaussiana. En este caso se va a obtener un p-valor exacto
- Si la población no es gaussiana pero la muestra es grande. En este caso se va a obtener un p-valor aproximado

#### Ejemplo de duración promedio de una batería cuando se conoce $\sigma$

Un investigador desea determinar si la duración media es inferior a 10

$$H_0: \mu \ge 10\mbox{ , }H_1: \mu < 10$$

Toma $n=100$ y resulta que cuando toma la muestra obtiene $\overline{x}=9.7$. Supongamos que sabe que la desviación $\sigma$ vale 1.5

Entonces $$z_{obs}=\frac{\overline{x}-10}{1.5/\sqrt{100}}=-2$$

## Prueba T

La **prueba T** la vamos a aplicar cuando desconocemos la desviación $\sigma$ siempre que
- La población es gaussiana. En este caso se va a obtener un p-valor exacto
- Si la población no es gaussiana pero la muestra es grande. En este caso se va a obtener un p-valor aproximado

El estadístico de constraste es $$T=\frac{\overline{X}-\mu_0}{S/\sqrt{n}}$$

Si $H_0$ es verdadera, entonces $T$ es una student con $n-1$ grados de libertad

#### Ejemplo de duración promedio de una batería cuando no se conoce $\sigma$

Un investigador desea determinar si la duración media es inferior a 10

$$H_0: \mu \ge 10\mbox{ , }H_1: \mu < 10$$

Digamos que el investigador toma $n=25$ baterías y le resulta $\overline{x}=9.2$ y $s=2$

El $p$-valor es $p=P(T_{24}\le t_{obs})$

Pero $$t_{obs}=\frac{\overline{x}-\mu_0}{s/\sqrt{n}}=\frac{9.2-10}{2/\sqrt{25}}=-2$$
