## Estandarización de una variable aleatoria

Sea $X$ es una variable aleatoria con $E[X]=\mu$ y desviación $\sigma$ y $Z=\frac{X-\mu}{\sigma}$. Entonces $E[Z]=0$ y $Desv(Z)=1$.


Al proceso de restar la media y dividir entre la desviación se le conoce como **estandarización**.

**Ejemplo.**

Tomemos datos = 80, 38, 22, 23, 55, 44, 15, 66, 12.9, -1.809, 456.9, 8.15, 1000.3,-8794014, 3.14, -1.333,1.445

Sea $\mathbb{M}$ la media y $\mathbb{S}$ la desviación estándar. Es decir



$$\mathbb{M}=\frac{x_1+x_2+...+x_n}{n}\mbox{ y }\mathbb{S}=\sqrt{\frac{(x_1-\mathbb{M})^2+(x_2-\mathbb{M})^2+...+(x_n-\mathbb{M})^2}{n}}$$



Entonces el promedio de $\frac{dato-\mathbb{M}}{\mathbb{S}}$ es 0 y la desviación es 1

**Demostración**

Tengo $Z=\frac{X-\mu}{\sigma}$. Y entonces $$E\left[Z\right]=E\left[\frac{X-\mu}{\sigma}\right]=\frac{E[X]-E[\mu]}{\sigma}=\frac{\mu-\mu}{\sigma}=0$$

$$Var(Z)=Var\left(\frac{X-\mu}{\sigma}\right)=\frac{Var(X)}{\sigma^2}=\frac{\sigma^2}{\sigma^2}=1$$


## Media y desviación de la media muestral

Sea $X_1,X_2,...,X_n$ una muestra aleatoria que pertenecen a una población con media $\mu$ y desviación $\sigma$

Hemos visto anteriormente que $E[\overline{X}]=\mu$ y que además $Desv(\overline{X})=\frac{\sigma}{\sqrt{n}}$. En los datos, ¿qué significa esto?

Tomemos $m$ muestras, todas de tamaño $n$, de una población (por ejemplo, tomo $m=20$ muestras de tamaño $n=60$ de la tabla de sismos). Calculemos la media de cada una de esas muestras. Por lo tanto obtenemos $m$ valores.

Que $E[\overline{X}]=\mu$ significa que a medida que $m$ sea grande, el promedio de los promedios de esas $m$ medias se aproxima más a $\mu$.

Ahora, calculemos la desviación de esos $m$ valores. Que $Desv(\overline{X})=\frac{\sigma}{\sqrt{n}}$ significa que esas $m$ medias se dispersan al rededor de $\mu$ mas o menos en $\frac{\sigma}{\sqrt{n}}$



## Suma y promedio de gaussianas

Una característica $X$ de una población se dice *gaussiana* o *normal* si 

$$P(X\le a)=\frac{1}{\sigma\sqrt{2\pi}}\int_{-\infty}^a e^{-\frac{(x-\mu)^2}{2\sigma^2}}\mathrm{d}x$$

donde $\mu$ es el valor promedio de la característica $X$ y $\sigma$ es su desviación. Esto se denota por $$X\sim N(\mu,\sigma^2)$$

---

En general, si $X\sim N(\mu_X,\sigma^2_X)$ y $Y\sim N(\mu_Y,\sigma^2_Y)$ son independientes, entonces $X+Y\sim N(\mu_X+\mu_Y,\sigma_X^2+\sigma^2_Y)$

Más aún... si $a$ es un número cualquiera ($a\neq 0$) entonces $a\cdot X\sim N(a\mu_X,a^2\sigma^2_X)$

Por lo tanto, las combinaciones lineales de gaussias son gaussianas. Es decir, si tengo las características $$X_1,X_2,...,X_n$$  y son gaussianas independientes y $a_1,a_2,...,a_n$ son números no todos cero, entonces $$a_1X_1+a_2X_2+...+a_nX_n$$ vuelve a ser gaussiana

**Otra propiedad bien importante** Si $X_1,X_2,...,X_n$ son gaussianas independientes entonces $\frac{X_1+X_2+...+X_n}{n}\sim N$
