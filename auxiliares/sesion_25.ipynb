Recordemos que las pruebas de Gaussianidad tienen la forma $$H_0:\mbox{ la población es gaussiana; }H_1:\mbox{ la población no es gaussiana}$$

Recordmos que si $p$-valor es pequeño, tenemos evidencia para rechazar $H_0$ en favor de $H_1$. Es decir, si el $p$-valor es pequeño, es muy difícil que realmente la población sea gaussiana

Por el contrario, si el $p$-valor es grande, no podemos rechazar $H_0$. Es decir, no puedo aceptero la **no gaussianidad**. Esto no significa que la población sea gaussiana
