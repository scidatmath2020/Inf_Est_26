# Intervalos de confianza para la media poblacional con varianza desconocida

## ¿Qué ocurre cuando desconocemos la desviación estándar poblacional?

En la sección anterior aprendimos a construir intervalos de confianza para la media poblacional $\mu$ cuando la desviación estándar poblacional $\sigma$ es conocida. Si $X_1,\dots,X_n$ son variables aleatorias independientes e idénticamente distribuidas con distribución $N(\mu,\sigma^2)$, entonces

$$
\overline X\sim N\left(\mu,\frac{\sigma^2}{n}\right)
$$

y, por tanto,

$$
Z=\frac{\overline X-\mu}{\sigma/\sqrt n}\sim N(0,1).
$$

A partir de esta distribución, construimos un intervalo de confianza de nivel $1-\alpha$:

$$
\left(\overline X-z_{1-\alpha/2}\frac{\sigma}{\sqrt n},\quad \overline X+z_{1-\alpha/2}\frac{\sigma}{\sqrt n}\right).
$$

Sin embargo, conocer $\sigma$ es una condición muy exigente en problemas reales. Pensemos nuevamente en los tiempos de traslado de los estudiantes de una universidad. Podemos encuestar a una muestra y calcular su promedio, pero normalmente tampoco conocemos la desviación estándar de los tiempos de traslado de *todos* los estudiantes.

Surge entonces una pregunta natural: ¿podemos sustituir $\sigma$ por una cantidad calculada a partir de la muestra?

La respuesta es sí, pero hacerlo modifica la distribución del estadístico que utilizaremos para construir el intervalo.

## La varianza muestral

Sea $X_1,\dots,X_n$ una muestra aleatoria de una población con media $\mu$ y varianza $\sigma^2$ finita. La varianza muestral se define como

$$
S^2=\frac{1}{n-1}\sum_{i=1}^{n}(X_i-\overline X)^2.
$$

Recordemos por qué aparece el denominador $n-1$. La identidad

$$
\sum_{i=1}^{n}(X_i-\overline X)^2
=\sum_{i=1}^{n}(X_i-\mu)^2-n(\overline X-\mu)^2
$$

nos permite calcular su esperanza. Como $E[(X_i-\mu)^2]=\sigma^2$ y $E[(\overline X-\mu)^2]=\sigma^2/n$, obtenemos

$$
\begin{aligned}
E\left[\sum_{i=1}^{n}(X_i-\overline X)^2\right]
&=n\sigma^2-n\frac{\sigma^2}{n}\\
&=(n-1)\sigma^2.
\end{aligned}
$$

Por consiguiente,

$$
E(S^2)=\sigma^2.
$$

Es decir, $S^2$ es un estimador insesgado de la varianza poblacional. Es importante distinguir esta propiedad de la de $S$: en general, $S$ no es un estimador insesgado de $\sigma$, aunque sí es consistente bajo las hipótesis usuales de muestreo independiente y varianza finita.

Si sustituimos $\sigma$ por $S$ en la expresión que utilizábamos para estandarizar la media muestral, obtenemos

$$
T=\frac{\overline X-\mu}{S/\sqrt n}.
$$

¿Por qué no podemos afirmar simplemente que $T$ sigue una distribución normal estándar?

Porque $\sigma$ es un parámetro fijo, mientras que $S$ es una variable aleatoria. En distintas muestras, no solo cambia $\overline X$: también cambia $S$. Al dividir entre un denominador aleatorio incorporamos una fuente adicional de variabilidad. Debemos averiguar cuál es la distribución de $T$.

## Dos resultados fundamentales de las muestras gaussianas

Por ahora supondremos que

$$
X_1,\dots,X_n\overset{\mathrm{iid}}{\sim}N(\mu,\sigma^2),\qquad n\geq 2,\quad\sigma>0.
$$

Bajo esta hipótesis se cumplen tres propiedades:

1. La media muestral estandarizada tiene distribución normal estándar:

$$
Z=\frac{\overline X-\mu}{\sigma/\sqrt n}\sim N(0,1).
$$

2. La varianza muestral estandarizada tiene distribución chi-cuadrada:

$$
V=\frac{(n-1)S^2}{\sigma^2}\sim\chi^2_{n-1}.
$$

3. La media muestral $\overline X$ y la varianza muestral $S^2$ son independientes.

La segunda y tercera propiedades son características especialmente importantes de las muestras gaussianas. Aquí las utilizaremos como resultados de teoría de distribuciones muestrales. Más adelante podremos estudiar con detenimiento la distribución chi-cuadrada al construir intervalos para la varianza poblacional.

### ¿Qué significa una distribución chi-cuadrada?

Si $Z_1,\dots,Z_k$ son variables aleatorias independientes con distribución normal estándar, entonces

$$
Z_1^2+\cdots+Z_k^2\sim\chi^2_k.
$$

El número $k$ recibe el nombre de *grados de libertad*. A diferencia de la normal estándar, una variable chi-cuadrada no puede tomar valores negativos y su distribución no es simétrica.

En nuestro caso aparecen $n-1$ grados de libertad. Esto se relaciona con la identidad

$$
\sum_{i=1}^{n}(X_i-\overline X)=0.
$$

Si conocemos $n-1$ desviaciones respecto a la media muestral, la última ya está determinada. Por ello solo hay $n-1$ desviaciones que pueden variar libremente. Esta explicación ayuda a entender los grados de libertad, aunque la demostración rigurosa de la distribución chi-cuadrada requiere un resultado adicional sobre muestras normales.

## La distribución t de Student

Consideremos dos variables aleatorias independientes tales que

$$
Z\sim N(0,1),\qquad V\sim\chi^2_{\nu}.
$$

Se define una variable aleatoria con distribución $t$ de Student y $\nu$ grados de libertad mediante

$$
T=\frac{Z}{\sqrt{V/\nu}}\sim t_{\nu}.
$$

Apliquemos esta definición a nuestra muestra gaussiana. Tomando

$$
Z=\frac{\overline X-\mu}{\sigma/\sqrt n},
\qquad
V=\frac{(n-1)S^2}{\sigma^2},
\qquad \nu=n-1,
$$

tenemos

$$
\begin{aligned}
T
&=\frac{Z}{\sqrt{V/(n-1)}}\\[4pt]
&=\frac{\dfrac{\overline X-\mu}{\sigma/\sqrt n}}{\sqrt{\dfrac{(n-1)S^2/\sigma^2}{n-1}}}\\[8pt]
&=\frac{\dfrac{\overline X-\mu}{\sigma/\sqrt n}}{S/\sigma}\\[8pt]
&=\frac{\overline X-\mu}{S/\sqrt n}.
\end{aligned}
$$

Por consiguiente,

$$
\frac{\overline X-\mu}{S/\sqrt n}\sim t_{n-1}.
$$

Este resultado es exacto para muestras independientes procedentes de una población normal. No se necesita que $n$ sea grande.

## ¿Cómo se compara la distribución t con la normal estándar?

La distribución $t$ de Student es simétrica respecto de cero y tiene forma de campana, igual que la normal estándar. Sin embargo, sus colas son más pesadas: asigna mayor probabilidad a valores muy alejados de cero. Esto refleja la incertidumbre adicional que surge al estimar $\sigma$ mediante $S$.

La forma exacta de la distribución depende de sus grados de libertad $\nu$. Cuando $\nu$ es pequeño, sus colas son notoriamente pesadas; cuando $\nu$ crece, la distribución se acerca a la normal estándar:

$$
t_{\nu}\xrightarrow[\nu\to\infty]{\mathcal D}N(0,1).
$$

Esto también es razonable: al aumentar el tamaño de la muestra, la desviación estándar muestral tiende a estabilizarse alrededor de la desviación estándar poblacional.

### Exploración gráfica en R

Podemos comparar la densidad de la normal estándar con las densidades de distribuciones $t$ con distintos grados de libertad. En R, `dnorm()` evalúa la densidad normal y `dt()` evalúa la densidad $t$.

```r
# Valores sobre los que evaluaremos las densidades
x <- seq(-4, 4, length.out = 1000)

# Densidad normal estándar
normal <- dnorm(x)

# Densidades t con diferentes grados de libertad
t_2 <- dt(x, df = 2)
t_5 <- dt(x, df = 5)
t_30 <- dt(x, df = 30)

plot(x, normal, type = "l", lwd = 3, col = "black",
     xlab = "Valor", ylab = "Densidad",
     main = "Normal estándar y t de Student")
lines(x, t_2, col = "red", lwd = 2)
lines(x, t_5, col = "blue", lwd = 2)
lines(x, t_30, col = "darkgreen", lwd = 2)
legend("topright", legend = c("Normal", "t(2)", "t(5)", "t(30)"),
       col = c("black", "red", "blue", "darkgreen"), lwd = c(3, 2, 2, 2))
```

Hay que prestar atención al centro y a las colas de las curvas. Para pocos grados de libertad, la curva $t$ es menos concentrada en el centro y asigna más probabilidad a valores extremos. Para $30$ grados de libertad, las diferencias ya son considerablemente menores.

## Construcción del intervalo de confianza para la media

Supongamos de nuevo que

$$
X_1,\dots,X_n\overset{\mathrm{iid}}{\sim}N(\mu,\sigma^2),
$$

pero ahora tanto $\mu$ como $\sigma^2$ son desconocidos.

Sabemos que

$$
T=\frac{\overline X-\mu}{S/\sqrt n}\sim t_{n-1}.
$$

Sea $1-\alpha$ el nivel de confianza deseado. Denotemos por $t_{1-\alpha/2,n-1}$ el cuantil de orden $1-\alpha/2$ de una distribución $t$ con $n-1$ grados de libertad. Por la simetría de esta distribución,

$$
P\left(-t_{1-\alpha/2,n-1}\leq T\leq t_{1-\alpha/2,n-1}\right)=1-\alpha.
$$

Sustituyendo la expresión de $T$, obtenemos

$$
P\left(-t_{1-\alpha/2,n-1}\leq\frac{\overline X-\mu}{S/\sqrt n}\leq t_{1-\alpha/2,n-1}\right)=1-\alpha.
$$

Como $S>0$ con probabilidad uno bajo nuestro modelo normal no degenerado, multiplicamos las desigualdades por $S/\sqrt n$:

$$
P\left(-t_{1-\alpha/2,n-1}\frac{S}{\sqrt n}\leq\overline X-\mu\leq t_{1-\alpha/2,n-1}\frac{S}{\sqrt n}\right)=1-\alpha.
$$

Reordenando para dejar $\mu$ en el centro,

$$
P\left(\overline X-t_{1-\alpha/2,n-1}\frac{S}{\sqrt n}\leq\mu\leq\overline X+t_{1-\alpha/2,n-1}\frac{S}{\sqrt n}\right)=1-\alpha.
$$

Por tanto, un intervalo de confianza de nivel $1-\alpha$ para la media poblacional es

$$
\left(\overline X-t_{1-\alpha/2,n-1}\frac{S}{\sqrt n},\quad\overline X+t_{1-\alpha/2,n-1}\frac{S}{\sqrt n}\right).
$$

Su centro es la media muestral $\overline X$. Su margen de error es

$$
E=t_{1-\alpha/2,n-1}\frac{S}{\sqrt n},
$$

y su amplitud total es $2E$.

Observemos que seguimos una estructura conocida: estimación puntual $\pm$ margen de error. Lo que cambia es que ahora el error estándar se estima con $S/\sqrt n$ y el valor crítico procede de la distribución $t$.

## Ejemplo numérico: tiempos de traslado

Supongamos que seleccionamos aleatoriamente $25$ estudiantes de una universidad. Los tiempos de traslado se modelan mediante una distribución normal cuya media y varianza poblacionales desconocemos. La muestra proporciona

$$
n=25,\qquad\overline x=35\text{ minutos},\qquad s=12\text{ minutos}.
$$

Construiremos un intervalo de confianza del $95\%$ para el tiempo promedio de traslado de todos los estudiantes.

El nivel de confianza es $1-\alpha=0.95$, así que $\alpha=0.05$. Como $n=25$, tenemos $24$ grados de libertad. El valor crítico es

$$
t_{0.975,24}\approx2.0639.
$$

El error estándar estimado es

$$
\frac{s}{\sqrt n}=\frac{12}{\sqrt{25}}=2.4\text{ minutos}.
$$

El margen de error es

$$
E=2.0639(2.4)\approx4.9534\text{ minutos}.
$$

En consecuencia, el intervalo es

$$
(35-4.9534,\quad35+4.9534)\approx(30.05,\quad39.95).
$$

Interpretación: utilizando este procedimiento tenemos un nivel de confianza del $95\%$ de que el intervalo construido contiene el verdadero tiempo promedio poblacional de traslado.

La precisión conceptual es importante: una vez observada la muestra, $\mu$ es un parámetro fijo y el intervalo concreto está fijado. El $95\%$ se refiere a la proporción de intervalos que contendrían a $\mu$ si repitiéramos el procedimiento de muestreo muchas veces.

### Reproducción en R

```r
# Resumen de la muestra
n <- 25
media <- 35
s <- 12

# Nivel de confianza
confianza <- 0.95
alpha <- 1 - confianza

# Grados de libertad y valor crítico
gl <- n - 1
t_critico <- qt(1 - alpha / 2, df = gl)

# Error estándar estimado y margen de error
error_estandar <- s / sqrt(n)
margen <- t_critico * error_estandar

# Límites del intervalo
LI <- media - margen
LS <- media + margen

round(c(inferior = LI, superior = LS), 2)
```

El resultado es aproximadamente:

```text
inferior superior
   30.05    39.95
```

La función `qt(p, df)` calcula el cuantil de orden `p` de una distribución $t$ con los grados de libertad indicados en `df`.

## Comparación con el caso de desviación estándar conocida

Cuando la población es normal y $\sigma$ es conocida, utilizamos

$$
\overline X\pm z_{1-\alpha/2}\frac{\sigma}{\sqrt n}.
$$

Cuando la población es normal y $\sigma$ es desconocida, utilizamos

$$
\overline X\pm t_{1-\alpha/2,n-1}\frac{S}{\sqrt n}.
$$

En el segundo caso, no solo cambia el valor crítico: la cantidad que representa la dispersión también se estima con los datos. No podemos asegurar que un intervalo particular sea más largo que otro sin conocer los valores observados de $S$ y $\sigma$. Sin embargo, para iguales valores numéricos del error estándar y con el mismo nivel de confianza, el valor crítico $t$ es mayor que el normal, por lo que el margen de error sería mayor.

## ¿Qué sucede si la población no es gaussiana?

El resultado exacto

$$
\frac{\overline X-\mu}{S/\sqrt n}\sim t_{n-1}
$$

depende de la hipótesis de normalidad poblacional. Si la población no es normal, esta igualdad en distribución no es válida en general.

Sin embargo, supongamos que las observaciones son independientes e idénticamente distribuidas y que la población tiene media $\mu$ y varianza finita $0<\sigma^2<\infty$. Por el Teorema del Límite Central,

$$
\frac{\overline X-\mu}{\sigma/\sqrt n}\xrightarrow[n\to\infty]{\mathcal D}N(0,1).
$$

Además, la consistencia de la varianza muestral implica

$$
\frac{S}{\sigma}\xrightarrow[n\to\infty]{P}1.
$$

Por el teorema de Slutsky,

$$
\frac{\overline X-\mu}{S/\sqrt n}\xrightarrow[n\to\infty]{\mathcal D}N(0,1).
$$

Esto permite construir intervalos aproximadamente válidos para muestras suficientemente grandes. Como los cuantiles $t_{1-\alpha/2,n-1}$ convergen a los normales, el intervalo basado en $t$ también tiene cobertura asintótica $1-\alpha$ bajo estas condiciones.

No existe, sin embargo, un número universal de observaciones que garantice una buena aproximación. Una población muy asimétrica o con valores extremadamente grandes puede requerir tamaños muestrales considerables. Si la varianza poblacional es infinita, los argumentos anteriores dejan de aplicarse.

## Exploración en R: el papel del tamaño muestral

El siguiente experimento permite comparar el valor crítico de la normal estándar con el de Student cuando cambia $n$. No necesita simular datos: se concentra exclusivamente en la diferencia entre los cuantiles.

```r
# Tamaños muestrales que vamos a comparar
n <- c(3, 5, 10, 20, 30, 50, 100, 500)

# Valores críticos para un nivel de confianza del 95 %
z_critico <- qnorm(0.975)
t_critico <- qt(0.975, df = n - 1)

# Tabla comparativa
comparacion <- data.frame(
  n = n,
  grados_libertad = n - 1,
  z = rep(z_critico, length(n)),
  t = t_critico
)

print(round(comparacion, 4))
```

¿Qué debemos observar? El valor crítico normal permanece fijo porque no depende de $n$. El valor crítico de Student sí depende de $n-1$: es más grande para muestras pequeñas y se aproxima al normal conforme $n$ aumenta.

## Ejercicio de consolidación

Una empresa fabrica piezas metálicas cuyo peso se distribuye normalmente. Se desconoce la media y la desviación estándar poblacionales. Una muestra aleatoria de $16$ piezas produce

$$
\overline x=250\text{ gramos},\qquad s=8\text{ gramos}.
$$

Construye intervalos de confianza para la media poblacional a niveles del $90\%$, $95\%$ y $99\%$.

Para cada nivel, identifica $\alpha$, los grados de libertad, el valor crítico de Student, el error estándar estimado, el margen de error y los extremos del intervalo. Después interpreta los resultados y responde:

1. ¿Por qué se utilizan $15$ grados de libertad?
2. ¿Por qué no usamos directamente los cuantiles de la normal estándar?
3. ¿Qué sucede con el margen de error al aumentar el nivel de confianza?
4. Si aumentáramos el tamaño de la muestra y la desviación estándar muestral permaneciera aproximadamente constante, ¿qué esperaríamos que ocurriera con la amplitud del intervalo?
5. ¿Por qué debemos distinguir entre la probabilidad de cobertura del procedimiento y la interpretación de un intervalo que ya fue calculado?

Puedes verificar tus cálculos con el siguiente código en R:

```r
n <- 16
media <- 250
s <- 8
confianzas <- c(0.90, 0.95, 0.99)

alpha <- 1 - confianzas
gl <- n - 1
criticos <- qt(1 - alpha / 2, df = gl)
error_estandar <- s / sqrt(n)
margenes <- criticos * error_estandar

resultado <- data.frame(
  confianza = confianzas,
  grados_libertad = rep(gl, length(confianzas)),
  valor_critico = criticos,
  margen_error = margenes,
  limite_inferior = media - margenes,
  limite_superior = media + margenes
)

print(round(resultado, 3))
```

## Para cerrar

En la sección anterior construimos intervalos de confianza para la media suponiendo conocida la desviación estándar poblacional. Ahora eliminamos esa suposición, algo fundamental para el análisis de datos reales.

El punto central fue comprender por qué la sustitución de $\sigma$ por $S$ no es un simple cambio de símbolos: modifica la distribución del estadístico. Bajo la hipótesis de normalidad poblacional, la independencia de $\overline X$ y $S^2$, junto con la distribución chi-cuadrada de la varianza muestral estandarizada, nos condujo a la distribución $t$ de Student.
