# Aplicaciones de Pruebas de hipótesis

En el capítulo anterior estudiamos los fundamentos de las pruebas de hipótesis. Aprendimos que una prueba estadística es un procedimiento que permite evaluar si la evidencia proporcionada por una muestra es suficiente para rechazar una afirmación sobre una población.

También introdujimos los conceptos de hipótesis nula, hipótesis alternativa, nivel de significancia, valor-p y errores de tipo I y II.

Ahora estudiaremos cómo aplicar estos conceptos a problemas concretos de inferencia estadística.

En particular, consideraremos pruebas de hipótesis sobre tres parámetros poblacionales:

1. La media poblacional $\mu$.
2. La proporción poblacional $p$.
3. La varianza poblacional $\sigma^2$.

En cada caso, nuestro objetivo será formular correctamente las hipótesis, identificar el estadístico de prueba, determinar su distribución bajo la hipótesis nula, calcular el valor-p e interpretar la decisión estadística.

## ¿Cómo elegir una hipótesis nula?

Supongamos que recibimos una base de datos que contiene información sobre 1000 sismos registrados en una región. Para cada sismo conocemos su magnitud, profundidad y fecha de ocurrencia. Antes de realizar cualquier prueba estadística, debemos preguntarnos qué queremos investigar. 

Por ejemplo, podríamos estudiar la magnitud media de los sismos, determinar qué proporción supera cierta magnitud o analizar la variabilidad de sus profundidades. Cada una de estas preguntas conduce a una prueba de hipótesis diferente.

Es importante comprender que **una muestra no determina por sí sola qué hipótesis debemos contrastar**. Las hipótesis surgen de una pregunta científica, de un conocimiento previo o de una afirmación que deseamos evaluar.

Cuando recibimos una base de datos por primera vez, lo más recomendable es comenzar con un análisis exploratorio y con la estimación de parámetros poblacionales. Posteriormente, podemos formular preguntas que justifiquen una prueba de hipótesis.

Sin embargo, debemos tener cuidado: si elegimos una hipótesis porque encontramos un resultado particularmente llamativo en los mismos datos, la prueba resultante tiene un carácter exploratorio. Para realizar inferencia confirmatoria puede ser necesario utilizar información independiente o procedimientos que consideren la selección realizada.

### Hipótesis bilaterales

Supongamos que queremos determinar si un parámetro poblacional $\theta$ es diferente de un valor de referencia $\theta_0$.

Planteamos

$$
\begin{array}{l}H_0:\theta=\theta_0\\H_1:\theta\neq\theta_0\end{array}
$$

Este contraste se conoce como **prueba bilateral**, porque buscamos evidencia de diferencias en cualquiera de las dos direcciones. Por ejemplo, si queremos determinar si la magnitud media de los sismos es diferente de 4, planteamos

$$
H_0:\mu=4,
\qquad
H_1:\mu\neq4.
$$

El valor 4 debe responder a una referencia o pregunta científica previamente establecida.

### Hipótesis unilaterales a la derecha

Supongamos que queremos demostrar que un parámetro poblacional supera cierto valor.

Planteamos

$$
\begin{array}{l}H_0:\theta\leq\theta_0\\H_1:\theta>\theta_0\end{array}
$$

En este caso, solamente nos interesa encontrar evidencia de que $\theta$ es mayor que $\theta_0$.

Por ejemplo, si queremos investigar si la magnitud media de los sismos supera 4, planteamos

$$
H_0:\mu\leq4,
\qquad
H_1:\mu>4.
$$

### Hipótesis unilaterales a la izquierda

Finalmente, si queremos demostrar que un parámetro es inferior a cierto valor, planteamos

$$
\begin{array}{l}H_0:\theta\geq\theta_0\\H_1:\theta<\theta_0\end{array}
$$

Por ejemplo, si queremos investigar si la profundidad media de los sismos es inferior a 20 kilómetros, planteamos

$$
H_0:\mu\geq20,
\qquad
H_1:\mu<20.
$$

En las pruebas unilaterales clásicas, frecuentemente calculamos el valor-p utilizando el caso límite $\theta=\theta_0$. Esto se debe a que, en muchos modelos, dicho valor permite controlar la probabilidad de cometer un error de tipo I para todos los valores incluidos en $H_0$. No obstante, esta propiedad debe justificarse para el procedimiento utilizado.

## El procedimiento general de una prueba de hipótesis

Aunque estudiaremos diferentes parámetros poblacionales, todas las pruebas siguen una misma estructura.

- Primero establecemos una pregunta científica y formulamos las hipótesis nula y alternativa.

- Después seleccionamos un nivel de significancia $\alpha$, que determina la probabilidad máxima tolerada de cometer un error de tipo I.

- A continuación, elegimos un estadístico de prueba $T$, cuya distribución bajo la hipótesis nula debe ser conocida, exactamente o mediante una aproximación justificada.

- Una vez obtenida la muestra, calculamos el valor observado del estadístico, que denotaremos por $t_{\text{obs}}$.

- Finalmente, calculamos el valor-p y lo comparamos con $\alpha$.

La regla general de decisión es

$$
p<\alpha
\quad\Longrightarrow\quad
\text{Rechazar }H_0.
$$

En caso contrario,

$$
p\geq\alpha
\quad\Longrightarrow\quad
\text{No rechazar }H_0.
$$

La igualdad $p=\alpha$ puede incluirse en la región de rechazo según la convención adoptada. En este capítulo utilizaremos la regla estricta $p<\alpha$.

Recordemos que no rechazar $H_0$ no significa demostrar que $H_0$ es verdadera. Significa únicamente que la evidencia disponible no fue suficiente para rechazarla mediante el procedimiento utilizado.


## Prueba de hipótesis para la media poblacional

Supongamos que queremos investigar la media poblacional de una variable cuantitativa. Por ejemplo, podríamos estudiar la magnitud media de los sismos registrados en una región. Sea $X_1,X_2,\ldots,X_n$ una muestra aleatoria de una población con media $\mu$ y varianza $\sigma^2$.

La media muestral es $\overline X=\frac{1}{n}\sum_{i=1}^nX_i$. Nuestro objetivo consiste en contrastar una afirmación sobre $\mu$.

### Formulación de las hipótesis

Supongamos que existe un valor de referencia $\mu_0$. Podemos plantear una prueba bilateral:

$$
H_0:\mu=\mu_0,
\qquad
H_1:\mu\neq\mu_0.
$$

También podemos considerar una prueba unilateral derecha:

$$
H_0:\mu\leq\mu_0,
\qquad
H_1:\mu>\mu_0.
$$

O una prueba unilateral izquierda:

$$
H_0:\mu\geq\mu_0,
\qquad
H_1:\mu<\mu_0.
$$

La elección depende de la pregunta científica y debe realizarse antes de examinar los resultados de la prueba.

### Estadístico de prueba cuando la varianza es conocida

Recordemos que, si las observaciones son independientes y proceden de una población normal, $\overline X\sim N\left(\mu,\frac{\sigma^2}{n}\right)$. Por lo tanto, bajo la hipótesis

$$
H_0:\mu=\mu_0,
$$

el estadístico

$$
Z=\frac{\overline X-\mu_0}{\sigma/\sqrt n}
$$

sigue una distribución normal estándar $Z\sim N(0,1)$. Esta es la base de la **prueba Z para una media poblacional**.

Si la población no es normal, pero las observaciones son independientes e idénticamente distribuidas, tienen varianza finita y el tamaño de muestra es suficientemente grande, podemos utilizar el Teorema del Límite Central para obtener una aproximación normal.

Es importante distinguir ambas situaciones. Cuando la población es normal, la distribución del estadístico es exactamente normal para cualquier tamaño de muestra; cuando utilizamos el teorema del límite central, la distribución normal es una aproximación que mejora conforme aumenta el tamaño de muestra, bajo las condiciones correspondientes.

### Cálculo del valor-p para la prueba Z

Sea $z_{\text{obs}}$ el valor observado del estadístico.

Para una prueba bilateral,

$$
H_0:\mu=\mu_0,
\qquad
H_1:\mu\neq\mu_0,
$$

el valor-p es

$$
p=2P(Z\geq|z_{\text{obs}}|).
$$

Para una prueba unilateral derecha,

$$
H_0:\mu\leq\mu_0,
\qquad
H_1:\mu>\mu_0,
$$

tenemos

$$
p=P(Z\geq z_{\text{obs}}).
$$

Finalmente, para una prueba unilateral izquierda,

$$
H_0:\mu\geq\mu_0,
\qquad
H_1:\mu<\mu_0,
$$

obtenemos

$$
p=P(Z\leq z_{\text{obs}}).
$$

En las pruebas unilaterales, estas probabilidades se calculan utilizando el caso límite $\mu=\mu_0$.

#### Ejemplo: duración media de una batería

Un fabricante afirma que sus baterías tienen una duración media de 10 horas. Un investigador desea determinar si la duración media es inferior a ese valor. Selecciona una muestra aleatoria de 100 baterías y obtiene $$n=100,\qquad\overline x=9.7.$$

Supongamos que conocemos la desviación estándar poblacional $\sigma=1.5$. 

Las hipótesis son

$$
H_0:\mu\geq10,
\qquad
H_1:\mu<10.
$$

El estadístico observado es

$$z_{\text{obs}}=\frac{9.7-10}{1.5/\sqrt{100}}.$$

Por lo tanto,

$$z_{\text{obs}}=\frac{-0.3}{0.15}=-2.$$

Como se trata de una prueba unilateral izquierda,

$$
p=P(Z\leq-2).
$$

Utilizando la distribución normal estándar,

$$
p\approx0.0228.
$$

Si establecemos

$$
\alpha=0.05,
$$

obtenemos

$$
p<\alpha.
$$

Por lo tanto, rechazamos $H_0$.

**Conclusión:** existe evidencia estadísticamente significativa, al nivel del $5\%$, de que la duración media de las baterías es inferior a 10 horas.

Observa que no afirmamos que la probabilidad de que el fabricante tenga razón sea del $2.28\%$.

El valor-p representa una probabilidad calculada bajo la hipótesis nula, no la probabilidad de que esta sea verdadera.

### Estadístico de prueba cuando la varianza es desconocida

En la práctica, generalmente desconocemos la varianza poblacional. Por ello, utilizamos la varianza muestral
$$
S^2=\frac{1}{n-1}\sum_{i=1}^n(X_i-\overline X)^2.
$$

El estadístico de prueba es

$$
T=\frac{\overline X-\mu_0}{S/\sqrt n}.
$$

Si la muestra procede de una población normal, entonces, bajo

$$
H_0:\mu=\mu_0,
$$

se cumple que

$$
\boxed{
T\sim t_{n-1}.
}
$$

Esta prueba se conoce como **prueba t de Student para una muestra**.

La distribución t de Student es simétrica alrededor de cero, pero presenta colas más pesadas que la distribución normal estándar. La diferencia es especialmente importante cuando el tamaño de muestra es pequeño. Conforme aumentan los grados de libertad, la distribución t de Student se aproxima a la normal estándar. Para muestras grandes, el estadístico también puede tener una distribución aproximadamente normal bajo condiciones apropiadas, aunque la población original no sea normal.

### Cálculo del valor-p para la prueba t

Supongamos que el estadístico observado es $t_{\text{obs}}$.

Para una prueba bilateral,

$$
p=2P(T_{n-1}\geq|t_{\text{obs}}|).
$$

Para una prueba unilateral derecha,

$$
p=P(T_{n-1}\geq t_{\text{obs}}).
$$

Y para una prueba unilateral izquierda,

$$
p=P(T_{n-1}\leq t_{\text{obs}}).
$$

Estas probabilidades se calculan utilizando la distribución t de Student con $n-1$ grados de libertad.

#### Ejemplo: duración media de una batería con varianza desconocida

Retomemos el problema de las baterías, pero ahora supongamos que desconocemos la desviación estándar poblacional. Seleccionamos una muestra aleatoria de 25 baterías y obtenemos

$$
n=25,\qquad\overline x=9.2,\qquad s=2.
$$

Supondremos que las duraciones son independientes y proceden de una población aproximadamente normal. Las hipótesis son

$$
H_0:\mu\geq10,
\qquad
H_1:\mu<10.
$$

El estadístico observado es

$$
t_{\text{obs}}=\frac{9.2-10}{2/\sqrt{25}}.
$$

Por lo tanto,

$$
t_{\text{obs}}=\frac{-0.8}{0.4}=-2.
$$

Como la varianza poblacional es desconocida, utilizamos una distribución t de Student con $n-1=24$ grados de libertad.

El valor-p es

$$
p=P(T_{24}\leq-2).
$$

Por lo tanto,

$$
p\approx0.0285
$$

Si utilizamos

$$
\alpha=0.05,
$$

obtenemos

$$
p<\alpha.
$$

Por lo tanto, rechazamos $H_0$.

**Conclusión:** existe evidencia estadísticamente significativa de que la duración media de las baterías es inferior a 10 horas.


# 5. Prueba de hipótesis para una proporción poblacional

En numerosas investigaciones, el parámetro de interés no es una media de una variable cuantitativa, sino una proporción.

Por ejemplo, podríamos preguntarnos qué proporción de los sismos registrados en una región supera una magnitud de 5.

Para cada sismo, definimos

$$
X_i=
\begin{cases}
1,&\text{si su magnitud supera 5},\\
0,&\text{en caso contrario}.
\end{cases}
$$

Si las observaciones son independientes y tienen una misma probabilidad de éxito $p$, entonces

$$
X_i\sim\operatorname{Bernoulli}(p).
$$

El parámetro de interés es

$$
p=P(X_i=1).
$$

Si observamos una muestra de tamaño $n$, el número total de éxitos es

$$
X=\sum_{i=1}^nX_i.
$$

Por lo tanto,

$$
\boxed{
X\sim\operatorname{Binomial}(n,p).
}
$$

La proporción muestral es

$$
\widehat p=\frac{X}{n}.
$$

## 5.1. Formulación de las hipótesis

Supongamos que queremos contrastar una proporción poblacional con un valor de referencia $p_0$.

Para una prueba bilateral,

$$
H_0:p=p_0,
\qquad
H_1:p\neq p_0.
$$

Para una prueba unilateral derecha,

$$
H_0:p\leq p_0,
\qquad
H_1:p>p_0.
$$

Y para una prueba unilateral izquierda,

$$
H_0:p\geq p_0,
\qquad
H_1:p<p_0.
$$

## 5.2. La prueba binomial exacta

Recordemos que, bajo el caso límite

$$
p=p_0,
$$

tenemos

$$
X\sim\operatorname{Binomial}(n,p_0).
$$

Por lo tanto,

$$
P(X=k)
=
\binom nkp_0^k(1-p_0)^{n-k}.
$$

Esta distribución permite calcular el valor-p directamente, sin recurrir a una aproximación normal.

La prueba resultante se conoce como **prueba binomial exacta**.

### Prueba unilateral derecha

Si observamos $x$ éxitos y queremos contrastar

$$
H_0:p\leq p_0,
\qquad
H_1:p>p_0,
$$

el valor-p es

$$
\boxed{
p_{\text{valor}}
=
P(X\geq x\mid p=p_0).
}
$$

Por lo tanto,

$$
\boxed{
p_{\text{valor}}
=
\sum_{k=x}^{n}
\binom nkp_0^k(1-p_0)^{n-k}.
}
$$

### Prueba unilateral izquierda

Si queremos contrastar

$$
H_0:p\geq p_0,
\qquad
H_1:p<p_0,
$$

el valor-p es

$$
\boxed{
p_{\text{valor}}
=
P(X\leq x\mid p=p_0).
}
$$

Por lo tanto,

$$
\boxed{
p_{\text{valor}}
=
\sum_{k=0}^{x}
\binom nkp_0^k(1-p_0)^{n-k}.
}
$$

### Prueba bilateral

Para una prueba bilateral,

$$
H_0:p=p_0,
\qquad
H_1:p\neq p_0,
$$

debemos tener más cuidado.

Como la distribución binomial es discreta y puede ser asimétrica, no siempre es apropiado duplicar una de las probabilidades de cola.

Una definición habitual del valor-p bilateral exacto consiste en sumar las probabilidades de todos los resultados que, bajo $H_0$, son tan probables o menos probables que el resultado observado.

Es decir,

$$
\boxed{
p_{\text{valor}}
=
\sum_{\substack{k=0,\ldots,n\\
P_{p_0}(X=k)\leq P_{p_0}(X=x)}}
P_{p_0}(X=k).
}
$$

Existen otras convenciones para las pruebas binomiales bilaterales exactas, por lo que es importante identificar cuál se está utilizando.

## 5.3. Ejemplo: proporción de productos defectuosos

Una empresa afirma que la proporción de productos defectuosos no supera el $10\%$.

Un inspector selecciona aleatoriamente 20 productos y encuentra que 5 son defectuosos.

Queremos determinar si existe evidencia de que la proporción de defectuosos supera el $10\%$.

Las hipótesis son

$$
H_0:p\leq0.10,
\qquad
H_1:p>0.10.
$$

Tenemos

$$
n=20,\qquad x=5.
$$

La proporción muestral es

$$
\widehat p=\frac{5}{20}=0.25.
$$

Aunque el $25\%$ observado es considerablemente mayor que el $10\%$ declarado, debemos determinar si esta diferencia constituye evidencia estadística suficiente.

Bajo el caso límite de la hipótesis nula,

$$
X\sim\operatorname{Binomial}(20,0.10).
$$

El valor-p es

$$
\begin{aligned}
p_{\text{valor}}
&=P(X\geq5)\\
&=\sum_{k=5}^{20}
\binom{20}{k}(0.10)^k(0.90)^{20-k}\\
&\approx0.0432.
\end{aligned}
$$

Si utilizamos

$$
\alpha=0.05,
$$

obtenemos

$$
p_{\text{valor}}<\alpha.
$$

Por lo tanto, rechazamos $H_0$.

**Conclusión:** existe evidencia estadísticamente significativa de que la proporción poblacional de productos defectuosos supera el $10\%$.

Este procedimiento supone que los productos pueden modelarse mediante ensayos de Bernoulli independientes con una misma probabilidad de defecto.

Si se extraen sin reemplazo de un lote pequeño y finito, podría resultar más apropiado un modelo hipergeométrico.

# 6. Prueba de hipótesis para la varianza poblacional

Hasta ahora hemos estudiado pruebas sobre parámetros que describen el comportamiento central de una población o la frecuencia de un evento.

Sin embargo, en muchos problemas también interesa estudiar la variabilidad de las observaciones.

Por ejemplo, una empresa puede necesitar comprobar que las dimensiones de sus productos no presentan una dispersión excesiva.

También podríamos investigar si la variabilidad de las profundidades de los sismos supera cierto valor de referencia.

El parámetro de interés es la varianza poblacional

$$
\sigma^2.
$$

## 6.1. Formulación de las hipótesis

Supongamos que queremos comparar la varianza poblacional con un valor $\sigma_0^2$.

Podemos formular una prueba bilateral:

$$
H_0:\sigma^2=\sigma_0^2,
\qquad
H_1:\sigma^2\neq\sigma_0^2.
$$

También podemos investigar si la variabilidad es excesiva:

$$
H_0:\sigma^2\leq\sigma_0^2,
\qquad
H_1:\sigma^2>\sigma_0^2.
$$

O si la variabilidad es inferior al valor de referencia:

$$
H_0:\sigma^2\geq\sigma_0^2,
\qquad
H_1:\sigma^2<\sigma_0^2.
$$

## 6.2. Estadístico de prueba

Supongamos que

$$
X_1,\ldots,X_n
$$

es una muestra aleatoria de una población normal con media $\mu$ y varianza $\sigma^2$.

Recordemos que

$$
S^2=\frac{1}{n-1}\sum_{i=1}^n(X_i-\overline X)^2.
$$

Bajo la hipótesis

$$
H_0:\sigma^2=\sigma_0^2,
$$

el estadístico

$$
\boxed{
Q=\frac{(n-1)S^2}{\sigma_0^2}
}
$$

sigue una distribución ji-cuadrada con $n-1$ grados de libertad:

$$
\boxed{
Q\sim\chi^2_{n-1}.
}
$$

Esta distribución permite calcular el valor-p correspondiente.

A diferencia de la prueba para la media, **la prueba ji-cuadrada clásica para una varianza depende especialmente del supuesto de normalidad poblacional**.

No basta con tener una muestra grande para afirmar que el estadístico anterior sigue aproximadamente una distribución ji-cuadrada cuando la población es arbitraria.

## 6.3. Cálculo del valor-p

Sea $q_{\text{obs}}$ el valor observado del estadístico.

Para una prueba unilateral derecha,

$$
H_0:\sigma^2\leq\sigma_0^2,
\qquad
H_1:\sigma^2>\sigma_0^2,
$$

el valor-p es

$$
\boxed{
p=P(Q\geq q_{\text{obs}}).
}
$$

Para una prueba unilateral izquierda,

$$
H_0:\sigma^2\geq\sigma_0^2,
\qquad
H_1:\sigma^2<\sigma_0^2,
$$

el valor-p es

$$
\boxed{
p=P(Q\leq q_{\text{obs}}).
}
$$

Para una prueba bilateral, una convención habitual consiste en duplicar la menor probabilidad de cola:

$$
\boxed{
p=
\min\left\{
1,\,
2\min\left(
P(Q\leq q_{\text{obs}}),
P(Q\geq q_{\text{obs}})
\right)
\right\}.
}
$$

Esta es una prueba bilateral válida, aunque no es la única manera posible de construir un contraste bilateral para una distribución asimétrica.

## 6.4. Ejemplo: variabilidad en el llenado de envases

Una máquina llena envases con cierta cantidad de líquido.

El fabricante establece que la varianza del volumen de llenado no debe superar 4 unidades cuadradas.

Un inspector selecciona una muestra aleatoria de 20 envases y obtiene

$$
n=20,\qquad s^2=8.
$$

Supondremos que los volúmenes de llenado siguen una distribución normal.

Queremos investigar si la varianza poblacional supera el valor permitido.

Las hipótesis son

$$
H_0:\sigma^2\leq4,
\qquad
H_1:\sigma^2>4.
$$

El estadístico observado es

$$
q_{\text{obs}}
=
\frac{(20-1)(8)}{4}
=
38.
$$

Bajo el caso límite de la hipótesis nula,

$$
Q\sim\chi^2_{19}.
$$

Por lo tanto,

$$
p=P(\chi^2_{19}\geq38)\approx0.0059.
$$

Si utilizamos

$$
\alpha=0.05,
$$

tenemos

$$
p<\alpha.
$$

Por lo tanto, rechazamos $H_0$.

**Conclusión:** existe evidencia estadísticamente significativa de que la varianza poblacional del volumen de llenado supera el valor permitido.

# 7. La interpretación del valor-p

Aunque hemos estudiado diferentes parámetros poblacionales, todas las pruebas comparten una misma lógica.

Primero formulamos una hipótesis nula y una alternativa.

Después calculamos un estadístico de prueba cuya distribución conocemos, exactamente o de manera aproximada, bajo la hipótesis nula.

Finalmente, utilizamos la distribución del estadístico para calcular el valor-p.

El valor-p representa qué tan probable sería obtener un resultado igual o más desfavorable para la hipótesis nula que el observado, suponiendo que esta fuera verdadera.

La regla de decisión es

$$
\boxed{
p<\alpha
\quad\Longrightarrow\quad
\text{Rechazar }H_0.
}
$$

En caso contrario,

$$
\boxed{
p\geq\alpha
\quad\Longrightarrow\quad
\text{No rechazar }H_0.
}
$$

Sin embargo, debemos recordar que el valor-p no representa la probabilidad de que la hipótesis nula sea verdadera.

Tampoco mide directamente la magnitud o la importancia práctica de un efecto.

Por ejemplo, una diferencia muy pequeña entre una media muestral y un valor de referencia puede resultar estadísticamente significativa cuando el tamaño de muestra es suficientemente grande.

Por ello, una conclusión estadística debe considerar tanto la significancia como la magnitud del efecto y su contexto científico.

## 8. Relación con los intervalos de confianza

En el capítulo anterior estudiamos cómo construir intervalos de confianza para parámetros poblacionales.

Ahora podemos establecer una relación entre esos intervalos y las pruebas de hipótesis.

Consideremos una prueba bilateral:

$$
H_0:\theta=\theta_0,
\qquad
H_1:\theta\neq\theta_0.
$$

Supongamos que construimos un intervalo de confianza bilateral de nivel

$$
1-\alpha.
$$

Cuando el intervalo y la prueba se construyen mediante procedimientos compatibles, se cumple que:

- Si $\theta_0$ pertenece al intervalo de confianza, no rechazamos $H_0$.
- Si $\theta_0$ no pertenece al intervalo de confianza, rechazamos $H_0$.

Por ejemplo, supongamos que obtenemos un intervalo de confianza del $95\%$ para una media poblacional:

$$
IC_{95\%}(\mu)=(12.4,15.8).
$$

Si queremos contrastar

$$
H_0:\mu=14,
$$

observamos que

$$
14\in(12.4,15.8).
$$

Por lo tanto, no rechazamos $H_0$ al nivel de significancia $\alpha=0.05$, utilizando la prueba bilateral correspondiente.

En cambio, si queremos contrastar

$$
H_0:\mu=12,
$$

observamos que

$$
12\notin(12.4,15.8).
$$

Por lo tanto, rechazamos $H_0$.

Esta relación muestra que los intervalos de confianza y las pruebas de hipótesis son dos herramientas estrechamente relacionadas.

No obstante, la equivalencia requiere utilizar procedimientos compatibles. No debe suponerse automáticamente para cualquier combinación de intervalo y prueba, especialmente cuando se utilizan métodos exactos para distribuciones discretas.

## 9. Consideraciones finales

En este capítulo estudiamos tres problemas fundamentales de inferencia estadística.

La prueba para una media poblacional permite contrastar afirmaciones sobre el valor esperado de una variable cuantitativa.

La prueba para una proporción poblacional permite estudiar la frecuencia relativa de un evento de interés. Cuando las observaciones siguen un modelo binomial, podemos calcular el valor-p mediante una prueba exacta.

Finalmente, la prueba para una varianza poblacional permite investigar la dispersión de una variable, utilizando la distribución ji-cuadrada cuando se cumplen los supuestos correspondientes.

Aunque los estadísticos y sus distribuciones cambian de una prueba a otra, la lógica general permanece:

$$
\boxed{
\text{Pregunta científica}
\longrightarrow
\text{Hipótesis}
\longrightarrow
\text{Estadístico}
\longrightarrow
\text{Valor-p}
\longrightarrow
\text{Decisión}.
}
$$

La prueba de hipótesis no determina si una afirmación es verdadera o falsa con absoluta certeza.

Su propósito consiste en evaluar si los datos proporcionan evidencia estadística suficiente para rechazar una hipótesis, utilizando un procedimiento cuyo comportamiento probabilístico ha sido establecido previamente.

Por ello, la conclusión de una prueba debe expresarse siempre en términos de evidencia estadística, evitando interpretar el rechazo o la falta de rechazo como una demostración definitiva.
