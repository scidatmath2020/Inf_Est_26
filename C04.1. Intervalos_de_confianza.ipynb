# Intervalos de confianza

## De la estimación puntual a la estimación por intervalos

En la clase anterior estudiamos cómo utilizar una muestra aleatoria para estimar parámetros desconocidos de una población. En particular, abordamos la estimación puntual y el método de máxima verosimilitud.

Recordemos que, si $X_1, X_2, \dots, X_n$ es una muestra aleatoria de una población cuya distribución depende de un parámetro desconocido $\theta$, nuestro objetivo es construir un estimador

$$ \hat{\theta} = T(X_1, X_2, \dots, X_n) $$

que nos permita aproximar el verdadero valor de $\theta$.

Por ejemplo, si deseamos estimar la media poblacional $\mu$, podemos utilizar la media muestral:

$$ \overline{X} = \frac{1}{n} \sum_{i=1}^{n} X_i. $$

Una vez obtenida la muestra, calculamos un valor numérico $\overline{x}$ que constituye nuestra estimación puntual.

### 1.1. El problema de la estimación puntual

Imaginemos que deseamos conocer el tiempo promedio que tardan los estudiantes de una universidad en trasladarse de sus casas al campus. Como sería complicado entrevistar a todos los estudiantes, seleccionamos aleatoriamente a 100 de ellos y registramos sus tiempos de traslado.

Supongamos que obtenemos una media muestral de 35 minutos. Nuestra estimación puntual es entonces:

$$ \hat{\mu} = \overline{x} = 35. $$

Pero debemos tener cuidado con la interpretación de este resultado. **¿Podemos afirmar que el verdadero tiempo promedio de traslado de todos los estudiantes es exactamente 35 minutos?** Por supuesto que no.

Si seleccionáramos otra muestra de 100 estudiantes, probablemente obtendríamos una media diferente. Quizá 33.8 minutos, 36.2 minutos o 34.5 minutos. 

Esto ocurre porque la media muestral es una variable aleatoria: su valor depende de los individuos seleccionados. Por tanto, aunque nuestro estimador sea adecuado, siempre existirá cierta incertidumbre en la estimación.

Luego, surge una pregunta: **En lugar de proporcionar únicamente un número, ¿podríamos construir un intervalo de valores que, con un nivel de confianza determinado, contenga al verdadero parámetro poblacional?**

Esta es precisamente la idea de la estimación por intervalos.

## La distribución muestral de la media

Para construir intervalos de confianza, primero necesitamos comprender cómo se comporta la media muestral cuando repetimos el proceso de muestreo.

Supongamos que

$$ X_1, X_2, \dots, X_n $$

son variables aleatorias independientes e idénticamente distribuidas, con

$$ E(X_i) = \mu, \qquad \text{Var}(X_i) = \sigma^2. $$

Consideremos la media muestral

$$ \overline{X} = \frac{X_1 + \cdots + X_n}{n}. $$

### Esperanza de la media muestral

Utilizando la linealidad de la esperanza:

$$
\begin{aligned}
E(\overline{X}) &= E\left( \frac{1}{n} \sum_{i=1}^{n} X_i \right) \\
&= \frac{1}{n} \sum_{i=1}^{n} E(X_i) \\
&= \frac{1}{n} (n\mu) \\
&= \mu.
\end{aligned}
$$

Por tanto,

$$ E(\overline{X}) = \mu. $$

Esto significa que la media muestral es un estimador insesgado de la media poblacional. En cristiano: **Si tomaramos muchas muestras del mismo tamaño y en cada una calculáramos la media, y luego promediamos todas esas medias, este último promedio se aproximaría a $\mu$ conforme aumentara el número de muestras**. Sin embargo, esto no significa que cada media muestral sea cercana a $\mu$. Algunas pueden encontrarse por encima y otras por debajo. ¿Qué tanto pueden alejarse? Para responder esta pregunta necesitamos estudiar la varianza de la media muestral.

### Varianza de la media muestral

Por independencia de las observaciones:

$$
\begin{aligned}
\text{Var}(\overline{X}) &= \text{Var}\left( \frac{1}{n} \sum_{i=1}^{n} X_i \right) \\
&= \frac{1}{n^2} \sum_{i=1}^{n} \text{Var}(X_i) \\
&= \frac{n\sigma^2}{n^2} \\
&= \frac{\sigma^2}{n}.
\end{aligned}
$$

Consecuentemente,

$$ \text{Var}(\overline{X}) = \frac{\sigma^2}{n}. $$

Y su desviación estándar es

$$ \sigma_{\overline{X}} = \frac{\sigma}{\sqrt{n}}. $$

Esta última cantidad recibe el nombre de error estándar de la media muestral.

### Interpretación del error estándar

Conviene distinguir dos conceptos que frecuentemente se confunden.

La desviación estándar poblacional $\sigma$ mide la dispersión de las observaciones individuales alrededor de $\mu$.

En cambio, el error estándar $\sigma / \sqrt{n}$ mide la dispersión de las medias muestrales alrededor de $\mu$.

Por ejemplo, si

$$ \sigma = 12, $$

tenemos:

| Tamaño de muestra $n$ | Error estándar $\sigma / \sqrt{n}$ |
| :---: | :---: |
| 4 | 6 |
| 16 | 3 |
| 36 | 2 |
| 100 | 1.2 |
| 400 | 0.6 |

Observemos que, conforme aumenta el tamaño de la muestra, disminuye el error estándar.

Esto explica por qué las muestras grandes suelen proporcionar estimaciones más precisas.

Sin embargo, la disminución no es lineal. Para reducir el error estándar a la mitad, necesitamos multiplicar por cuatro el tamaño de la muestra.

## La distribución de la media muestral

Conocer la esperanza y la varianza de $\overline{X}$ es importante, pero todavía necesitamos algo más.

Para construir un intervalo de confianza, debemos conocer, al menos aproximadamente, la distribución de probabilidad de nuestro estimador.

Comenzaremos con un caso especialmente sencillo.

### Población con distribución gaussiana

Supongamos que las observaciones provienen de una distribución normal:

$$ X_1, \dots, X_n \overset{\text{iid}}{\sim} N(\mu, \sigma^2). $$

Sabemos que una combinación lineal de variables aleatorias normales independientes también tiene distribución normal.

Por tanto,

$$ \overline{X} \sim N\left( \mu, \frac{\sigma^2}{n} \right). $$

Este resultado es exacto para cualquier tamaño de muestra $n \geq 1$.

Ahora estandarizaremos la media muestral:

$$ Z = \frac{\overline{X} - \mu}{\sigma / \sqrt{n}}. $$

Entonces

$$ Z \sim N(0, 1). $$

Este es el resultado que nos permitirá construir nuestro primer intervalo de confianza.

### ¿Qué ocurre cuando la población no es normal?

Aquí entra en juego el Teorema del Límite Central.

Si $X_1, \dots, X_n$ son independientes e idénticamente distribuidas, con media finita $\mu$ y varianza finita y positiva $\sigma^2$, entonces

$$ \frac{\overline{X} - \mu}{\sigma / \sqrt{n}} \xrightarrow{d} N(0, 1) $$

cuando $n \to \infty$.

Es decir, para tamaños de muestra suficientemente grandes, la distribución de la media muestral puede aproximarse mediante una normal, aunque la distribución original de las observaciones no sea normal.

No existe un tamaño mínimo universal que garantice una buena aproximación: depende de la distribución poblacional.

## Construcción de nuestro primer intervalo de confianza

Supongamos que deseamos estimar la media poblacional $\mu$, bajo las siguientes condiciones:

- Disponemos de una muestra aleatoria de una población normal.
- La desviación estándar poblacional $\sigma$ es conocida.
- La media poblacional $\mu$ es desconocida.

Como demostramos anteriormente,

$$ Z = \frac{\overline{X} - \mu}{\sigma / \sqrt{n}} \sim N(0, 1). $$

Nuestro objetivo será encontrar dos cantidades, calculables a partir de la muestra, entre las cuales podamos encerrar a $\mu$ con una probabilidad determinada.

### La idea del 95 % central

Recordemos que la distribución normal estándar es simétrica respecto del cero.

Podemos encontrar un número positivo $c$ tal que

$$ P(-c \leq Z \leq c) = 0.95. $$

Para la normal estándar, dicho número es aproximadamente

$$ c = 1.96. $$

![imagenes](im022.png)

Por tanto,

$$ P(-1.96 \leq Z \leq 1.96) \approx 0.95. $$

Sustituyendo la definición de $Z$:

$$ P\left( -1.96 \leq \frac{\overline{X} - \mu}{\sigma / \sqrt{n}} \leq 1.96 \right) \approx 0.95. $$

Ahora realizaremos algunas manipulaciones algebraicas.

Multiplicamos las tres expresiones de la desigualdad por $\sigma / \sqrt{n}$, que es una cantidad positiva:

$$ P\left( -1.96 \frac{\sigma}{\sqrt{n}} \leq \overline{X} - \mu \leq 1.96 \frac{\sigma}{\sqrt{n}} \right) \approx 0.95. $$

Restando $\overline{X}$ y multiplicando por $-1$, invirtiendo el sentido de las desigualdades, obtenemos:

$$ P\left( \overline{X} - 1.96 \frac{\sigma}{\sqrt{n}} \leq \mu \leq \overline{X} + 1.96 \frac{\sigma}{\sqrt{n}} \right) \approx 0.95. $$

Hemos conseguido algo fundamental. Construimos dos variables aleatorias:

$$ L = \overline{X} - 1.96 \frac{\sigma}{\sqrt{n}} $$

y

$$ U = \overline{X} + 1.96 \frac{\sigma}{\sqrt{n}}, $$

tales que

$$ P(L \leq \mu \leq U) \approx 0.95. $$

El intervalo aleatorio

$$ (L, U) $$

es un intervalo de confianza del 95 % para la media poblacional.

Observemos que sus extremos son aleatorios porque dependen de $\overline{X}$, mientras que $\mu$ es un parámetro fijo.

## Ejemplo numérico

Regresemos al problema de los tiempos de traslado de los estudiantes. Supongamos que conocemos la desviación estándar poblacional:

$$ \sigma = 12 \text{ minutos}. $$

Seleccionamos una muestra aleatoria de

$$ n = 100 $$

estudiantes y obtenemos

$$ \overline{x} = 35 \text{ minutos}. $$

**Supondremos que los tiempos de traslado siguen una distribución normal.**

Nuestro intervalo de confianza del 95 % es

$$ \left( \overline{x} - 1.96 \frac{\sigma}{\sqrt{n}}, \overline{x} + 1.96 \frac{\sigma}{\sqrt{n}} \right). $$

Sustituyendo:

$$
\begin{aligned}
L &= 35 - 1.96 \frac{12}{10} = 32.648, \\
U &= 35 + 1.96 \frac{12}{10} = 37.352.
\end{aligned}
$$

Por tanto, el intervalo obtenido es

$$ (32.648, \ 37.352). $$

**Interpretación:** con un nivel de confianza del 95 %, estimamos que el verdadero tiempo promedio de traslado de los estudiantes se encuentra entre aproximadamente 32.65 y 37.35 minutos.

Notemos que no estamos afirmando que el promedio poblacional sea exactamente 35 minutos. Estamos proporcionando un rango de valores compatibles con la información muestral y con el procedimiento estadístico utilizado.

## ¿Qué significa realmente tener un 95 % de confianza?

Este es probablemente el punto conceptual más importante. Una interpretación muy frecuente es afirmar:

*«Existe un 95 % de probabilidad de que la verdadera media poblacional esté dentro del intervalo calculado».*

Sin embargo, desde el punto de vista de la estadística frecuentista, esa afirmación no es correcta. ¿Por qué?

Porque el parámetro poblacional $\mu$ es una cantidad fija, aunque desconocida. Una vez que observamos nuestra muestra y calculamos el intervalo, sus extremos también quedan determinados.

En ese momento, el intervalo contiene a $\mu$ o no lo contiene.

La probabilidad del 95 % describe el comportamiento del procedimiento de construcción del intervalo antes de observar la muestra.

### Interpretación mediante muestreos repetidos

Imaginemos el siguiente experimento:

1. Seleccionamos una muestra aleatoria de tamaño 100.
2. Calculamos su media muestral.
3. Construimos un intervalo de confianza del 95 %.
4. Repetimos el procedimiento muchas veces, siempre con muestras independientes de la misma población.

Cada muestra producirá un intervalo diferente.

Algunos intervalos contendrán a la verdadera media poblacional y otros no.

Si construimos, por ejemplo, 1,000 intervalos, esperamos que aproximadamente 950 contengan al verdadero valor de $\mu$.

No necesariamente serán exactamente 950, porque el propio número de intervalos que contienen al parámetro es aleatorio.

La siguiente visualización permite explorar esta idea.

**Simulación de 150 intervalos de confianza**

Cada línea horizontal representa un intervalo construido con una muestra distinta. La línea vertical marca la verdadera media poblacional $\mu$.

![imagenes](im021.png)

Observemos dos fenómenos:

**Primero:** el número de intervalos que contienen al parámetro puede variar entre experimentos, pero la proporción observada tiende a acercarse al nivel de confianza cuando repetimos el procedimiento muchas veces.

**Segundo:** al incrementar el nivel de confianza de 90 % a 95 % o 99 %, los intervalos se hacen más amplios.

Esto tiene sentido: si deseamos aumentar la frecuencia con la que nuestros intervalos contienen al verdadero parámetro, debemos aceptar un rango de valores más grande.

## Generalización: intervalos de confianza de nivel $1 - \alpha$

Hasta ahora construimos un intervalo con nivel de confianza del 95 %. Sin embargo, podemos generalizar el procedimiento.

Sea

$$ 0 < \alpha < 1. $$

Deseamos construir un intervalo de confianza con nivel

$$ 1 - \alpha. $$

Denotemos por $z_{1-\alpha/2}$ el cuantil de orden $1 - \alpha/2$ de la distribución normal estándar. Es decir,

$$ P(Z \leq z_{1-\alpha/2}) = 1 - \frac{\alpha}{2}. $$

Por simetría,

$$ P\left( -z_{1-\alpha/2} \leq Z \leq z_{1-\alpha/2} \right) = 1 - \alpha. $$

Repitiendo nuestro procedimiento algebraico obtenemos:

$$ P\left( \overline{X} - z_{1-\alpha/2} \frac{\sigma}{\sqrt{n}} \leq \mu \leq \overline{X} + z_{1-\alpha/2} \frac{\sigma}{\sqrt{n}} \right) = 1 - \alpha. $$

Por tanto, el intervalo de confianza para la media poblacional, con varianza conocida y bajo normalidad, es

$$ IC_{1-\alpha}(\mu) = \left[ \overline{X} - z_{1-\alpha/2} \frac{\sigma}{\sqrt{n}}, \ \overline{X} + z_{1-\alpha/2} \frac{\sigma}{\sqrt{n}} \right]. $$

Los valores críticos más utilizados son:

| Nivel de confianza | $\alpha$ | $z_{1-\alpha/2}$ |
| :---: | :---: | :---: |
| 90 % | 0.10 | 1.645 |
| 95 % | 0.05 | 1.960 |
| 99 % | 0.01 | 2.576 |

### Margen de error

Observemos que nuestro intervalo tiene la forma

$$ \overline{X} \pm E, $$

donde

$$ E = z_{1-\alpha/2} \frac{\sigma}{\sqrt{n}} $$

recibe el nombre de **margen de error**.

Esta cantidad expresa qué tan lejos se extiende el intervalo a cada lado de la estimación puntual.

Su amplitud total es

$$ A = 2E. $$

De esta expresión podemos obtener tres conclusiones:

- Si aumenta el nivel de confianza, aumenta el valor crítico y, por tanto, la amplitud del intervalo.
- Si aumenta la variabilidad poblacional $\sigma$, aumenta la amplitud.
- Si aumenta el tamaño de la muestra $n$, disminuye la amplitud.

Estas relaciones son fundamentales para el diseño de estudios estadísticos y para determinar tamaños de muestra.

## Ejercicio para consolidar

Una empresa produce envases cuyo contenido se distribuye normalmente. Se sabe que la desviación estándar poblacional es de 8 mililitros.

Se selecciona una muestra aleatoria de 64 envases y se obtiene un contenido promedio de 502 mililitros.

Construir e interpretar los intervalos de confianza para la media poblacional con niveles de 90 %, 95 % y 99 %.

Además, responder:

¿Cuál de los tres intervalos es más preciso y cuál ofrece un mayor nivel de confianza? ¿Es posible mejorar simultáneamente ambas características sin cambiar el tamaño de la muestra?

Este ejercicio permite apreciar que la precisión y la confianza son dos propiedades relacionadas, pero diferentes.
