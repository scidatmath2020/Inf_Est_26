# Regresión lineal

## 8.1. Introducción

Hasta ahora hemos estudiado herramientas estadísticas que permiten describir datos, estimar parámetros y comparar poblaciones. Por ejemplo, mediante ANOVA podemos investigar si la magnitud promedio de los sismos es la misma durante los doce meses del año.

Sin embargo, existen problemas en los que no solamente interesa comparar grupos, sino también **comprender cómo se relacionan distintas variables y utilizar esas relaciones para realizar predicciones**.

Supongamos que queremos estudiar el precio de una vivienda. Es razonable pensar que su precio depende de características como la superficie construida, el número de habitaciones, la antigüedad y la ubicación.

La regresión lineal es una herramienta estadística que permite estudiar este tipo de relaciones.

Su objetivo consiste en construir un modelo matemático que explique, al menos parcialmente, el comportamiento de una variable a partir de una o varias variables adicionales.

Conviene distinguir dos propósitos:

- **Explicación:** investigar cómo se relaciona una variable con otras.
- **Predicción:** estimar el valor de una variable para observaciones nuevas.

Aunque ambos propósitos utilizan modelos semejantes, no necesariamente conducen a las mismas decisiones estadísticas.

## Regresión lineal simple

Comencemos con un problema sencillo.

Supongamos que disponemos de información sobre las horas de estudio y las calificaciones de un grupo de estudiantes.

Es posible que los estudiantes que dedican más horas al estudio tiendan a obtener mejores calificaciones. No obstante, la relación no será necesariamente perfecta: dos estudiantes que estudian el mismo tiempo pueden obtener resultados diferentes.

La regresión lineal simple propone representar esta relación mediante una recta:

$$
\hat{y}=b_0+b_1x.
$$

En este modelo, $\hat{y}$ representa la calificación estimada, $x$ las horas de estudio, $b_0$ el intercepto y $b_1$ la pendiente.

Por ejemplo, si obtenemos

$$
\hat{y}=45+6x,
$$

podemos interpretar que, según el modelo, cada hora adicional de estudio se asocia con un incremento promedio de seis puntos en la calificación.

Para un estudiante que estudia cinco horas, la predicción sería

$$
\hat{y}=45+6(5)=75.
$$

Esto no significa que necesariamente obtendrá 75 puntos. El modelo solamente proporciona una estimación.

Tampoco podemos concluir, únicamente a partir de esta relación, que estudiar una hora adicional cause un incremento de seis puntos.

## ¿Cómo se obtiene la recta?

Una vez que tenemos los datos, necesitamos elegir la recta que mejor los represente.

Para ello utilizamos el método de **mínimos cuadrados ordinarios**.

La idea es sencilla: cada observación tiene un valor real $y_i$ y el modelo proporciona una predicción $\hat{y}_i$. La diferencia entre ambos se denomina residuo:

$$
e_i=y_i-\hat{y}_i.
$$

Buscamos una recta que haga pequeña la suma de los cuadrados de estas diferencias:

$$
\mathrm{SCE}=\sum_{i=1}^{n}(y_i-\hat{y}_i)^2.
$$

¿Por qué utilizamos cuadrados? Principalmente porque evitan que los errores positivos y negativos se cancelen y porque permiten obtener una solución matemática conveniente.

Sin embargo, debemos tener presente que los errores grandes reciben una penalización considerable. Por esta razón, los valores atípicos pueden afectar notablemente el ajuste.

## Regresión lineal múltiple

En muchos problemas, una sola variable resulta insuficiente para explicar el fenómeno estudiado.

Retomemos el ejemplo del precio de una vivienda. En lugar de utilizar solamente su superficie, podemos incorporar otras características.

El modelo de regresión lineal múltiple tiene la forma

$$
Y=\beta_0+\beta_1X_1+\cdots+\beta_pX_p+\varepsilon.
$$

Aquí, $Y$ representa la variable que queremos estudiar, $X_1,\ldots,X_p$ son las variables explicativas y $\varepsilon$ representa la parte que el modelo no consigue explicar.

Por ejemplo, podríamos obtener

$$
\widehat{\text{Precio}}
=250000+12000X_1+80000X_2-5000X_3,
$$

donde $X_1$ representa la superficie en metros cuadrados, $X_2$ el número de baños y $X_3$ la antigüedad en años.

La interpretación de los coeficientes requiere cuidado.

El coeficiente de la superficie indica que, **manteniendo constantes las demás variables**, un metro cuadrado adicional se asocia con un incremento estimado de $12,000 en el precio.

Esta condición es fundamental: en regresión múltiple, cada coeficiente describe una relación ajustada por las demás variables del modelo.

## Correlación entre variables

Antes de construir un modelo de regresión múltiple, conviene explorar las relaciones entre las variables.

Una herramienta conocida es el coeficiente de correlación de Pearson, que toma valores entre $-1$ y $1$.

Un valor cercano a $1$ indica una relación lineal positiva fuerte, mientras que uno cercano a $-1$ indica una relación lineal negativa fuerte. Los valores cercanos a cero indican una relación lineal débil, aunque pueden existir relaciones de otro tipo.

Supongamos que tenemos las siguientes variables:

- Superficie construida.
- Número de habitaciones.
- Número de baños.
- Antigüedad.
- Precio.

Es razonable encontrar una correlación positiva entre la superficie construida y el número de habitaciones.

Pero esto nos lleva a una pregunta importante: **¿qué sucede cuando dos variables explicativas contienen prácticamente la misma información?**

## Columnas altamente correlacionadas: multicolinealidad

La multicolinealidad aparece cuando una variable explicativa puede relacionarse linealmente de manera muy fuerte con una o varias de las demás variables explicativas.

Imaginemos que queremos predecir el precio de una vivienda utilizando dos columnas:

$$
X_1=\text{Superficie en metros cuadrados},
$$

$$
X_2=\text{Superficie en pies cuadrados}.
$$

Como ambas columnas representan exactamente la misma característica, tenemos

$$
X_2\approx10.764X_1.
$$

En este caso existe una dependencia lineal exacta si utilizamos el factor de conversión completo, salvo los posibles errores de redondeo.

Incluir ambas columnas no proporciona información nueva.

De hecho, cuando existe dependencia lineal exacta, los coeficientes del modelo no pueden determinarse de manera única mediante mínimos cuadrados ordinarios.

### ¿Por qué es un problema?

Consideremos ahora un caso menos extremo.

Supongamos que queremos explicar el precio de una vivienda utilizando la superficie construida y el número de habitaciones.

Ambas variables están relacionadas, pero no son idénticas.

Si su correlación es muy alta, puede resultar difícil distinguir qué parte del efecto estimado corresponde a la superficie y cuál al número de habitaciones.

Como consecuencia, pueden presentarse las siguientes situaciones:

1. Los coeficientes cambian considerablemente al modificar ligeramente los datos.
2. Los errores estándar de los coeficientes aumentan.
3. Algunas variables parecen no ser estadísticamente significativas, aunque el modelo completo explique una proporción importante de la variabilidad.
4. Los coeficientes pueden adquirir signos inesperados.

Es importante destacar que **la multicolinealidad no implica necesariamente malas predicciones**. Un modelo puede predecir razonablemente bien y, al mismo tiempo, tener coeficientes difíciles de interpretar.

### ¿Cómo identificarla?

Una primera aproximación consiste en construir una matriz de correlaciones entre las variables explicativas.

Por ejemplo:

| Variables | Correlación |
|---|---:|
| Superficie y habitaciones | 0.94 |
| Superficie y baños | 0.82 |
| Habitaciones y baños | 0.79 |
| Superficie y antigüedad | -0.12 |

En este ejemplo, la correlación de 0.94 merece una revisión especial.

No existe un valor universal que determine cuándo una correlación es demasiado alta. En la práctica, valores absolutos superiores a 0.8 o 0.9 suelen utilizarse como señales de advertencia, pero no constituyen reglas definitivas.

Además, una matriz de correlaciones puede pasar por alto problemas en los que una columna está relacionada con una combinación de varias columnas, aunque ninguna correlación por pares sea especialmente elevada.

### ¿Qué podemos hacer?

La solución depende del objetivo del análisis.

Si queremos interpretar los coeficientes, puede resultar conveniente eliminar alguna variable redundante, combinar variables relacionadas o replantear el modelo utilizando conocimiento del problema.

Si nuestro objetivo principal es predecir, podemos considerar técnicas como la regresión Ridge, que reduce la inestabilidad de los coeficientes, o la regresión Lasso, que también puede ayudar a seleccionar variables.

Otra alternativa es el análisis de componentes principales, que permite construir nuevas variables no correlacionadas a partir de las originales.

Sin embargo, no debemos eliminar columnas simplemente porque estén altamente correlacionadas. Antes debemos preguntarnos si representan conceptos diferentes y qué información perderíamos al retirar alguna.

## Supuestos de la regresión lineal

Al igual que ocurre con las pruebas de hipótesis y ANOVA, la regresión lineal requiere revisar ciertas condiciones para que sus conclusiones estadísticas sean confiables.

Es conveniente distinguir entre las condiciones necesarias para estimar un modelo y las condiciones adicionales utilizadas para realizar inferencia estadística.

### Linealidad

El modelo supone que la media de la variable respuesta puede expresarse como una combinación lineal de las variables explicativas.

Esto no significa que todas las relaciones deban representarse mediante rectas simples. Por ejemplo, podemos incorporar $X^2$ y continuar trabajando con un modelo lineal en sus coeficientes.

### Independencia

Los errores de las observaciones deben ser independientes cuando utilizamos los procedimientos inferenciales habituales que suponen esta condición.

En datos temporales, como las mediciones de temperatura registradas cada hora, esta hipótesis puede ser problemática.

Si las observaciones están relacionadas, los errores estándar y las pruebas estadísticas convencionales pueden resultar incorrectos.

### Homocedasticidad

La homocedasticidad significa que la variabilidad de los errores permanece constante para los distintos valores de las variables explicativas.

Si la dispersión de los residuos aumenta o disminuye sistemáticamente, hablamos de heterocedasticidad.

Esta situación no invalida automáticamente los coeficientes obtenidos por mínimos cuadrados, pero puede afectar la precisión de los errores estándar y de los intervalos de confianza habituales.

### Gaussianidad de los errores

Para realizar ciertas pruebas de hipótesis e intervalos de confianza exactos en muestras pequeñas, el modelo clásico supone que los errores siguen una distribución gaussiana.

Es importante aclarar que **no necesitamos que las variables explicativas ni la variable respuesta sean gaussianas**.

La hipótesis se refiere a los errores del modelo, condicionados a los valores de las variables explicativas.

Tampoco es indispensable la gaussianidad para calcular los coeficientes mediante mínimos cuadrados. En muestras grandes, bajo condiciones apropiadas, podemos utilizar aproximaciones asintóticas.

### Ausencia de multicolinealidad perfecta

Las variables explicativas no deben presentar dependencias lineales exactas si queremos determinar de manera única todos los coeficientes del modelo.

La multicolinealidad elevada, aunque no sea perfecta, merece atención especial por sus efectos sobre la estabilidad de las estimaciones.

## Análisis de residuos

Una vez ajustado el modelo, debemos investigar qué tan razonables son sus supuestos.

Los residuos constituyen una herramienta fundamental para realizar este diagnóstico.

Un buen análisis de residuos debe investigar principalmente si aparecen patrones sistemáticos, si la dispersión cambia con los valores ajustados, si existen observaciones atípicas y si resulta razonable suponer gaussianidad cuando sea necesaria para la inferencia.

Para estudiar estas cuestiones podemos utilizar gráficos de residuos, histogramas, gráficos Q-Q y pruebas estadísticas.

Entre las herramientas más conocidas se encuentran:

- **Shapiro-Wilk:** para estudiar gaussianidad.
- **Breusch-Pagan:** para detectar heterocedasticidad.
- **Durbin-Watson:** para investigar autocorrelación de primer orden.

No obstante, debemos evitar utilizar estas pruebas de manera mecánica. Sus resultados deben interpretarse junto con los gráficos, el tamaño de muestra y el contexto del problema.

## ¿Cómo evaluamos la calidad del modelo?

Existen diferentes maneras de evaluar un modelo de regresión lineal.

### Coeficiente de determinación

El coeficiente $R^2$ mide la proporción de la variabilidad observada en la respuesta que explica el modelo, tomando como referencia la predicción mediante la media muestral.

En un modelo habitual con intercepto,

$$
R^2=1-\frac{\mathrm{SCE}}{\mathrm{SCT}}.
$$

Un valor de $R^2=0.85$ significa que el modelo explica el 85 % de la variabilidad observada de la respuesta en los datos evaluados.

Esto no significa que las predicciones tengan un 85 % de precisión ni que el modelo explique causalmente el fenómeno.

Además, al incorporar nuevas variables explicativas, el $R^2$ de entrenamiento no disminuye, aunque las nuevas variables aporten poca información útil.

Por esta razón también se utiliza el $R^2$ ajustado, que penaliza la incorporación de variables adicionales.

### Error absoluto medio (MAE)

El MAE representa el promedio de las diferencias absolutas entre los valores observados y los predichos.

$$
\mathrm{MAE}=\frac{1}{n}\sum_{i=1}^{n}|y_i-\hat{y}_i|.
$$

Si estamos prediciendo precios de viviendas y obtenemos un MAE de $80,000, significa que las predicciones se equivocan, en promedio, en $80,000 en términos absolutos.

### Raíz del error cuadrático medio (RMSE)

El RMSE se define mediante

$$
\mathrm{RMSE}=
\sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2}.
$$

A diferencia del MAE, el RMSE penaliza con mayor intensidad los errores grandes.

Ninguna de estas métricas es universalmente superior. La elección depende del problema y de qué tipo de errores consideramos más importantes.

## Significancia estadística de los coeficientes

Una vez construido el modelo, podemos preguntarnos si existe evidencia estadística de que una variable explicativa está asociada con la respuesta, después de considerar las demás variables.

Para ello podemos plantear la hipótesis

$$
H_0:\beta_j=0
$$

frente a

$$
H_1:\beta_j\neq0.
$$

Habitualmente se utiliza una prueba basada en la distribución $t$ de Student, bajo los supuestos correspondientes.

Si obtenemos un p-valor pequeño, tenemos evidencia contra la hipótesis de que el coeficiente es cero.

Sin embargo, debemos distinguir entre significancia estadística e importancia práctica.

Un coeficiente puede ser estadísticamente significativo y representar un efecto demasiado pequeño para tener relevancia en el problema.

También puede ocurrir que una variable importante presente un p-valor grande debido a una muestra pequeña, errores muy variables o multicolinealidad elevada.

Además de las pruebas individuales, existe una prueba $F$ para investigar si el conjunto de variables explicativas aporta información al modelo frente a un modelo que solamente utiliza el intercepto.

Esta prueba guarda una relación estrecha con ANOVA, pues ambas metodologías comparan la variabilidad explicada con la variabilidad residual.

## Predicción y sobreajuste

Un modelo puede describir muy bien los datos utilizados para construirlo y, sin embargo, realizar malas predicciones sobre observaciones nuevas.

Este fenómeno se conoce como **sobreajuste**.

Por ejemplo, imaginemos que tenemos información sobre 100 viviendas y construimos un modelo utilizando 70 variables explicativas.

Aunque obtengamos un $R^2$ elevado, existe el riesgo de que el modelo esté capturando particularidades de esas 100 viviendas en lugar de relaciones que puedan generalizarse.

Una estrategia habitual consiste en separar los datos en dos conjuntos: entrenamiento y prueba.

El conjunto de entrenamiento se utiliza para construir el modelo. El conjunto de prueba se reserva para evaluar qué tan bien predice observaciones que no participaron en el ajuste.

Si necesitamos comparar distintos modelos o elegir variables, conviene utilizar además un conjunto de validación o validación cruzada, manteniendo los datos de prueba separados hasta la evaluación final.

En problemas temporales, esta separación debe respetar el orden cronológico cuando el objetivo sea predecir el futuro.

## Procedimiento general para construir un modelo

En la práctica, conviene seguir una secuencia ordenada que evite tomar decisiones únicamente a partir de los p-valores.

**Paso 1. Comprender el problema.**

Identificar qué variable queremos explicar o predecir y cuáles podrían estar relacionadas con ella.

**Paso 2. Explorar los datos.**

Revisar histogramas, diagramas de dispersión, valores faltantes y posibles observaciones atípicas.

**Paso 3. Estudiar las relaciones entre columnas.**

Construir una matriz de correlaciones y revisar la posible multicolinealidad.

**Paso 4. Construir un primer modelo.**

Ajustar una regresión lineal con las variables seleccionadas.

**Paso 5. Analizar los residuos.**

Investigar linealidad, independencia, homocedasticidad y gaussianidad cuando corresponda.

**Paso 6. Evaluar los resultados.**

Interpretar los coeficientes, revisar las métricas y estudiar la incertidumbre de las estimaciones.

**Paso 7. Validar el modelo.**

Evaluar su comportamiento sobre datos no utilizados para ajustarlo y considerar modificaciones si son necesarias.

Este procedimiento no debe entenderse como una receta inflexible. En ocasiones será necesario regresar a etapas anteriores, modificar las variables o incluso considerar otro tipo de modelo.
