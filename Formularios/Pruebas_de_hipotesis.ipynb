# Pruebas de hipótesis: aplicaciones

En este apartado aplicaremos las pruebas de hipótesis para la media, la proporción y la desviación estándar poblacionales. En cada problema identificaremos el parámetro de interés, formularemos las hipótesis, seleccionaremos la prueba adecuada y obtendremos el valor-p y, cuando corresponda, un intervalo de confianza.

Fijado un nivel de significancia $\alpha$, nuestra regla de decisión será:

- Si el valor-p es menor o igual que $\alpha$, **rechazamos $H_0$** a favor de $H_1$.
- Si el valor-p es mayor que $\alpha$, **no rechazamos $H_0$**: la muestra no proporciona evidencia suficiente a favor de $H_1$ al nivel elegido.

No rechazar $H_0$ **no significa** demostrar que es verdadera, y mucho menos rechazar $H_1$. Para las pruebas bilaterales, el contraste y el intervalo de confianza bilateral del nivel $1-\alpha$ conducen a la misma decisión. En pruebas unilaterales debemos utilizar el intervalo unilateral correspondiente. En particular, el valor propuesto bajo $H_0$ se contrasta con el intervalo: si queda **fuera**, se rechaza $H_0$; si queda **dentro**, no se rechaza.

El nivel de significancia debe elegirse antes de observar los datos. Si no está especificado, debemos fijarlo explícitamente según el contexto; el valor-p, por sí solo, no determina una decisión universal.

## Resumen de pruebas

| Parámetro | Supuestos | Estadístico bajo el valor de frontera de $H_0$ | Procedimiento |
|:--|:--|:--|:--|
| Media $\mu$, $\sigma$ conocida | Población normal, o aproximación normal justificada para $\overline X$ | $Z=\dfrac{\overline X-\mu_0}{\sigma/\sqrt n}$ | Prueba Z |
| Media $\mu$, $\sigma$ desconocida | Población normal; con muestras grandes, aproximación apropiada | $T=\dfrac{\overline X-\mu_0}{S/\sqrt n}$ | Prueba t con $n-1$ grados de libertad |
| Proporción $p$ | $n$ ensayos Bernoulli independientes con una probabilidad común | $X\sim\operatorname{Bin}(n,p_0)$ | Prueba binomial exacta |
| Varianza $\sigma^2$ o desviación $\sigma$ | Población normal | $Q=\dfrac{(n-1)S^2}{\sigma_0^2}\sim\chi^2_{n-1}$ | Prueba ji-cuadrada |

Las hipótesis alternativas pueden ser $\theta>\theta_0$, $\theta<\theta_0$ o $\theta\ne\theta_0$. Las colas de la distribución del estadístico se eligen de acuerdo con dicha alternativa. Para la prueba binomial bilateral exacta, la definición del valor-p utilizada por `binom.test` suma probabilidades de resultados al menos tan incompatibles con $H_0$ como el observado; **no siempre coincide** con el doble de la menor cola.

En las pruebas sobre la media, con una alternativa $\mu>\mu_0$, el intervalo unilateral inferior tiene la forma $(L,\infty)$; con $\mu<\mu_0$, el intervalo unilateral superior es $(-\infty,U)$. Los límites se calculan con $\overline X$, nunca con el estadístico $T$.

## Pruebas de hipótesis en R

En R podemos utilizar `z.test` del paquete `BSDA` cuando la desviación poblacional es conocida; `t.test` de R base cuando es desconocida; `binom.test` de R base para proporciones; y `sigma.test` del paquete `TeachingDemos` para la desviación estándar de una población normal.

| Función | Paquete |
|:--|:--|
| `z.test(x, alternative, mu, sigma.x, conf.level)` | `BSDA` |
| `t.test(x, alternative, mu, conf.level)` | `stats` (R base) |
| `binom.test(x, n, p, alternative, conf.level)` | `stats` (R base) |
| `sigma.test(x, sigma, alternative, conf.level)` | `TeachingDemos` |

En estas funciones, `x` representa los datos observados, excepto en `binom.test`, donde es el número de éxitos. Los argumentos `mu`, `p` y `sigma` expresan, respectivamente, los valores propuestos $\mu_0$, $p_0$ y $\sigma_0$; `sigma.x` es la desviación poblacional conocida; `n` es el número de ensayos; `conf.level = 1 - alpha` indica el nivel de confianza. Finalmente, `alternative` toma los valores `"greater"`, `"less"` o `"two.sided"` según $H_1$.

Antes de ejecutar los ejemplos que emplean paquetes externos, deben estar instalados y cargados `BSDA` y `TeachingDemos`.

## Ejemplo 1. Rendimiento de un nutriente agrícola

Un productor obtiene habitualmente un rendimiento medio de **2000 kg/ha**. Se le ofrece un nutriente que, según el proveedor, permite superar ese rendimiento. Se sabe que los rendimientos individuales siguen aproximadamente una distribución normal y que su desviación estándar poblacional es **210 kg/ha**.

El productor selecciona aleatoriamente 9 parcelas tratadas con el nutriente y registra los rendimientos (en kg/ha):

$2150,\ 1950,\ 2170,\ 1860,\ 2050,\ 2120,\ 1920,\ 1850,\ 2230.$

**Pregunta:** ¿Existe evidencia estadísticamente significativa, al nivel $\alpha=0.05$, de que el rendimiento medio con el nutriente supera los 2000 kg/ha?

### Planteamiento

Sea $\mu$ el rendimiento medio poblacional con el nutriente. Consideramos

$$
H_0:\mu\le 2000,\qquad H_1:\mu>2000.
$$

Como la desviación estándar poblacional es conocida y se supone normalidad, corresponde una **prueba Z unilateral derecha**. Para el cálculo del valor-p se utiliza el valor de frontera $\mu_0=2000$.

```r
library(BSDA)

muestra <- c(2150, 1950, 2170, 1860, 2050, 2120, 1920, 1850, 2230)
mu0 <- 2000
sigma <- 210
alpha <- 0.05

z.test(muestra, alternative = "greater", mu = mu0,
       sigma.x = sigma, conf.level = 1 - alpha)
```

### Resultados e interpretación

La media muestral es $\overline x=2033.33$ kg/ha, el estadístico observado es $z\approx0.476$ y el **valor-p es aproximadamente 0.317**. El intervalo unilateral inferior al 95% es $(1918.19,\infty)$ kg/ha.

Como $0.317>0.05$, **no rechazamos $H_0$**. Aunque la media muestral supera los 2000 kg/ha, la diferencia observada no es estadísticamente significativa al nivel elegido. Esto no prueba que el nutriente carezca de efecto.

## Ejemplo 2. Estatura media de una población

Se desea estudiar si la estatura media de los hombres de 18 años o más de cierto país es **menor que 180 cm**. La desviación estándar poblacional es desconocida. Se seleccionan aleatoriamente 15 hombres y se obtienen las siguientes estaturas (en cm):

$168,\ 168,\ 169,\ 167,\ 171,\ 172,\ 182,\ 175,\ 177,\ 173,\ 168,\ 195,\ 167,\ 175,\ 175.$

Supondremos que la distribución poblacional de las estaturas es aproximadamente normal, una condición importante dado el tamaño reducido de la muestra.

**Pregunta:** ¿Hay evidencia al nivel $\alpha=0.05$ de que la estatura media poblacional es menor que 180 cm?

### Planteamiento

Sea $\mu$ la estatura media poblacional. Contrastamos

$$
H_0:\mu\ge180,\qquad H_1:\mu<180.
$$

Como $\sigma$ es desconocida, utilizamos una **prueba t unilateral izquierda**, con 14 grados de libertad y el valor de frontera $\mu_0=180$.

```r
muestra <- c(168, 168, 169, 167, 171, 172, 182, 175,
             177, 173, 168, 195, 167, 175, 175)
mu0 <- 180
alpha <- 0.05

t.test(muestra, alternative = "less", mu = mu0,
       conf.level = 1 - alpha)
```

### Resultados e interpretación

Se obtiene $\overline x\approx173.47$ cm, $s\approx7.36$ cm, un estadístico $t\approx-3.439$ y un **valor-p aproximado de 0.00199**. El intervalo unilateral superior al 95% es $(-\infty,176.81)$ cm.

Dado que $0.00199<0.05$, **rechazamos $H_0$**. La muestra aporta evidencia estadísticamente significativa de que la estatura media es inferior a 180 cm. El resultado no permite asegurar que todas las estaturas sean menores que 180 cm ni que la media sea inferior a 170 cm.

## Ejemplo 3. Proporción de universitarios que trabajan

Un estudio sostiene que **al menos 3 de cada 10** estudiantes universitarios trabajan; es decir, que la proporción poblacional satisface $p\ge0.30$. Para examinar la afirmación, se seleccionan aleatoriamente 600 estudiantes y se encuentra que **200 trabajan**.

**Pregunta:** ¿Ofrecen los datos evidencia para cuestionar la afirmación de que al menos el 30% de los estudiantes universitarios trabajan? Utiliza $\alpha=0.025$.

### Planteamiento

Sea $p$ la proporción poblacional de universitarios que trabajan. Para buscar evidencia **en contra** de la afirmación original, consideramos

$$
H_0:p\ge0.30,\qquad H_1:p<0.30.
$$

Usamos una **prueba binomial exacta unilateral izquierda**, evaluando la distribución binomial bajo el valor de frontera $p_0=0.30$. Asumimos observaciones independientes con una probabilidad común de éxito.

```r
p0 <- 0.30
exitos <- 200
tamano_muestra <- 600
alpha <- 0.025

binom.test(x = exitos, n = tamano_muestra, p = p0,
           alternative = "less", conf.level = 1 - alpha)
```

### Resultados e interpretación

La proporción muestral es $\widehat p=200/600\approx0.3333$, superior a $0.30$. El **valor-p unilateral izquierdo es aproximadamente 0.9651**.

Como $0.9651>0.025$, **no rechazamos $H_0$**: la muestra no proporciona evidencia de que la proporción poblacional sea menor del 30%. Esto no equivale a haber demostrado que la afirmación $p\ge0.30$ sea verdadera. Es importante distinguir este contraste de uno con alternativa $p>0.30$, que respondería a una pregunta diferente.

## Ejemplo 4. Variabilidad en la duración de baterías

Un fabricante afirma que la **desviación estándar** de la vida útil de sus baterías es **mayor que 0.8 años**. Se sabe que la duración de las baterías sigue una distribución normal. Para examinar la afirmación se seleccionan aleatoriamente 8 baterías y se registran estas duraciones (en años):

$2.32,\ 4.26,\ 4.02,\ 4.44,\ 3.68,\ 2.72,\ 1.90,\ 1.21.$

**Pregunta:** ¿Existe evidencia estadísticamente significativa, al nivel $\alpha=0.05$, de que la desviación estándar poblacional excede 0.8 años?

### Planteamiento

Sea $\sigma$ la desviación estándar poblacional de la duración. Formulamos

$$
H_0:\sigma\le0.8,\qquad H_1:\sigma>0.8.
$$

Como la población es normal y se estudia su dispersión, utilizamos la **prueba ji-cuadrada unilateral derecha** bajo el valor de frontera $\sigma_0=0.8$; el estadístico tiene 7 grados de libertad.

```r
library(TeachingDemos)

muestra <- c(2.32, 4.26, 4.02, 4.44,
             3.68, 2.72, 1.90, 1.21)
sigma0 <- 0.8
alpha <- 0.05

sigma.test(x = muestra, sigma = sigma0,
           alternative = "greater", conf.level = 1 - alpha)
```

### Resultados e interpretación

La desviación estándar muestral es $s\approx1.2001$ años, por lo que $s^2\approx1.4402$ años$^2$. El estadístico de contraste es

$$
q=\frac{(8-1)s^2}{0.8^2}\approx15.752,
$$

con un **valor-p aproximado de 0.0275**. El límite inferior unilateral al 95% para la desviación estándar poblacional es aproximadamente $0.8465$ años, y el intervalo es $(0.8465,\infty)$.

Como $0.0275<0.05$, **rechazamos $H_0$**. Los datos proporcionan evidencia de que la desviación estándar poblacional supera 0.8 años. Esta conclusión se refiere exclusivamente a la **variabilidad**, no a que la duración media de las baterías sea de 2.5 años.
