# C06. Pruebas de gaussianidad

## 1. Introducción

En los capítulos anteriores estudiamos procedimientos de inferencia estadística que permiten obtener información sobre una población a partir de una muestra aleatoria. En particular, construimos intervalos de confianza y realizamos pruebas de hipótesis sobre parámetros poblacionales desconocidos.

Algunos de estos procedimientos requieren que las observaciones provengan de una población con distribución gaussiana. Por ejemplo, la prueba t de Student para una media poblacional tiene una distribución de referencia exacta cuando las observaciones son independientes y proceden de una población gaussiana.

Sin embargo, en las aplicaciones reales generalmente desconocemos la distribución de la población. Podemos disponer de una muestra de estaturas, temperaturas, ingresos o tiempos de espera, pero esto no significa que conozcamos la distribución probabilística que generó esas observaciones.

Surge entonces una pregunta fundamental:

**¿Cómo podemos determinar si una muestra es compatible con una población que tiene distribución gaussiana?**

Una primera posibilidad consiste en representar los datos mediante un histograma y observar si presentan una forma aproximadamente simétrica y acampanada. Sin embargo, este procedimiento tiene limitaciones importantes, especialmente cuando el tamaño de la muestra es pequeño.

Una muestra de ocho observaciones, por ejemplo, difícilmente producirá un histograma que reproduzca claramente la forma de una campana de Gauss, incluso cuando las observaciones procedan de una población perfectamente gaussiana.

Por esta razón, necesitamos procedimientos estadísticos que permitan evaluar formalmente la compatibilidad de los datos con una distribución gaussiana.

Estos procedimientos se conocen como **pruebas de gaussianidad** o **pruebas de normalidad**.

En este capítulo estudiaremos sus fundamentos matemáticos, sus principales aplicaciones y las dificultades que aparecen cuando trabajamos con muestras pequeñas.

## 2. Planteamiento de las hipótesis

Sea $X_1,X_2,\ldots,X_n$ una muestra aleatoria de una población con función de distribución desconocida $F$.

Deseamos determinar si existe una media $\mu\in\mathbb{R}$ y una varianza $\sigma^2>0$ tales que la población tenga distribución gaussiana.

Formalmente, planteamos

$$
H_0:X\sim N(\mu,\sigma^2)
$$

frente a

$$
H_1:X\not\sim N(\mu,\sigma^2).
$$

La hipótesis nula establece que la distribución poblacional pertenece a la familia de distribuciones gaussianas, mientras que la alternativa establece que no pertenece a dicha familia.

Es importante observar que no estamos contrastando un valor específico de la media o de la varianza. En general, ambos parámetros son desconocidos.

Para realizar la prueba seleccionamos un nivel de significancia $\alpha$, calculamos un estadístico de prueba y obtenemos su correspondiente p-valor.

La regla de decisión es

$$
\boxed{
\begin{cases}
p<\alpha & \Rightarrow \text{Rechazar }H_0,\\
p\geq\alpha & \Rightarrow \text{No rechazar }H_0.
\end{cases}
}
$$

Si rechazamos $H_0$, concluimos que existe evidencia estadísticamente significativa contra la gaussianidad de la población.

Si no rechazamos $H_0$, concluimos que los datos no proporcionan evidencia suficiente para descartar una distribución gaussiana.

Sin embargo, no rechazar $H_0$ no demuestra que la población sea gaussiana.

Esta distinción será particularmente importante cuando trabajemos con muestras pequeñas.

## 3. El problema de las muestras pequeñas

Supongamos que deseamos investigar la distribución de una característica poblacional y disponemos únicamente de ocho observaciones:

$$
n=8.
$$

Aunque la población tenga distribución gaussiana, las observaciones pueden presentar una configuración aparentemente irregular.

Por ejemplo, algunas muestras pueden contener valores concentrados en un extremo, mientras que otras pueden presentar separaciones considerables entre observaciones consecutivas.

Estas configuraciones no necesariamente contradicen la gaussianidad. Son consecuencias de la variabilidad muestral.

El problema consiste en que, con pocas observaciones, resulta difícil distinguir entre una irregularidad producida por el azar y una característica auténtica de la distribución poblacional.

Esta dificultad se relaciona directamente con la potencia estadística.

### 3.1. Error de tipo I y error de tipo II

Recordemos que existen dos errores posibles en una prueba de hipótesis.

El error de tipo I consiste en rechazar una hipótesis nula verdadera. En nuestro contexto, significa concluir que la población no es gaussiana cuando realmente sí lo es.

El error de tipo II consiste en no rechazar una hipótesis nula falsa. En nuestro contexto, significa no detectar que la población tiene una distribución diferente de la gaussiana.

El nivel de significancia $\alpha$ controla la probabilidad máxima de cometer un error de tipo I:

$$
\sup_{F\in H_0}P_F(\text{Rechazar }H_0)\leq\alpha.
$$

Por otra parte, para una distribución alternativa específica $F$, definimos

$$
\beta(F)=P_F(\text{No rechazar }H_0).
$$

La potencia estadística correspondiente es

$$
\boxed{
\pi(F)=1-\beta(F)
=P_F(\text{Rechazar }H_0).
}
$$

Por lo tanto, la potencia representa la probabilidad de detectar correctamente una desviación de la gaussianidad.

### 3.2. Relación entre potencia y tamaño muestral

La potencia depende de diversos factores, entre ellos el tamaño de la muestra, el nivel de significancia, la prueba utilizada y la distribución alternativa considerada.

En general, cuando mantenemos fija una distribución alternativa no gaussiana y el nivel de significancia, aumentar el tamaño muestral permite detectar con mayor facilidad las desviaciones de la gaussianidad.

Consideremos dos investigadores que estudian la misma población no gaussiana.

El primero dispone de una muestra de tamaño $n=8$, mientras que el segundo dispone de una muestra de tamaño $n=100$.

Ambos utilizan la misma prueba y establecen

$$
\alpha=0.05.
$$

El primer investigador podría obtener un p-valor grande y no rechazar la gaussianidad, simplemente porque su muestra contiene poca información para identificar la desviación existente.

El segundo investigador, al disponer de más observaciones, podría detectar esa misma desviación.

Sin embargo, esto no significa que las muestras grandes sean siempre preferibles para cualquier propósito.

Cuando el tamaño muestral es muy grande, las pruebas pueden detectar desviaciones extremadamente pequeñas que, aunque estadísticamente significativas, carezcan de importancia práctica.

Por ello, debemos distinguir entre significancia estadística e importancia práctica.

### 3.3. Interpretación de un p-valor grande

Supongamos que aplicamos una prueba de gaussianidad a una muestra de tamaño $n=8$ y obtenemos

$$
p=0.40.
$$

Como $p>0.05$, no rechazamos $H_0$.

Esto no significa que exista una probabilidad del 40% de que la población sea gaussiana.

Tampoco significa que hayamos demostrado la gaussianidad.

La conclusión correcta es que no encontramos evidencia suficiente para rechazar la hipótesis nula.

Dado que la muestra es pequeña, debemos considerar la posibilidad de que la prueba tenga poca potencia frente a ciertas distribuciones alternativas.

En cambio, si obtenemos

$$
p=0.01,
$$

rechazamos $H_0$ al nivel de significancia del 5%.

En este caso, incluso con una muestra pequeña, encontramos evidencia estadísticamente significativa contra la gaussianidad.

Por lo tanto, una muestra pequeña puede proporcionar evidencia suficiente para rechazar la gaussianidad, pero no rechazarla no garantiza que la población sea gaussiana.

## 4. La función de distribución empírica

Antes de estudiar las pruebas específicas, introduciremos una herramienta matemática fundamental: la función de distribución empírica.

Sea $X_1,\ldots,X_n$ una muestra aleatoria de una población con función de distribución $F$.

La función de distribución empírica se define mediante

$$
\boxed{
F_n(x)=\frac{1}{n}\sum_{i=1}^{n}I(X_i\leq x),
}
$$

donde $I(A)$ es la función indicadora del evento $A$:

$$
I(A)=
\begin{cases}
1 & \text{si ocurre }A,\\
0 & \text{en otro caso}.
\end{cases}
$$

Así, $F_n(x)$ representa la proporción de observaciones muestrales menores o iguales que $x$.

Por ejemplo, consideremos la muestra ordenada

$$
2,\ 3,\ 4,\ 6,\ 7.
$$

Como $n=5$, tenemos

$$
F_n(4)=\frac{3}{5}=0.60,
$$

pues tres de las cinco observaciones son menores o iguales que 4.

La función $F_n$ es escalonada y aumenta en los valores observados.

Su importancia radica en que proporciona una estimación no paramétrica de la función de distribución poblacional.

Además, el teorema de Glivenko-Cantelli establece que

$$
\boxed{
\sup_{x\in\mathbb{R}}|F_n(x)-F(x)|
\xrightarrow{c.s.}0.
}
$$

Es decir, conforme aumenta el tamaño muestral, la función de distribución empírica converge uniformemente a la función de distribución poblacional, casi seguramente.

Este resultado constituye uno de los fundamentos de las pruebas de bondad de ajuste.

## 5. Prueba de Kolmogórov-Smirnov

La prueba de Kolmogórov-Smirnov compara la función de distribución empírica con una función de distribución teórica completamente especificada.

Supongamos inicialmente que deseamos contrastar

$$
H_0:F=F_0
$$

frente a

$$
H_1:F\neq F_0,
$$

donde $F_0$ es una función de distribución continua conocida.

El estadístico de prueba es

$$
\boxed{
D_n=\sup_{x\in\mathbb{R}}|F_n(x)-F_0(x)|.
}
$$

Por lo tanto, $D_n$ representa la mayor distancia vertical entre ambas funciones de distribución.

Si $H_0$ es verdadera, esperamos que $F_n$ se encuentre relativamente cerca de $F_0$.

En cambio, valores grandes de $D_n$ proporcionan evidencia contra la hipótesis nula.

La región de rechazo tiene la forma

$$
D_n>c_{\alpha,n},
$$

donde $c_{\alpha,n}$ es un valor crítico que depende del tamaño muestral y del nivel de significancia.

### 5.1. Aplicación a la gaussianidad

Si conocemos los parámetros $\mu$ y $\sigma$, podemos considerar

$$
F_0(x)=\Phi\left(\frac{x-\mu}{\sigma}\right),
$$

donde $\Phi$ representa la función de distribución de una gaussiana estándar.

Sin embargo, en la mayoría de las aplicaciones desconocemos ambos parámetros y debemos estimarlos a partir de la muestra.

Una posibilidad consiste en utilizar

$$
\overline{X}=\frac{1}{n}\sum_{i=1}^{n}X_i
$$

y

$$
S^2=\frac{1}{n-1}\sum_{i=1}^{n}(X_i-\overline{X})^2.
$$

De esta manera, obtenemos el estadístico

$$
D_n^*=
\sup_x
\left|
F_n(x)-
\Phi\left(\frac{x-\overline{X}}{S}\right)
\right|.
$$

No obstante, aparece una dificultad importante: los parámetros de la distribución teórica se estimaron utilizando las mismas observaciones que construyen $F_n$.

Por lo tanto, la distribución nula del estadístico ya no coincide con la correspondiente a la prueba clásica de Kolmogórov-Smirnov con parámetros conocidos.

En consecuencia, no debemos utilizar directamente los valores críticos clásicos de Kolmogórov-Smirnov para este problema.

## 6. Prueba de Lilliefors

La prueba de Lilliefors adapta el enfoque de Kolmogórov-Smirnov al caso en que los parámetros de la distribución gaussiana son desconocidos y se estiman a partir de la muestra.

Consideramos nuevamente

$$
H_0:X\sim N(\mu,\sigma^2)
$$

frente a

$$
H_1:X\not\sim N(\mu,\sigma^2).
$$

El estadístico se construye comparando la distribución empírica con una distribución gaussiana ajustada a los datos:

$$
\boxed{
D_n^*=
\sup_x
\left|
F_n(x)-
\Phi\left(\frac{x-\overline{X}}{S}\right)
\right|.
}
$$

La diferencia fundamental respecto de la prueba clásica de Kolmogórov-Smirnov se encuentra en la distribución de referencia utilizada para obtener los valores críticos y los p-valores.

La prueba de Lilliefors considera explícitamente que los parámetros se estimaron a partir de las observaciones.

Así, proporciona un procedimiento apropiado para contrastar gaussianidad cuando desconocemos la media y la varianza poblacionales.

Sin embargo, al igual que otras pruebas de bondad de ajuste, su potencia depende del tipo de desviación y del tamaño muestral.

## 7. Prueba de Anderson-Darling

La prueba de Anderson-Darling también compara la distribución empírica con una distribución teórica, pero introduce una característica adicional: asigna mayor importancia a las diferencias que aparecen en las colas de la distribución.

Sea $F_0$ una función de distribución continua completamente especificada.

El estadístico de Anderson-Darling puede expresarse como

$$
\boxed{
A^2=
n\int_{-\infty}^{\infty}
\frac{[F_n(x)-F_0(x)]^2}
{F_0(x)[1-F_0(x)]}\,dF_0(x).
}
$$

Observemos el denominador

$$
F_0(x)[1-F_0(x)].
$$

Cuando $F_0(x)$ se encuentra cerca de 0 o de 1, este producto es pequeño.

Por lo tanto, las discrepancias en las colas reciben un peso relativamente mayor.

Esta característica distingue a Anderson-Darling de Kolmogórov-Smirnov, cuyo estadístico considera la máxima distancia vertical sin introducir ese mismo mecanismo de ponderación.

### 7.1. Forma computacional del estadístico

Sean

$$
X_{(1)}\leq X_{(2)}\leq\cdots\leq X_{(n)}
$$

las observaciones ordenadas.

Para una distribución teórica continua completamente especificada, el estadístico puede calcularse mediante

$$
\boxed{
A^2=
-n-\frac{1}{n}
\sum_{i=1}^{n}(2i-1)
\left[
\ln F_0(X_{(i)})
+
\ln\left(1-F_0(X_{(n+1-i)})\right)
\right].
}
$$

Los valores grandes de $A^2$ indican discrepancias importantes entre la distribución empírica y la distribución teórica.

Cuando contrastamos gaussianidad con parámetros desconocidos, utilizamos una distribución gaussiana ajustada a la muestra.

En este caso, al igual que ocurre con Lilliefors, necesitamos valores críticos o p-valores apropiados para el procedimiento de estimación utilizado.

### 7.2. Interpretación

La prueba de Anderson-Darling resulta especialmente interesante cuando deseamos detectar desviaciones relacionadas con las colas de la distribución.

Por ejemplo, una población puede ser aproximadamente simétrica y presentar una región central semejante a la de una gaussiana, pero producir valores extremos con mayor frecuencia.

En una situación de este tipo, Anderson-Darling puede resultar más sensible que otros procedimientos.

Sin embargo, no existe una garantía de que sea la prueba más potente frente a todas las distribuciones alternativas.

## 8. Prueba de Shapiro-Wilk

La prueba de Shapiro-Wilk es uno de los procedimientos más utilizados para contrastar gaussianidad, particularmente cuando trabajamos con muestras pequeñas.

A diferencia de las pruebas anteriores, su construcción no se basa directamente en la distancia entre la función de distribución empírica y una función teórica.

Su idea central consiste en estudiar la relación entre las observaciones ordenadas y los valores esperados de los estadísticos de orden de una muestra gaussiana.

### 8.1. Estadísticos de orden

Sea

$$
X_1,X_2,\ldots,X_n
$$

una muestra aleatoria.

Denotamos sus observaciones ordenadas mediante

$$
X_{(1)}\leq X_{(2)}\leq\cdots\leq X_{(n)}.
$$

Supongamos ahora que

$$
X_i\sim N(\mu,\sigma^2).
$$

Entonces podemos escribir

$$
X_i=\mu+\sigma Z_i,
$$

donde

$$
Z_i\sim N(0,1).
$$

Como la transformación $z\mapsto\mu+\sigma z$ es creciente, también tenemos

$$
X_{(i)}=\mu+\sigma Z_{(i)}.
$$

Por lo tanto,

$$
E[X_{(i)}]=\mu+\sigma E[Z_{(i)}].
$$

Esto significa que, bajo gaussianidad, las posiciones esperadas de las observaciones ordenadas están relacionadas con los estadísticos de orden de una gaussiana estándar.

Esta propiedad constituye el fundamento de Shapiro-Wilk.

### 8.2. Construcción del estadístico

Sean

$$
m_i=E[Z_{(i)}]
$$

los valores esperados de los estadísticos de orden de una muestra de tamaño $n$ procedente de una gaussiana estándar.

Definimos el vector

$$
\mathbf{m}=
(m_1,\ldots,m_n)^T
$$

y la matriz de covarianzas

$$
V=\operatorname{Cov}
(Z_{(1)},\ldots,Z_{(n)}).
$$

Los coeficientes de Shapiro-Wilk se construyen a partir de $\mathbf{m}$ y $V$.

En particular, podemos definir

$$
\mathbf{a}=
\frac{V^{-1}\mathbf{m}}
{\sqrt{
\mathbf{m}^TV^{-1}V^{-1}\mathbf{m}
}}.
$$

El estadístico de Shapiro-Wilk es

$$
\boxed{
W=
\frac{
\left(\displaystyle\sum_{i=1}^{n}a_iX_{(i)}\right)^2
}{
\displaystyle\sum_{i=1}^{n}(X_i-\overline{X})^2
}.
}
$$

Los coeficientes $a_i$ dependen únicamente del tamaño muestral y de las propiedades de los estadísticos de orden de la gaussiana estándar.

El estadístico compara una combinación lineal de las observaciones ordenadas con la variabilidad total de la muestra.

Bajo gaussianidad, esperamos que las observaciones ordenadas mantengan una relación estrecha con los valores esperados de los estadísticos de orden gaussianos.

Cuando esa relación se deteriora, el valor de $W$ tiende a disminuir.

### 8.3. Propiedades del estadístico

Para una muestra no constante, el estadístico satisface

$$
0<W\leq1.
$$

En términos generales, valores de $W$ cercanos a 1 indican una configuración de las observaciones compatible con la gaussianidad.

En cambio, valores pequeños indican desviaciones respecto del patrón esperado bajo una distribución gaussiana.

Sin embargo, no debemos establecer un umbral arbitrario como $W=0.90$ para decidir si rechazamos o no la gaussianidad.

La decisión debe basarse en la distribución nula del estadístico, que depende del tamaño de la muestra.

Por ello, necesitamos calcular el p-valor correspondiente.

La región de rechazo tiene la forma

$$
W<c_{\alpha,n}.
$$

### 8.4. Invariancia ante cambios de localización y escala

Una propiedad importante de Shapiro-Wilk es que su estadístico no cambia si transformamos todas las observaciones mediante

$$
Y_i=a+bX_i,
\qquad b\neq0.
$$

En efecto, el denominador se multiplica por $b^2$.

Por otra parte, los coeficientes de Shapiro-Wilk suman cero, de modo que el término de localización desaparece del numerador. El cambio de escala multiplica su cuadrado por $b^2$.

Por lo tanto,

$$
\boxed{
W(Y_1,\ldots,Y_n)=W(X_1,\ldots,X_n).
}
$$

Esta propiedad es coherente con el problema que estamos estudiando: la gaussianidad no depende de la unidad de medida ni de la posición de la distribución.

Si una población es gaussiana, cualquier transformación afín no degenerada de esa población también es gaussiana.

### 8.5. Shapiro-Wilk para muestras pequeñas

La prueba de Shapiro-Wilk se desarrolló originalmente para tamaños muestrales desde $n=3$ y suele presentar buena potencia frente a diversas alternativas no gaussianas.

Por esta razón, constituye una elección especialmente útil cuando disponemos de pocas observaciones.

Sin embargo, debemos recordar que ninguna prueba puede garantizar la detección de todas las desviaciones de la gaussianidad con una muestra pequeña.

Por ejemplo, si utilizamos $n=8$ observaciones procedentes de una población ligeramente asimétrica, Shapiro-Wilk podría no rechazar la hipótesis nula.

Esto no representa necesariamente un defecto de la prueba, sino una consecuencia de la información limitada que contiene la muestra.

## 9. Herramientas gráficas para evaluar gaussianidad

Las pruebas formales pueden complementarse con representaciones gráficas que permitan identificar características de la distribución muestral.

Entre las herramientas más utilizadas se encuentran los histogramas y los gráficos de cuantiles teóricos frente a cuantiles observados, conocidos como gráficos Q-Q.

### 9.1. Histogramas

Un histograma representa la distribución de frecuencias de las observaciones.

Cuando los datos proceden de una población gaussiana y el tamaño muestral es suficientemente grande, esperamos observar una forma aproximadamente simétrica y acampanada.

Sin embargo, con muestras pequeñas la forma del histograma puede cambiar considerablemente dependiendo de la amplitud de los intervalos utilizados.

Por ello, un histograma irregular no constituye por sí mismo evidencia suficiente contra la gaussianidad.

### 9.2. Gráficos Q-Q

Un gráfico Q-Q compara los cuantiles observados con los cuantiles teóricos de una distribución de referencia.

Para una muestra ordenada

$$
X_{(1)}\leq\cdots\leq X_{(n)},
$$

podemos considerar las posiciones aproximadas

$$
p_i=\frac{i-\frac12}{n}
$$

y calcular los cuantiles gaussianos estándar

$$
q_i=\Phi^{-1}(p_i).
$$

Después representamos los puntos

$$
(q_i,X_{(i)}).
$$

Si las observaciones proceden de una distribución $N(\mu,\sigma^2)$, esperamos que los puntos se encuentren aproximadamente sobre una recta de la forma

$$
y=\mu+\sigma x.
$$

Desviaciones sistemáticas respecto de esta recta pueden sugerir asimetría, colas pesadas u otras diferencias con la gaussianidad.

Sin embargo, un gráfico Q-Q tampoco constituye una prueba formal de hipótesis.

Su interpretación debe considerar el tamaño de la muestra y la variabilidad esperada de los estadísticos de orden.

## 10. Comparación entre las pruebas

Las pruebas estudiadas comparten el objetivo de detectar desviaciones de la gaussianidad, pero utilizan diferentes propiedades de la distribución muestral.

La prueba de Kolmogórov-Smirnov compara la máxima distancia entre la función de distribución empírica y una distribución teórica completamente especificada.

La prueba de Lilliefors adapta este enfoque cuando los parámetros gaussianos se estiman a partir de la muestra.

La prueba de Anderson-Darling utiliza una distancia ponderada que concede mayor importancia a las discrepancias en las colas.

Finalmente, Shapiro-Wilk estudia la estructura de las observaciones ordenadas y su relación con los estadísticos de orden gaussianos.

No existe una prueba uniformemente más potente frente a todas las distribuciones alternativas.

No obstante, Shapiro-Wilk suele ser una elección apropiada para muestras pequeñas, mientras que Anderson-Darling puede resultar especialmente interesante cuando las desviaciones afectan a las colas.

La elección de la prueba debe considerar el objetivo del análisis, las características de los datos y las alternativas que deseamos detectar.

## 11. Ejemplo de interpretación con una muestra pequeña

Supongamos que un investigador desea determinar si cierta característica poblacional tiene distribución gaussiana.

Para ello obtiene una muestra aleatoria de tamaño

$$
n=8.
$$

Decide utilizar Shapiro-Wilk con un nivel de significancia

$$
\alpha=0.05.
$$

Las hipótesis son

$$
H_0:X\sim N(\mu,\sigma^2)
$$

y

$$
H_1:X\not\sim N(\mu,\sigma^2).
$$

Supongamos que, después de calcular el estadístico y su distribución de referencia, obtiene

$$
W=0.94,
\qquad p=0.62.
$$

Estos valores son hipotéticos y se utilizan únicamente para ilustrar la interpretación.

Como

$$
p=0.62>0.05,
$$

no rechazamos $H_0$.

La conclusión es que la muestra no proporciona evidencia estadísticamente significativa contra la gaussianidad.

Sin embargo, como únicamente disponemos de ocho observaciones, debemos considerar que la prueba podría tener poca potencia frente a determinadas distribuciones alternativas.

En consecuencia, no podemos afirmar que la población sea gaussiana.

Ahora consideremos una segunda situación hipotética en la que obtenemos

$$
p=0.02.
$$

En este caso,

$$
p=0.02<0.05,
$$

por lo que rechazamos la hipótesis nula.

Concluimos que existe evidencia estadísticamente significativa contra la gaussianidad de la población.

La diferencia entre ambos casos muestra que el tamaño pequeño de una muestra no impide rechazar la gaussianidad, aunque sí puede dificultar la detección de desviaciones moderadas.

## 12. Gaussianidad y Teorema del Límite Central

Una confusión frecuente consiste en pensar que las pruebas de gaussianidad dejan de ser necesarias cuando el tamaño muestral es grande debido al Teorema del Límite Central.

Sin embargo, debemos distinguir entre la distribución de las observaciones individuales y la distribución de la media muestral.

Sea $X_1,\ldots,X_n$ una muestra aleatoria independiente e idénticamente distribuida, con

$$
E[X_i]=\mu,
\qquad
\operatorname{Var}(X_i)=\sigma^2<\infty,
$$

donde $\sigma>0$.

El Teorema del Límite Central establece que

$$
\boxed{
\frac{\overline{X}-\mu}{\sigma/\sqrt{n}}
\xrightarrow{d}N(0,1).
}
$$

Este resultado no afirma que las observaciones individuales se vuelvan gaussianas cuando aumenta $n$.

Afirma que la distribución de la media muestral estandarizada converge a una gaussiana estándar.

Por ejemplo, si las observaciones proceden de una distribución exponencial, su distribución poblacional seguirá siendo exponencial independientemente del tamaño de la muestra.

Sin embargo, para tamaños muestrales suficientemente grandes, la distribución de la media muestral estandarizada será aproximadamente gaussiana.

Esta distinción es fundamental para decidir cuándo necesitamos suponer gaussianidad.

### 12.1. Relación con la prueba t de Student

Si las observaciones proceden de una población gaussiana, entonces

$$
T=
\frac{\overline{X}-\mu}{S/\sqrt{n}}
\sim t_{n-1}.
$$

Esta distribución es exacta para cualquier $n\geq2$, siempre que se cumplan las condiciones correspondientes.

En cambio, si la población no es gaussiana, el estadístico anterior no necesariamente tiene distribución t de Student.

Bajo condiciones apropiadas y con varianza poblacional finita, su distribución converge a una gaussiana estándar cuando aumenta el tamaño muestral.

Por ello, cuando $n$ es pequeño, la suposición de gaussianidad puede ser especialmente importante para justificar el uso exacto de determinados procedimientos paramétricos.

Sin embargo, no rechazar una prueba de gaussianidad no demuestra que las condiciones de una prueba t se cumplan.

La elección del procedimiento inferencial también debe considerar la independencia de las observaciones, la presencia de valores extremos y las características del fenómeno estudiado.

## 13. Recomendaciones para muestras pequeñas

Cuando disponemos de una muestra pequeña, conviene comenzar examinando el origen de las observaciones y verificando que el diseño de muestreo sea compatible con los supuestos del análisis.

Posteriormente, podemos utilizar un gráfico Q-Q para identificar desviaciones evidentes de la gaussianidad y complementar esta exploración con una prueba formal.

Shapiro-Wilk constituye una elección razonable en muchas aplicaciones con muestras pequeñas.

Sin embargo, debemos interpretar sus resultados considerando la potencia estadística.

Un p-valor pequeño proporciona evidencia contra la gaussianidad, mientras que un p-valor grande solamente indica que no encontramos evidencia suficiente para rechazarla.

También debemos evitar aplicar numerosas pruebas de gaussianidad y seleccionar posteriormente la conclusión que más nos convenga.

Las distintas pruebas pueden producir resultados diferentes porque responden de manera distinta a las características de la distribución alternativa.

Finalmente, es importante recordar que la gaussianidad no es un requisito universal de la inferencia estadística. Su importancia depende del procedimiento que deseamos utilizar y de las condiciones bajo las cuales sus resultados son válidos.

## 14. Conclusiones

Las pruebas de gaussianidad permiten evaluar si una muestra proporciona evidencia estadística contra la hipótesis de que la población tiene distribución gaussiana.

En este capítulo estudiamos cuatro procedimientos principales: Kolmogórov-Smirnov, Lilliefors, Anderson-Darling y Shapiro-Wilk.

Cada uno utiliza una estrategia diferente para identificar discrepancias entre los datos y una distribución gaussiana.

La prueba de Shapiro-Wilk resulta especialmente útil cuando trabajamos con muestras pequeñas, aunque su capacidad para detectar desviaciones depende de la potencia estadística.

La potencia se define mediante

$$
\boxed{
1-\beta=P(\text{Rechazar }H_0\mid H_1),
}
$$

entendiendo que esta probabilidad debe evaluarse respecto de una distribución alternativa específica.

Cuando el tamaño muestral es pequeño, la potencia puede ser insuficiente para detectar ciertas desviaciones de la gaussianidad.

Por lo tanto, no rechazar la hipótesis nula no equivale a demostrar que la población tiene distribución gaussiana.

Finalmente, distinguimos entre la gaussianidad de la población y la gaussianidad asintótica de la media muestral, establecida por el Teorema del Límite Central.

Esta distinción permite comprender por qué algunos procedimientos inferenciales requieren suposiciones distribucionales especialmente importantes cuando el tamaño de la muestra es pequeño.

En conclusión, las pruebas de gaussianidad deben entenderse como herramientas para evaluar la compatibilidad de los datos con un modelo probabilístico, y no como mecanismos capaces de demostrar definitivamente la forma de una distribución poblacional.


## Resumen de las pruebas de gaussianidad

| Prueba | ¿Cuándo utilizarla? | Aplicación en R |
|---|---|---|
| **Shapiro-Wilk** | Especialmente recomendable para muestras pequeñas. Presenta buena potencia frente a diversas desviaciones de la gaussianidad. | `shapiro.test(x)` |
| **Kolmogórov-Smirnov** | Cuando se desea comparar la muestra con una distribución gaussiana cuyos parámetros $\mu$ y $\sigma$ son conocidos de antemano. | `ks.test(x, "pnorm", mean=mu, sd=sigma)` |
| **Lilliefors** | Cuando los parámetros $\mu$ y $\sigma$ son desconocidos y deben estimarse a partir de la muestra. | `nortest::lillie.test(x)` |
| **Anderson-Darling** | Cuando interesa detectar desviaciones de la gaussianidad, especialmente en las colas de la distribución. | `nortest::ad.test(x)` |

### Consideraciones importantes

- **Shapiro-Wilk:** la función `shapiro.test()` de R admite entre 3 y 5000 observaciones no faltantes. Es una de las pruebas más recomendadas para muestras pequeñas.

- **Kolmogórov-Smirnov:** requiere que los parámetros de la distribución teórica estén completamente especificados. No es correcto estimar $\mu$ y $\sigma$ con la misma muestra y utilizar directamente el p-valor clásico.

- **Lilliefors:** es una adaptación de Kolmogórov-Smirnov que permite contrastar gaussianidad cuando los parámetros poblacionales son desconocidos.

- **Anderson-Darling:** asigna mayor importancia a las discrepancias entre la distribución empírica y la teórica en las colas.

Las funciones `lillie.test()` y `ad.test()` pertenecen al paquete `nortest`, que debe instalarse previamente.

### Interpretación de los resultados

Todas estas pruebas consideran las hipótesis

$$
H_0:X\sim N(\mu,\sigma^2)
$$

$$
H_1:X\not\sim N(\mu,\sigma^2).
$$

Para un nivel de significancia $\alpha=0.05$, la regla de decisión es

$$
\boxed{
\begin{cases}
p<0.05 & \Rightarrow \text{Rechazar }H_0,\\
p\geq0.05 & \Rightarrow \text{No rechazar }H_0.
\end{cases}
}
$$

**Recomendación:** cuando el tamaño muestral es pequeño, Shapiro-Wilk constituye una elección apropiada, complementada con un gráfico Q-Q. Sin embargo, no rechazar la gaussianidad no demuestra que la población sea gaussiana, especialmente cuando la potencia estadística es baja.



### ¿Qué se entiende por muestra pequeña?

En estadística, no existe un tamaño muestral universal que permita clasificar una muestra como pequeña o grande. Esta clasificación depende del procedimiento estadístico, de la distribución poblacional y del objetivo del análisis.

Tradicionalmente, se utiliza la siguiente clasificación orientativa:

| Tamaño muestral | Clasificación |
|---|---|
| $n<30$ | Muestra pequeña |
| $30\leq n<100$ | Muestra moderada |
| $n\geq100$ | Muestra grande |

Sin embargo, el umbral de $n=30$ es una regla práctica y no un resultado matemático general.

#### Muestras pequeñas en las pruebas de gaussianidad

En el contexto de las pruebas de gaussianidad, la principal dificultad de trabajar con muestras pequeñas es la posible baja potencia estadística.

Recordemos que la potencia se define como

$$
1-\beta=P(\text{Rechazar }H_0\mid H_1),
$$

donde la probabilidad se calcula para una distribución alternativa específica.

Por ejemplo, una muestra de tamaño $n=20$ podría ser suficiente para detectar una desviación muy pronunciada de la gaussianidad, pero insuficiente para detectar una desviación moderada.

Incluso con $n=100$, una prueba puede presentar poca potencia frente a determinadas distribuciones alternativas.

Por lo tanto, no podemos afirmar que una prueba de gaussianidad tenga suficiente potencia simplemente porque $n\geq30$.

#### Consideración sobre el Teorema del Límite Central

El umbral de $n=30$ suele asociarse al Teorema del Límite Central. Sin embargo, este teorema no establece que, a partir de 30 observaciones, la distribución de la media muestral sea necesariamente aproximadamente gaussiana.

La rapidez de convergencia depende de las características de la distribución poblacional.

En consecuencia, para nuestro estudio adoptaremos $n<30$ como una convención didáctica para referirnos a muestras pequeñas, sin considerarlo un límite teórico.

**Conclusión:** en las pruebas de gaussianidad, el tamaño muestral debe interpretarse en relación con la potencia estadística. Shapiro-Wilk suele ser una elección apropiada para muestras pequeñas, pero continúa siendo aplicable para tamaños muestrales mayores.


## Ejemplos de aplicación de las pruebas de gaussianidad

En los siguientes ejemplos utilizaremos un nivel de significancia de $\alpha=0.05$.

Las hipótesis generales son

$$
H_0:X\sim N(\mu,\sigma^2)
$$

$$
H_1:X\not\sim N(\mu,\sigma^2).
$$

La regla de decisión es

$$
\boxed{
\begin{cases}
p<\alpha & \Rightarrow \text{Rechazar }H_0,\\
p\geq\alpha & \Rightarrow \text{No rechazar }H_0.
\end{cases}
}
$$

Los ejemplos utilizan datos simulados para que puedan reproducirse. Los resultados se obtendrán al ejecutar el código en R.

### 1. Prueba de Shapiro-Wilk

#### Ejemplo 1. Estaturas de estudiantes

Un investigador registra las estaturas, en centímetros, de 12 estudiantes seleccionados aleatoriamente. Desea determinar si la población de estaturas puede modelarse mediante una distribución gaussiana.

Como el tamaño muestral es pequeño, $n=12$, utilizaremos la prueba de Shapiro-Wilk.

```r
estaturas <- c(165, 170, 168, 172, 169, 174,
               166, 171, 173, 167, 170, 169)

shapiro.test(estaturas)
```

Las hipótesis son

$$
H_0:X\sim N(\mu,\sigma^2)
$$

$$
H_1:X\not\sim N(\mu,\sigma^2).
$$

Si el p-valor es mayor o igual que 0.05, no rechazamos la gaussianidad. Esto significa que las observaciones no proporcionan evidencia suficiente para descartar una distribución gaussiana.

Sin embargo, debido al tamaño pequeño de la muestra, debemos considerar que la prueba podría tener poca potencia frente a ciertas desviaciones.

#### Ejemplo 2. Tiempos de espera

Un investigador registra los tiempos de espera, en minutos, de 15 clientes de un establecimiento.

Desea investigar si los tiempos de espera proceden de una población gaussiana.

```r
tiempos <- c(1.2, 1.5, 1.8, 2.0, 2.1,
             2.3, 2.5, 2.8, 3.0, 3.2,
             4.5, 6.8, 9.2, 14.5, 22.0)

shapiro.test(tiempos)
```

Observemos que los datos presentan una cola hacia la derecha, pues existen algunos tiempos considerablemente mayores que los demás.

La prueba de Shapiro-Wilk permite determinar si esta configuración proporciona evidencia estadísticamente significativa contra la gaussianidad.

Si obtenemos $p<0.05$, rechazamos $H_0$ y concluimos que existe evidencia contra una distribución gaussiana.

### 2. Prueba de Kolmogórov-Smirnov

#### Ejemplo 3. Pesos de paquetes

Una empresa fabrica paquetes cuyo peso debe seguir, de acuerdo con las especificaciones del proceso, una distribución gaussiana con media de 500 gramos y desviación estándar de 10 gramos.

Se seleccionan aleatoriamente 40 paquetes y se desea contrastar si sus pesos siguen exactamente la distribución especificada.

En este caso, los parámetros son conocidos de antemano:

$$
\mu=500,\qquad \sigma=10.
$$

Por lo tanto, las hipótesis son

$$
H_0:X\sim N(500,10^2)
$$

$$
H_1:X\not\sim N(500,10^2).
$$

```r
set.seed(123)

pesos <- rnorm(40, mean = 500, sd = 10)

ks.test(pesos, "pnorm", mean = 500, sd = 10)
```

La prueba compara la función de distribución empírica con la función de distribución de $N(500,10^2)$.

Si $p\geq0.05$, no encontramos evidencia suficiente para rechazar que los pesos siguen la distribución especificada.

Es importante observar que aquí estamos contrastando una distribución completamente determinada, no solamente la pertenencia a la familia gaussiana.

#### Ejemplo 4. Errores de un instrumento

Un fabricante establece que los errores de medición de cierto instrumento siguen una distribución gaussiana con media cero y desviación estándar de 2 unidades.

Un investigador obtiene 50 mediciones y desea verificar esta especificación.

Las hipótesis son

$$
H_0:X\sim N(0,2^2)
$$

$$
H_1:X\not\sim N(0,2^2).
$$

```r
set.seed(456)

errores <- rexp(50, rate = 0.5) - 2

ks.test(errores, "pnorm", mean = 0, sd = 2)
```

Los datos se generan mediante una distribución exponencial desplazada, que es asimétrica y tiene media cero.

La prueba permite evaluar si estas observaciones son compatibles con la distribución gaussiana especificada por el fabricante.

Si $p<0.05$, rechazamos la hipótesis de que los errores siguen $N(0,2^2)$.

En ambos ejemplos de Kolmogórov-Smirnov, los parámetros teóricos se establecieron antes de observar la muestra. No fueron estimados a partir de los mismos datos.

### 3. Prueba de Lilliefors

#### Ejemplo 5. Calificaciones de estudiantes

Un profesor desea investigar si las calificaciones de 25 estudiantes proceden de una población gaussiana.

A diferencia de los ejemplos anteriores, desconoce la media y la varianza poblacionales.

Por lo tanto, utilizaremos la prueba de Lilliefors.

```r
install.packages("nortest")
library(nortest)

set.seed(789)

calificaciones <- rnorm(25, mean = 75, sd = 8)

lillie.test(calificaciones)
```

Las hipótesis son

$$
H_0:X\sim N(\mu,\sigma^2)
$$

$$
H_1:X\not\sim N(\mu,\sigma^2).
$$

La prueba estima los parámetros de la distribución gaussiana a partir de las observaciones y compara la función de distribución empírica con la distribución ajustada.

Si $p\geq0.05$, no rechazamos la gaussianidad.

#### Ejemplo 6. Ingresos de trabajadores

Un investigador obtiene una muestra aleatoria de 60 ingresos mensuales y desea determinar si su distribución es gaussiana.

Como desconoce los parámetros poblacionales, utiliza Lilliefors.

```r
set.seed(321)

ingresos <- rlnorm(60, meanlog = 9, sdlog = 0.8)

lillie.test(ingresos)
```

Los ingresos se generan mediante una distribución lognormal, que presenta asimetría positiva.

La prueba permite investigar si la diferencia entre la distribución empírica y una gaussiana ajustada es estadísticamente significativa.

Si $p<0.05$, rechazamos la gaussianidad.

Este ejemplo también permite observar que una distribución puede ser no gaussiana aunque sus parámetros de localización y dispersión puedan estimarse sin dificultad.

### 4. Prueba de Anderson-Darling

#### Ejemplo 7. Errores de medición con valores extremos

Un investigador estudia los errores de un instrumento de medición.

Aunque la mayoría de los errores son relativamente pequeños, sospecha que ocasionalmente aparecen errores extremos con mayor frecuencia que la esperada bajo una distribución gaussiana.

Para investigar esta posibilidad, utiliza Anderson-Darling, que concede especial importancia a las discrepancias en las colas.

```r
library(nortest)

set.seed(654)

errores <- rt(80, df = 3)

ad.test(errores)
```

Los datos se generan mediante una distribución t de Student con tres grados de libertad.

Esta distribución es simétrica, pero tiene colas más pesadas que una gaussiana.

Las hipótesis son

$$
H_0:X\sim N(\mu,\sigma^2)
$$

$$
H_1:X\not\sim N(\mu,\sigma^2).
$$

Si $p<0.05$, concluimos que existe evidencia estadísticamente significativa contra la gaussianidad.

Este ejemplo muestra que una distribución puede ser simétrica y, aun así, no ser gaussiana.

#### Ejemplo 8. Duración de componentes electrónicos

Un fabricante registra la duración de 100 componentes electrónicos.

Desea investigar si la distribución de las duraciones es gaussiana, prestando especial atención a las observaciones extremas.

```r
set.seed(987)

duraciones <- rweibull(100, shape = 1.5, scale = 1000)

ad.test(duraciones)
```

Las duraciones se generan mediante una distribución de Weibull, frecuentemente utilizada para modelar tiempos de vida y fenómenos de confiabilidad.

La prueba de Anderson-Darling compara la distribución empírica con una gaussiana cuyos parámetros se estiman a partir de la muestra.

Si $p<0.05$, rechazamos la hipótesis de gaussianidad.

En este caso, la distribución poblacional es asimétrica y está restringida a valores positivos, mientras que una distribución gaussiana tiene soporte en toda la recta real.

### 5. Consideraciones finales

Los ejemplos anteriores muestran que las pruebas de gaussianidad pueden aplicarse en contextos diferentes, pero no todas responden exactamente a la misma pregunta.

Shapiro-Wilk, Lilliefors y Anderson-Darling permiten contrastar la pertenencia de una distribución poblacional a la familia gaussiana, aun cuando sus parámetros sean desconocidos.

En cambio, la prueba clásica de Kolmogórov-Smirnov requiere especificar completamente la distribución teórica antes de utilizar sus valores críticos habituales.

Además, debemos recordar que los datos simulados no garantizan un resultado específico en términos del p-valor. Aunque conozcamos la distribución utilizada para generar las observaciones, una prueba puede no detectar su diferencia respecto de la gaussianidad.

Esto ocurre porque la decisión depende de la muestra observada y de la potencia estadística de la prueba.

Por último, no debemos interpretar un p-valor grande como una demostración de gaussianidad, ni un p-valor pequeño como una medida de la magnitud de la desviación.

La significancia estadística y la importancia práctica son conceptos diferentes.
