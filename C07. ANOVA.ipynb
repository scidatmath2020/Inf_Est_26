# Análisis de Varianza (ANOVA)

## Introducción

En los capítulos anteriores estudiamos las pruebas de hipótesis sobre parámetros poblacionales y las pruebas de gaussianidad. Ahora abordaremos un problema más general: determinar si varias poblaciones tienen la misma media.

El análisis de varianza, conocido como ANOVA (*Analysis of Variance*), permite contrastar simultáneamente las medias de varias poblaciones mediante la descomposición de la variabilidad de las observaciones.

Supongamos que una institución educativa utiliza tres métodos de enseñanza y desea determinar si producen el mismo rendimiento académico.

Para ello, selecciona tres grupos independientes de estudiantes, aplica un método distinto a cada grupo y registra sus calificaciones.

Sean $\mu_1,\mu_2,\mu_3$ las medias poblacionales correspondientes.

Nuestro interés consiste en contrastar

$$
H_0:\mu_1=\mu_2=\mu_3
$$

frente a

$$
H_1:\text{al menos dos medias son diferentes}.
$$

Una posibilidad sería realizar pruebas de hipótesis para cada pareja de poblaciones. Sin embargo, esta estrategia introduce un problema: al aumentar el número de comparaciones, también aumenta la probabilidad de cometer al menos un error de tipo I.

Por ejemplo, con tres poblaciones tendríamos tres comparaciones. Si fueran independientes y cada una tuviera un nivel de significancia $\alpha=0.05$, la probabilidad de cometer al menos un error de tipo I sería

$$
1-(1-0.05)^3\approx0.1426.
$$

Aunque las comparaciones por parejas generalmente no son independientes, el ejemplo ilustra el problema.

ANOVA permite realizar un único contraste global.

## Planteamiento matemático

Consideremos $k$ poblaciones y supongamos que de la población $i$ obtenemos una muestra aleatoria de tamaño $n_i$:

$$
X_{i1},X_{i2},\ldots,X_{in_i}.
$$

El número total de observaciones es

$$
N=\sum_{i=1}^k n_i.
$$

Denotamos por

$$
\overline X_i=\frac1{n_i}\sum_{j=1}^{n_i}X_{ij}
$$

la media muestral del grupo $i$, y por

$$
\overline X=\frac1N\sum_{i=1}^k\sum_{j=1}^{n_i}X_{ij}
$$

la media general.

Observemos que

$$
\overline X=\frac1N\sum_{i=1}^k n_i\overline X_i.
$$

Por tanto, la media general es una media ponderada de las medias de los grupos.

El problema consiste en contrastar

$$
H_0:\mu_1=\mu_2=\cdots=\mu_k
$$

contra

$$
H_1:\exists\,i\ne j\text{ tales que }\mu_i\ne\mu_j.
$$

Es importante observar que rechazar $H_0$ no significa que todas las medias sean diferentes. Solamente permite concluir que no todas son iguales.

## El modelo ANOVA de un factor

El modelo clásico de ANOVA de un factor es

$$
X_{ij}=\mu_i+\varepsilon_{ij},
$$

donde los errores son independientes y satisfacen

$$
\varepsilon_{ij}\sim N(0,\sigma^2).
$$

Equivalentemente, podemos escribir

$$
X_{ij}=\mu+\tau_i+\varepsilon_{ij},
$$

donde $\mu$ representa una media de referencia y $\tau_i$ el efecto del grupo $i$. Para identificar estos parámetros se impone una restricción, por ejemplo,

$$
\sum_{i=1}^k n_i\tau_i=0.
$$

Bajo esta parametrización, $\mu$ es la media ponderada de las medias poblacionales.

Las hipótesis fundamentales del modelo son:

1. **Independencia:** las observaciones son independientes entre sí.
2. **Gaussianidad:** las observaciones de cada población tienen distribución gaussiana.
3. **Homogeneidad de varianzas:** todas las poblaciones tienen la misma varianza $\sigma^2$.

La tercera condición es importante porque ANOVA compara dos estimaciones de una misma varianza poblacional.

## La idea fundamental: descomponer la variabilidad

La observación central es que podemos escribir

$$X_{ij}-\overline X=(X_{ij}-\overline X_i)
+
(\overline X_i-\overline X).
$$

Esta identidad separa la desviación de una observación respecto de la media general en dos componentes.

El primero,

$$
X_{ij}-\overline X_i,
$$

representa la variabilidad dentro del grupo.

El segundo,

$$
\overline X_i-\overline X,
$$

representa la separación de la media del grupo respecto de la media general.

Al elevar al cuadrado y sumar sobre todas las observaciones, obtenemos

$$
\boxed{SCT=SCD+SCE}
$$

donde

$$
SCT=\sum_{i=1}^k\sum_{j=1}^{n_i}
(X_{ij}-\overline X)^2
$$

es la suma de cuadrados total,

$$
SCD=\sum_{i=1}^k\sum_{j=1}^{n_i}
(X_{ij}-\overline X_i)^2
$$

es la suma de cuadrados dentro de los grupos, y

$$
SCE=\sum_{i=1}^k n_i
(\overline X_i-\overline X)^2
$$

es la suma de cuadrados entre los grupos.

La igualdad se obtiene porque los términos cruzados desaparecen:

$$
\sum_{j=1}^{n_i}(X_{ij}-\overline X_i)=0.
$$

Esta descomposición constituye el fundamento matemático de ANOVA.

## Los grados de libertad

Cada suma de cuadrados tiene asociados ciertos grados de libertad.

La suma de cuadrados total tiene $N-1$ grados de libertad porque la media general se estima a partir de las $N$ observaciones.

La suma de cuadrados dentro de los grupos tiene

$$
\sum_{i=1}^k(n_i-1)=N-k
$$

grados de libertad.

Por último, la suma de cuadrados entre los grupos tiene $k-1$ grados de libertad.

Por tanto,

$$
\boxed{N-1=(N-k)+(k-1)}.
$$

Esta identidad corresponde a la descomposición de los grados de libertad.

## Construcción del estadístico F

Definimos los cuadrados medios

$$
CME=\frac{SCE}{k-1}
$$

y

$$
CMD=\frac{SCD}{N-k}.
$$

Bajo las hipótesis del modelo, el cuadrado medio dentro de los grupos es un estimador insesgado de $\sigma^2$:

$$
E[CMD]=\sigma^2.
$$

Por otra parte,

$$E[CME]=
\sigma^2+
\frac{\sum_{i=1}^k n_i(\mu_i-\mu_\ast)^2}{k-1},
$$

donde

$$
\mu_\ast=\frac1N\sum_{i=1}^k n_i\mu_i.
$$

En particular, bajo $H_0$,

$$
E[CME]=E[CMD]=\sigma^2.
$$

Sin embargo, cuando las medias poblacionales son diferentes, el cuadrado medio entre los grupos tiende a ser mayor.

Esto sugiere utilizar el cociente

$$
\boxed{F=\frac{CME}{CMD}}.
$$

Bajo $H_0$, las hipótesis gaussianas permiten demostrar que

$$
\frac{SCE}{\sigma^2}\sim\chi^2_{k-1}
$$

y

$$
\frac{SCD}{\sigma^2}\sim\chi^2_{N-k},
$$

siendo ambas variables independientes.

En consecuencia,

$$
\boxed{F\sim F_{k-1,N-k}}.
$$

Esta es la distribución F de Fisher-Snedecor, con $k-1$ y $N-k$ grados de libertad.

## Regla de decisión

El estadístico F es no negativo y los valores grandes proporcionan evidencia contra la hipótesis nula.

Para un nivel de significancia $\alpha$, rechazamos $H_0$ cuando

$$
F_{\mathrm{obs}}>F_{1-\alpha;k-1,N-k},
$$

donde $F_{1-\alpha;k-1,N-k}$ es el cuantil correspondiente de la distribución F.

Equivalentemente, calculamos el p-valor

$$
p=P(F_{k-1,N-k}\ge F_{\mathrm{obs}})
$$

y aplicamos la regla

$$
\boxed{p<\alpha\implies\text{rechazar }H_0}.
$$

Si $p\ge\alpha$, no rechazamos $H_0$. Esto no demuestra que las medias poblacionales sean iguales.

## La tabla ANOVA

La información del contraste se organiza tradicionalmente en la siguiente tabla.

| Fuente | Suma de cuadrados | Grados de libertad | Cuadrado medio | Estadístico F |
|---|---|---|---|---|
| Entre grupos | $SCE$ | $k-1$ | $SCE/(k-1)$ | $CME/CMD$ |
| Dentro de grupos | $SCD$ | $N-k$ | $SCD/(N-k)$ | — |
| Total | $SCT$ | $N-1$ | — | — |

Esta tabla resume la descomposición de la variabilidad y proporciona el estadístico necesario para realizar el contraste.

## Ejemplo resuelto

Supongamos que queremos comparar el rendimiento de tres métodos de enseñanza. Se seleccionan cuatro estudiantes para cada método y se obtienen las siguientes calificaciones.

| Método A | Método B | Método C |
|---:|---:|---:|
| 6 | 8 | 10 |
| 7 | 9 | 11 |
| 8 | 10 | 12 |
| 9 | 11 | 13 |

Las medias muestrales son

$$
\overline X_1=7.5,\qquad
\overline X_2=9.5,\qquad
\overline X_3=11.5.
$$

Como los tamaños muestrales son iguales,

$$
\overline X=9.5.
$$

La suma de cuadrados entre grupos es

$$
SCE=4(7.5-9.5)^2+
4(9.5-9.5)^2+
4(11.5-9.5)^2=32.
$$

Por otra parte, cada grupo tiene una suma de cuadrados interna igual a 5. Por tanto,

$$
SCD=5+5+5=15.
$$

En consecuencia,

$$
SCT=32+15=47.
$$

Los cuadrados medios son

$$
CME=\frac{32}{2}=16,
$$

$$
CMD=\frac{15}{9}=\frac53.
$$

Así,

$$
F_{\mathrm{obs}}=\frac{16}{5/3}=9.6.
$$

Como $k=3$ y $N=12$, bajo la hipótesis nula tenemos

$$
F\sim F_{2,9}.
$$

Para $\alpha=0.05$, el valor crítico es aproximadamente $4.256$. Como

$$
9.6>4.256,
$$

rechazamos la hipótesis nula.

El p-valor es aproximadamente $0.0059$.

**Conclusión:** existe evidencia estadísticamente significativa de que los tres métodos de enseñanza no tienen todos la misma media poblacional.

Sin embargo, ANOVA no nos dice cuáles métodos difieren entre sí.

## Comparaciones posteriores: pruebas post hoc

Cuando rechazamos $H_0$, surge una nueva pregunta: ¿qué parejas de medias poblacionales son diferentes?

Podríamos contrastar

$$
H_0^{(ij)}:\mu_i=\mu_j
$$

para cada pareja $i,j$.

No obstante, debemos controlar el error asociado a realizar múltiples comparaciones.

Uno de los procedimientos más utilizados es la prueba de Tukey, que permite comparar todas las parejas de medias controlando la probabilidad de cometer al menos un error de tipo I dentro de la familia de comparaciones.

En el caso equilibrado, con $n_i=n$, dos medias muestrales se consideran significativamente diferentes mediante Tukey HSD cuando

$$
|\overline X_i-\overline X_j|
>
q_{1-\alpha;k,N-k}\sqrt{\frac{CMD}{n}},
$$

donde $q_{1-\alpha;k,N-k}$ es un cuantil de la distribución del rango studentizado.

Cuando los tamaños muestrales son distintos, puede utilizarse la extensión de Tukey-Kramer.

Es importante distinguir los objetivos: ANOVA contrasta una hipótesis global, mientras que las comparaciones post hoc buscan identificar diferencias específicas.

## Verificación de los supuestos

La validez exacta del contraste F clásico depende de las hipótesis del modelo.

### Independencia

Es fundamental que las observaciones sean independientes. Esta condición depende principalmente del diseño de muestreo o del experimento.

No puede garantizarse únicamente mediante una prueba estadística.

### Gaussianidad

La hipótesis se refiere a los errores del modelo, equivalentes a las desviaciones respecto de las medias de cada población.

En la práctica, pueden examinarse los residuos mediante histogramas, gráficos Q-Q y pruebas de gaussianidad como Shapiro-Wilk.

Debe recordarse lo estudiado en C06: no rechazar gaussianidad no equivale a demostrarla, especialmente cuando el tamaño de muestra es pequeño.

### Homogeneidad de varianzas

El modelo supone

$$
\sigma_1^2=\sigma_2^2=\cdots=\sigma_k^2.
$$

Esta condición puede estudiarse mediante pruebas como Levene o Brown-Forsythe, además de herramientas gráficas.

El ANOVA clásico suele ser razonablemente robusto ante desviaciones moderadas de gaussianidad, especialmente en diseños equilibrados, pero la heterogeneidad de varianzas combinada con tamaños muestrales muy distintos puede afectar el nivel real del contraste.

## ANOVA de Welch

Cuando las poblaciones presentan varianzas diferentes, una alternativa es el ANOVA de Welch.

Este procedimiento también contrasta

$$
H_0:\mu_1=\cdots=\mu_k,
$$

pero no exige que las varianzas poblacionales sean iguales.

La idea consiste en ponderar las medias de los grupos mediante cantidades relacionadas con

$$
w_i=\frac{n_i}{s_i^2},
$$

donde $s_i^2$ es la varianza muestral del grupo $i$.

El procedimiento construye un estadístico F modificado y utiliza una aproximación para sus grados de libertad.

Si se rechaza la hipótesis nula, las comparaciones posteriores pueden realizarse mediante el procedimiento de Games-Howell, diseñado para situaciones con varianzas y tamaños muestrales diferentes.

## Tamaño del efecto

Un resultado estadísticamente significativo no necesariamente representa una diferencia importante en términos prácticos.

Para cuantificar la proporción de variabilidad observada asociada a las diferencias entre grupos, podemos utilizar

$$
\eta^2=\frac{SCE}{SCT}.
$$

Este coeficiente satisface

$$
0\le\eta^2\le1.
$$

En nuestro ejemplo,

$$
\eta^2=\frac{32}{47}\approx0.681.
$$

Esto significa que aproximadamente el 68.1 % de la variabilidad total observada corresponde a la variabilidad entre las medias de los grupos.

No debemos interpretar automáticamente este porcentaje como un efecto causal. Esa interpretación requiere condiciones adicionales sobre el diseño experimental.

Otro estimador habitual es

$$
\omega^2=
\frac{SCE-(k-1)CMD}{SCT+CMD},
$$

que introduce una corrección por los grados de libertad y suele presentar menor sesgo que $\eta^2$ como estimador del tamaño del efecto poblacional.

## Alcances y limitaciones

ANOVA responde una pregunta precisa: si las medias de varias poblaciones pueden considerarse iguales desde el punto de vista de un contraste estadístico.

No determina por sí solo qué grupos difieren, no garantiza que las diferencias tengan importancia práctica y no establece relaciones causales sin un diseño adecuado.

Además, aunque el nombre del procedimiento hace referencia a la varianza, su objetivo principal es contrastar medias poblacionales.

La razón es que las diferencias entre las medias generan variabilidad entre los grupos. Cuando esa variabilidad resulta demasiado grande en relación con la variabilidad interna, encontramos evidencia contra la igualdad de las medias.

## Conclusión

ANOVA constituye una extensión natural de las pruebas de hipótesis estudiadas en C05. Su fundamento consiste en descomponer la variabilidad total en una componente dentro de los grupos y otra entre los grupos.

Bajo la hipótesis de igualdad de medias, ambas componentes, una vez ajustadas por sus grados de libertad, estiman una misma varianza poblacional. El cociente de estas estimaciones produce el estadístico F, cuya distribución permite construir un contraste de hipótesis.

De esta manera, el análisis de varianza reúne varias ideas fundamentales de la inferencia estadística: estimación, distribuciones muestrales, pruebas de hipótesis, errores de decisión, gaussianidad y comparación de poblaciones.
